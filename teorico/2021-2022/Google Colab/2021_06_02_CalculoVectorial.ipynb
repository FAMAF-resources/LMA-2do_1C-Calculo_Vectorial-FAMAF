# Cálculo Vectorial


## Teorema de Stokes

Por el Teorema de Green, sabemos que si tenemos un campo vectorial $F:\mathbb{R}^2 \to \mathbb{R}^2$ que actúa sobre una región $D\subset \mathbb{R}^2$ entonces
$$
\int_D \left( \frac{\partial F_2}{\partial x_1}(x) - \frac{\partial F_1}{\partial x_2}(x)\right) dx_1 dx_2
= \int_{\partial D} F_1(x) dx_1 + F_2(x) dx_2.
$$
Este resultado puede extenderse a regiones bidimensionales en $\mathbb{R}^3$, o sea, superficies. Para ello será de mucha utilidad usar una notación conveniente para la integral de un campo sobre una superficie.





Recuerde que si $\phi:U \subset\mathbb{R}^2 \to \mathbb{R}^3$ es la parametrización de una superficie regular $\mathcal{S}$ y $x=\phi(u)$, o sea $(x_1,x_2,x_3) = \bigl(\phi_1(u_1,u_2),\phi_2(u_1,u_2),\phi_3(u_1,u_2) \bigr)$, el vector normal a $\mathcal{S}$ en $x$ es
\begin{align*}
    N(x)   &= \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u) \\
        &= \det\left(\left[ \begin{array}{cc}
             \frac{\partial \phi_2}{\partial u_1} & \frac{\partial \phi_3}{\partial u_1} \\
             \frac{\partial \phi_2}{\partial u_2} & \frac{\partial \phi_3}{\partial u_2}
             \end{array}\right] \right) e^1
        - \det\left(\left[ \begin{array}{cc}
             \frac{\partial \phi_1}{\partial u_1} & \frac{\partial \phi_3}{\partial u_1} \\
             \frac{\partial \phi_1}{\partial u_2} & \frac{\partial \phi_3}{\partial u_2}
             \end{array}\right] \right) e^2
        + \det\left(\left[ \begin{array}{cc}
             \frac{\partial \phi_1}{\partial u_1} & \frac{\partial \phi_2}{\partial u_1} \\
             \frac{\partial \phi_1}{\partial u_2} & \frac{\partial \phi_2}{\partial u_2}
             \end{array}\right] \right) e^3\\
        &= \det\left(\left[ \begin{array}{cc}
             \frac{\partial \phi_2}{\partial u_1} & \frac{\partial \phi_2}{\partial u_2} \\
             \frac{\partial \phi_3}{\partial u_1} & \frac{\partial \phi_3}{\partial u_2}
             \end{array}\right] \right) e^1
        + \det\left(\left[ \begin{array}{cc}
             \frac{\partial \phi_3}{\partial u_1} & \frac{\partial \phi_3}{\partial u_2} \\
             \frac{\partial \phi_1}{\partial u_1} & \frac{\partial \phi_1}{\partial u_2}
             \end{array}\right] \right) e^2
        + \det\left(\left[ \begin{array}{cc}
             \frac{\partial \phi_1}{\partial u_1} & \frac{\partial \phi_1}{\partial u_2} \\
             \frac{\partial \phi_2}{\partial u_1} & \frac{\partial \phi_2}{\partial u_2}
             \end{array}\right] \right) e^3.
\end{align*}
Note que el escalar que multiplica a $e^1$ es igual al determinante de la matriz Jacobiana de $T_1(u) = \bigl( \phi_2(u), \phi_3(u) \bigr)$ y que esta transformación corresponde a la variables $(x_2,x_3)$ del punto en la superficie. Análogamente, tenemos que en la dirección $e^2$ y $e^3$ el cambio de variables es $(x_3,x_1)=\bigl( \phi_3(u), \phi_1(u) \bigr)$ y $(x_1,x_2)=\bigl( \phi_1(u), \phi_2(u) \bigr)$, respectivamente. Por lo tanto una notación coherente sería
\begin{align*}
    \int_{\mathcal{S}} F(x) dS &= \int_{\mathcal{S}} \langle F(x), N(x) \rangle dS \\
    &= \int_{\mathcal{S}} F_1(x) dx_2 dx_3 + F_2(x) dx_3 dx_1 + F_3(x) dx_1 dx_2.
\end{align*}



Tomando esto en cuenta, obtenemos lo siguiente.

>**Teorema** (Stokes)
>
> Sea $\mathcal{S}$ una superficie regular orientada positivamente parametrizada por una función dos veces continuamente diferenciable $\phi: U \subset \mathbb{R}^2 \to \mathbb{R}^3$. Suponga que $U$ es una unión finita de regiones simples, acotado por una curva suave a trozos. Si $F:\mathbb{R}^3 \to \mathbb{R}^3$ es continuamente diferenciable en $\mathcal{S}$, entonces
$$
 \int_{\mathcal{S}} \operatorname{rot}(F)(x) dS = \int_{\partial \mathcal{S}} F(x) ds,
$$
donde $\partial \mathcal{S}$ es la frontera de $\mathcal{S}$ con orientación positiva. Tomando $G=\operatorname{rot}(F)$, esta ecuación también puede escribirse como
 $$
 \int_{\mathcal{S}} G_1(x) dx_2 dx_3 + G_2(x) dx_3 dx_1 + G_3(x) dx_1 dx_2
 = \int_{\partial \mathcal{S}} F_1(x) dx_1 + F_2(x) dx_2 + F_3(x) dx_3.
 $$


**Demostración:**

Veremos que
$$
\int_{\partial \mathcal{S}} F_1(x) dx_1 = \int_{\mathcal{S}} -\frac{\partial F_1}{\partial x_2}(x) dx_1 dx_2 + \int_{\mathcal{S}} \frac{\partial F_1}{\partial x_3} dx_3 dx_1.
$$
De manera análoga se demuestra que
\begin{align*}
\int_{\partial \mathcal{S}} F_2(x) dx_2  &= \int_{\mathcal{S}} -\frac{\partial F_2}{\partial x_3}(x) dx_2 dx_3 + \int_{\mathcal{S}} \frac{\partial F_2}{\partial x_1} dx_1 dx_2, \\
\int_{\partial \mathcal{S}} F_3(x) dx_3  &= \int_{\mathcal{S}} -\frac{\partial F_3}{\partial x_1}(x) dx_3 dx_1 + \int_{\mathcal{S}} \frac{\partial F_3}{\partial x_2} dx_2 dx_3.
\end{align*}
Luego, sumando estas ecuaciones se obtiene
\begin{align*}
    \int_{\partial \mathcal{S}} F_1(x) dx_1 + F_2(x) dx_2 + F_3(x) dx_3
    =& \int_{\mathcal{S}} \left( \frac{\partial F_3}{\partial x_2} -\frac{\partial F_2}{\partial x_3}(x) \right) dx_2 dx_3 \\
     &+ \int_{\mathcal{S}} \left( \frac{\partial F_1}{\partial x_3} -\frac{\partial F_3}{\partial x_1}(x) \right) dx_3 dx_1 \\
     &+ \int_{\mathcal{S}} \left( \frac{\partial F_2}{\partial x_1} -\frac{\partial F_1}{\partial x_2}(x) \right)dx_1 dx_2.
\end{align*}

Para demostrar la primera ecuación, suponga que $\gamma:[a,b] \to \mathbb{R}^2$ es una parametrización de la curva $\partial U \subset \mathbb{R}^2$ recorrida en sentido antihorario. Luego $\beta = \phi \circ \gamma$ es una parametrización de la curva $\partial \mathcal{S} \subset \mathbb{R}^3$ orientada positivamente.
Entonces
\begin{align*}
    \int_{\partial \mathcal{S}} F_1(x) dx_1 &= \int_a^b F_1(\beta(t)) \beta_1'(t) dt \\
    &=\int_a^b F_1(\phi(\gamma(t))) \left(\frac{\partial \phi_1}{\partial u_1}(\gamma(t))\gamma_1'(t) + \frac{\partial \phi_1}{\partial u_2}(\gamma(t))\gamma_2'(t) \right)  dt \\
    &=\int_{\partial U} F_1(\phi(u))\frac{\partial \phi_1}{\partial u_1}(u) du_1
    + F_1(\phi(u))\frac{\partial \phi_1}{\partial u_2}(u) du_2 \\
    &=\int_{\partial U} H_1(u) du_1 + H_2(u) du_2 \\
    &= \int_U \left( \frac{\partial H_2}{\partial u_1}(u) - \frac{\partial H_1}{\partial u_2}(u)\right) du_1 du_2,
\end{align*}
donde en la última ecuación usamos el Teorema de Green para $H_1(u) = F_1(\phi(u))\frac{\partial \phi_1}{\partial u_1}(u)$ y $H_2(u) = F_1(\phi(u))\frac{\partial \phi_1}{\partial u_2}(u)$. Ahora, note que
\begin{align*}
    \frac{\partial H_2}{\partial u_1} &= \left(\frac{\partial F_1}{\partial x_1}\frac{\partial \phi_1}{\partial u_1} + \frac{\partial F_1}{\partial x_2}\frac{\partial \phi_2}{\partial u_1} + \frac{\partial F_1}{\partial x_3}\frac{\partial \phi_3}{\partial u_1} \right)\frac{\partial \phi_1}{\partial u_2} + F_1 \frac{\partial^2 \phi_1}{\partial u_1 \partial u_2} \\
    \frac{\partial H_1}{\partial u_2} &= \left(\frac{\partial F_1}{\partial x_1}\frac{\partial \phi_1}{\partial u_2} + \frac{\partial F_1}{\partial x_2}\frac{\partial \phi_2}{\partial u_2} + \frac{\partial F_1}{\partial x_3}\frac{\partial \phi_3}{\partial u_2} \right)\frac{\partial \phi_1}{\partial u_1} + F_1 \frac{\partial^2 \phi_1}{\partial u_2 \partial u_1}.
\end{align*}
Por lo tanto
\begin{align*}
\frac{\partial H_2}{\partial u_1} - \frac{\partial H_1}{\partial u_2}
    &= \frac{\partial F_1}{\partial x_2} \left( \frac{\partial \phi_2}{\partial u_1} \frac{\partial \phi_1}{\partial u_2} - \frac{\partial \phi_2}{\partial u_2}\frac{\partial \phi_1}{\partial u_1}
    \right)
    + \frac{\partial F_1}{\partial x_3} \left( \frac{\partial \phi_3}{\partial u_1} \frac{\partial \phi_1}{\partial u_2} - \frac{\partial \phi_3}{\partial u_2}\frac{\partial \phi_1}{\partial u_1}
    \right)\\
    &= -\frac{\partial F_1}{\partial x_2} \det\left(\left[ \begin{array}{cc}
             \frac{\partial \phi_1}{\partial u_1} & \frac{\partial \phi_1}{\partial u_2} \\
             \frac{\partial \phi_2}{\partial u_1} & \frac{\partial \phi_2}{\partial u_2}
             \end{array}\right] \right)
     + \frac{\partial F_1}{\partial x_3} \det\left(\left[ \begin{array}{cc}
             \frac{\partial \phi_3}{\partial u_1} & \frac{\partial \phi_3}{\partial u_2} \\
             \frac{\partial \phi_1}{\partial u_1} & \frac{\partial \phi_1}{\partial u_2}
             \end{array}\right] \right).
\end{align*}
Obteniendo que
$$
\int_{\partial \mathcal{S}} F_1(x) dx_1 = \int_{\mathcal{S}} -\frac{\partial F_1}{\partial x_2}(x) dx_1 dx_2 + \int_{\mathcal{S}} \frac{\partial F_1}{\partial x_3} dx_3 dx_1. \quad \spadesuit
$$






Este resultado nos permitirá simplificar los cálculos al resolver integrales de superficie.

#### Ejemplo

Suponga que $\mathcal{S}$ es la esfera truncada con orientación positiva mostrada en la figura.


```
#@title Esfera truncada { display-mode: "form" }
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

subplot_args = {'projection':'3d'}
fig, ax = plt.subplots(1,1,subplot_kw=subplot_args)

th = np.linspace(0, 2*np.pi, 50)
ph = np.linspace(0, 5*np.pi/6, 50)
r = 2
a = r*np.sqrt(3)/2
x = np.outer(r*np.sin(ph), np.cos(th))
y = np.outer(r*np.sin(ph), np.sin(th))
z = np.outer(r*np.cos(ph), np.ones(th.shape))

# phi
ax.plot_surface(x, y, z + a, color='b', alpha=0.3)
ax.plot3D([0,0],[-2.1,2.1], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([-2.1,2.1], [0,0], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([0,0], [0,0], [-0.1,4.1], color=(0.5,0.5,0.5))
ax.plot3D(x[-1,:], y[-1,:], z[-1,:]+a, color='r')
ax.view_init(30,30)
ax.text(2,1,0,r'$x_1^2 + x_2^2 = 1$',fontsize=14)
ax.axis('off');
#plt.savefig('ejm_esferatruncada.pdf')


```

Para el campo $F:\mathbb{R}^3 \to \mathbb{R}^3$ tal que $F(x) = (x_2, -x_1, e^{x_1 x_3})$ se desea calcular
$$
    \int_{\mathcal{S}} \operatorname{rot}(F)(x) dS.
$$
Si parametrizamos $\partial \mathcal{S}$ por $\gamma:[0, 2\pi] \to \mathbb{R}^3$ tal que $\gamma(t) = \bigl( \cos(t), \operatorname{sen}(t), 0 \bigr)$, obtenemos que
\begin{align*}
    \int_{\mathcal{S}} \operatorname{rot}(F)(x) dS &= \int_{\partial \mathcal{S}} F(x) ds \\
    &= \int_0^{2\pi} \langle \bigl( \operatorname{sen}(t), -\cos(t),  1 \bigr) , \bigl(  -\operatorname{sen}(t), \cos(t), 0 \bigr) \rangle dt \\
    &= - \int_0^{2\pi} ( \operatorname{sen}^2(t) + \cos^2(t) ) dt = -2\pi.
\end{align*}

---


Usando el Teorema de Stokes podemos dar una interpretación del campo $\operatorname{rot}(F)$. Primero notemos que si $\mathcal{S}$ es una superficie tal que $\operatorname{cl}(\mathcal{S})$ es compacto, entonces dada $f:\mathbb{R}^3 \to \mathbb{R}$ continua en $\operatorname{cl}(\mathcal{S})$ existe $\tilde{x} \in \operatorname{cl}(\mathcal{S})$ tal que
$$
\int_{\mathcal{S}} f(x) dS = f(\tilde{x}) \mathscr{a}(\mathcal{S}).
$$
Esto se desprende del hecho que existen $x_{\min}, x_{\max} \in \operatorname{cl}(\mathcal{S})$ tales que $f(x_{\min}) \leq f(x) \leq f(x_{\max})$ para todo $x \in \operatorname{cl}(\mathcal{S})$ y por lo tanto $f(x_{\min}) \mathscr{a}(\mathcal{S}) \leq \int_{\mathcal{S}} f(x) dS \leq f(x_{\max}) \mathscr{a}(\mathcal{S})$.


Suponga ahora que $F:\mathbb{R}^3 \to \mathbb{R}^3$ es el campo vectorial de velocidad de un fluído. Para un punto $\bar{x}$ y un vector unitario $N$, tomamos $\mathcal{S}_r(\bar{x})$ como el disco de centro $\bar{x}$ y radio $r$ perpendicular a $N$. Por el Teorema de Stokes tenemos que
$$
\int_{\mathcal{S}_r(\bar{x})} \langle \operatorname{rot}(F)(x) , N \rangle dS = \int_{\mathcal{S}_r(\bar{x})} \operatorname{rot}(F)(x) dS = \int_{\partial \mathcal{S}_r(\bar{x})} F(x) ds,
$$
donde $\partial \mathcal{S}_r(\bar{x})$ tiene orientación positiva en relación a $N$. Además, existe $\tilde{x}_r \in \mathcal{S}_r(\bar{x})$ tal que
$$
\langle \operatorname{rot}(F)(\tilde{x}_r) , N \rangle = \frac{1}{\mathscr{a}(\mathcal{S}_r(\bar{x}))} \int_{\mathcal{S}_r(\bar{x})} \langle \operatorname{rot}(F)(x) , N \rangle dS
= \frac{1}{\mathscr{a}(\mathcal{S}_r(\bar{x}))} \int_{\partial \mathcal{S}_r(\bar{x})} F(x) ds.
$$
Como $\tilde{x}_r \to \bar{x}$ cuando $r \to 0$, obtenemos que
$$
\langle \operatorname{rot}(F)(\bar{x}) , N \rangle = \lim_{r \to 0} \frac{1}{\mathscr{a}(\mathcal{S}_r(\bar{x}))} \int_{\partial \mathcal{S}_r(\bar{x})} F(x) ds.
$$

En general, si $\mathcal{C}$ es una curva cerrada orientada y $F$ apunta siempre en dirección tangente a $\mathcal{C}$ entonces $\int_{\mathcal{C}} F(x) ds >0$ y las partículas en $\mathcal{C}$ giran siguiendo la orientación de la curva. Si $F$ apunta en dirección opuesta se tiene $\int_{\mathcal{C}} F(x) ds <0$. Si $F$ es perpendicular a $\mathcal{C}$ entonces las partículas sobre $\mathcal{C}$ no giran y $\int_{\mathcal{C}} F(x) ds =0$. Luego $\int_{\mathcal{C}} F(x) ds$ representa la cantidad neta de giro del fluído y se la conoce como la circulación de $F$ alrededor de $\mathcal{C}$.

Con esto, $\langle \operatorname{rot}(F)(\bar{x}) , N \rangle$ es el límite de la circulación de $F$ alrededor de $\partial \mathcal{S}_r(\bar{x})$ por unidad de área en una superficie perpendicular a $N$. Concluyendo que $\langle \operatorname{rot}(F)(\bar{x}) , N \rangle$ representa el efecto de giro o rotación del fluído alrededor del eje dado por $N$.
Además, el valor máximo que puede tomar $\langle \operatorname{rot}(F)(\bar{x}) , N \rangle$ se alcanza cuando $N= \operatorname{rot}(F)(\bar{x}) / \|\operatorname{rot}(F)(\bar{x})\|$. Por ello $\operatorname{rot}(F)(x)$ es llamado el vector de vorticidad.


```
#@title Rotacional { display-mode: "form" }
import numpy as np
import matplotlib.pyplot as plt
n = 200
h = 1/(n-1)
t = np.linspace(0, 1, n)
gamma1 = (2+20*t*(t-0.5)*(t-1))*np.cos(2*np.pi*t)
gamma2 = (2-20*t*(t-0.5)*(t-1))*np.sin(2*np.pi*t)
dgamma1 = np.diff(gamma1)
dgamma2 = np.diff(gamma2)
X1, X2 = np.meshgrid(np.linspace(-4,4,100),np.linspace(-4,4,100))
fig, axs = plt.subplots(1,2,figsize=(10,4))
axs[0].plot(gamma1, gamma2, linewidth=2)
X30 = (X1+4)**2 + (X2-4)**2
axs[0].arrow(gamma1[49], gamma2[49], dgamma1[49], dgamma2[49], head_length=.5, head_width=.5, length_includes_head=True, color='b')
axs[0].arrow(gamma1[120], gamma2[120], dgamma1[120], dgamma2[120], head_length=.5, head_width=.5, length_includes_head=True, color='b')
axs[0].arrow(gamma1[190], gamma2[190], dgamma1[190], dgamma2[190], head_length=.5, head_width=.5, length_includes_head=True, color='b')
axs[0].contour(X1, X2, X30, levels=[i**2 for i in range(3,9)], colors='k')
for i in range(3,9):
  axs[0].arrow(i*np.cos(-i*np.pi/20)-4, i*np.sin(-i*np.pi/20)+4, -0.1*np.sin(-i*np.pi/20), 0.1*np.cos(-i*np.pi/20), head_length=.5, head_width=.5, length_includes_head=True, color='k')
axs[0].axis('equal')
axs[0].axis('off')
axs[1].plot((2+20*t*(t-0.5)*(t-1))*np.cos(2*np.pi*t), (2-20*t*(t-0.5)*(t-1))*np.sin(2*np.pi*t))
axs[1].arrow(gamma1[49], gamma2[49], dgamma1[49], dgamma2[49], head_length=.5, head_width=.5, length_includes_head=True, color='b')
axs[1].arrow(gamma1[120], gamma2[120], dgamma1[120], dgamma2[120], head_length=.5, head_width=.5, length_includes_head=True, color='b')
axs[1].arrow(gamma1[190], gamma2[190], dgamma1[190], dgamma2[190], head_length=.5, head_width=.5, length_includes_head=True, color='b')
X31 = X1**2 + X2**2
axs[1].contour(X1, X2, X31, levels=[i**2 for i in range(1,5)] , colors='k')
for i in range(1,5):
  axs[1].arrow(i*np.cos(i*np.pi/6), i*np.sin(i*np.pi/6), -0.1*np.sin(i*np.pi/6), 0.1*np.cos(i*np.pi/6), head_length=.5, head_width=.5, length_includes_head=True, color='k')
axs[1].axis('equal')
axs[1].axis('off');
#plt.savefig('curvarotacional.pdf')
```


Claramente, si $f:\mathbb{R}^3 \to \mathbb{R}$ es dos veces continuamente diferenciable, entonces
$$
\operatorname{rot}(\nabla f)
    = \left(
    \frac{\partial f}{\partial x_2 \partial x_3} - \frac{\partial f}{\partial x_3 \partial x_2}, \,
    \frac{\partial f}{\partial x_3 \partial x_1} - \frac{\partial f}{\partial x_1 \partial x_3},\, \frac{\partial f}{\partial x_1 \partial x_2} - \frac{\partial f}{\partial x_2 \partial x_1} \right)=0.
$$

Por otro lado $F:\mathbb{R}^3 \to \mathbb{R}^3$ es continuamente diferenciable y $\operatorname{rot}(F)=0$, entonces para cualquier curva cerrada $\mathcal{C} \subset \mathbb{R}^3$ se tiene que
$$
\int_{\mathcal{C}} F(x) ds = \int_{\mathcal{S}} \operatorname{rot}(F)(x) dS = 0,
$$
donde $\mathcal{S}$ una superficie cuya frontera es $\mathcal{C}$. Veremos que esta propiedad es suficiente para garantizar que $F = \nabla f$ para alguna función $f:\mathbb{R}^3 \to \mathbb{R}$.




## Campo gradiente

En esta sección veremos condiciones necesarias y suficientes para garantizar que un campo vectorial $F$ puede escribirse como $\nabla f$ para una función a valores reales $f$.

>**Definición**
>
> Diremos que un campo $F:\mathbb{R}^n \to \mathbb{R}^n$ es un **campo gradiente** (o **campo conservativo**) si existe $f:\mathbb{R}^n \to \mathbb{R}$ tal que $F = \nabla f$. En este caso $f$ es llamado el **potencial** de $F$.





Los campos gradiente son caracterizados por el siguiente resultado.

> **Teorema**
>
> Sea $F:\mathbb{R}^n \to \mathbb{R}^n$ continua y definida sobre un conjunto $D$ abierto y poligonalmente conexo. Entonces las siguientes afirmaciones son equivalentes.
1. $F$ es un campo gradiente para una función $f$ continuamente diferenciable en $D$.
2. Para cualquier curva cerrada diferenciable a trozos $\mathcal{C} \subset D$, vale que $\int_{\mathcal{C}} F(x) ds =0$.
3. Dados $\check{x}, \hat{x} \in D$, la integral de línea de $F$ desde $\check{x}$ a $\hat{x}$ es independiente de la curva diferenciable a trozos en $D$ que une esos puntos.


**Demostración:**


(1 $\Rightarrow$ 2). Si $\mathcal{C}$ es una curva cerrada se puede parametrizar por $\gamma:[0,1] \to \mathbb{R}^n$ tal que $\gamma(0)=\gamma(1)$. Como $F=\nabla f$, entonces
$$
 \int_{\mathcal{C}} F(x) ds = \int_{\mathcal{C}} \nabla f(x) ds = \int_0^1 \langle \nabla f(\gamma(t)), \gamma'(t) \rangle dt  = f(\gamma(1)) - f(\gamma(0))=0.
$$

(2 $\Rightarrow$ 3). Sean $\mathcal{C}_1$ y $\mathcal{C}_2$ curvas de $\check{x}$ a $\hat{x}$, entonces la curva $\mathcal{C}$ definida al recorrer $\mathcal{C}_1$ y luego $\mathcal{C}_2$ en sentido contrario, es cerrada y cumple
$$
0 = \int_{\mathcal{C}} F(x) ds = \int_{\mathcal{C}_1} F(x) ds + \int_{\mathcal{C}_2^-} F(x) ds = \int_{\mathcal{C}_1} F(x) ds - \int_{\mathcal{C}_2} F(x) ds.
$$

(3 $\Rightarrow$ 1). Como la integral no depende de la curva, definimos
$$
f(\hat{x})= \int_{\mathcal{C}[\check{x},\hat{x}]} F(x) ds,
$$
donde $\mathcal{C}[\check{x},\hat{x}]$ es una curva diferenciable a trozos arbitraria de $\check{x}$ a $\hat{x}$.
Sea $u\in \mathbb{R}^n$ tal que $\|u\|=1$. Como $\hat{x} \in D$ con $D$ abierto, entonces existe $t>0$ suficientemente pequeño tal que $\hat{x} + tu \in D$. Luego, si $\mathcal{R}$ es el segmento parametrizado por $\gamma:[0,t]\to \mathbb{R}^n$ tal que $\gamma(s) = \hat{x} + su$, entonces $\mathcal{C}[\check{x},\hat{x}] \cup \mathcal{R} = \mathcal{C}[\check{x},\hat{x}+tu]$.
Con esto, obtenemos que
$$
    f(\hat{x}+ t u) - f(\hat{x}) = \int_{\mathcal{C}[\check{x},\hat{x}+tu]} F(x) ds - \int_{\mathcal{C}[\check{x},\hat{x}]} F(x) ds
    = \int_{\mathcal{R}} F(x) ds
    = \int_0^t \langle F(\hat{x} + s u), u \rangle ds.  
$$
Definiendo $\eta(t) = \int_0^t \langle F(\hat{x} + s u), u \rangle ds$, tenemos que $\eta'(t) = \langle F(\hat{x} + t u), u \rangle $ y
$$
    f'(\hat{x}; u) = \lim_{t \to 0} \frac{f(\hat{x}+ t u) - f(\hat{x})}{t}
    = \lim_{t \to 0} \frac{\eta(t) - \eta(0)}{t} = \eta'(0)
     = \langle F(\hat{x}), u \rangle.
$$
Como esto vale para todo $u$, tomando $u=e^j$, tenemos que $\frac{\partial f}{\partial x_j}(\hat{x}) = F_j(\hat{x})$. Concluyendo que $F(\hat{x}) = \nabla f(\hat{x})$ para cualquier $\hat{x} \in D$. $\spadesuit$




Una consecuencia directa de estas equivalencias es que todo campo $F:\mathbb{R}^3 \to \mathbb{R}^3$ con $\operatorname{rot}(F)=0$ es un campo gradiente.




Aunque los campos gradientes son caracterizados por el resultado anterior, es de utilidad contar con formas prácticas de verificar cuándo un campo es gradiente. La siguiente es una condición necesaria.

>**Proposición**
>
> Si $F:\mathbb{R}^n \to \mathbb{R}^n$ es un campo gradiente continuamente diferenciable, entonces $F'(x)$ es una matriz simétrica para todo $x\in\mathbb{R}^n$.

**Demostración:**

Como $F_j(x) = \frac{\partial f}{\partial x_j}(x)$ para todo $j=1, \ldots,n$, entonces para todo $i,j$
$$
    \frac{\partial F_j}{\partial x_i} (x) = \frac{\partial^2 f}{\partial x_i \partial x_j}(x)
    = \frac{\partial^2 f}{\partial x_j \partial x_i}(x) =  \frac{\partial F_i}{\partial x_j} (x). \quad \spadesuit
$$




De esta manera, si la matriz Jacobiana de un campo no es simétrica, entonces tal campo no puede ser un campo gradiente. Veamos que la condición de tener la matriz Jacobiana del campo es simétrica en un conjunto convexo, es una condición suficiente para garantizar que un campo es gradiente. Recuerde que un conjunto es convexo si dados dos puntos en el conjunto el segmento que los une también estará en el conjunto.

> **Proposición**
>
> Sea $D\subset \mathbb{R}^n$ abierto convexo y $F:\mathbb{R}^n \to \mathbb{R}^n$ continuamente diferenciable en $D$. Si $F'(x)$ es simétrica para todo $x\in R$, entonces $F$ es un campo gradiente.

**Demostración:**

Fije $\check{x} \in D$ y para $x\in D$ defina $\mathcal{C}[\check{x},x]$ como el segmento que une $\check{x}$ con $x$ y $f:\mathbb{R}^n \to \mathbb{R}$ tal que
\begin{align*}
    f(x) &= \int_{\mathcal{C}[\check{x},x]} F(y) ds  
    =  \int_0^1 \langle F(\check{x} + t(x-\check{x})) , x-\check{x} \rangle dt \\
    &=  \int_0^1 \sum_{j=1}^n F_j(\check{x} + t(x-\check{x})) (x_j-\check{x}_j) dt.
\end{align*}
Usando la regla de Leibnitz, tenemos que
\begin{align*}
    \frac{\partial f}{\partial x_k}(x)
    &=\int_0^1 \left( \sum_{j=1}^n t \frac{\partial F_j}{\partial x_k}(\check{x} + t(x-\check{x})) (x_j-\check{x}_j) + F_k(\check{x} + t(x-\check{x})) \right) dt \\
    &=\int_0^1 \left( F_k(\check{x} + t(x-\check{x})) + \sum_{j=1}^n t \frac{\partial F_k}{\partial x_j}(\check{x} + t(x-\check{x})) (x_j-\check{x}_j)  \right) dt,
\end{align*}
donde en la última ecuación usamos que $\frac{\partial F_j}{\partial x_k}(x) = \frac{\partial F_k}{\partial x_j}(x)$, por ser $F'(x)$ una matriz simétrica.

Por otro lado, para la función auxiliar $\varphi:\mathbb{R} \to \mathbb{R}$ tal que $\varphi(t) = t F_k(\check{x} + t(x-\check{x}))$, se tiene
$$
\varphi'(t) = F_k(\check{x} + t(x-\check{x})) + t \sum_{j=1}^n \frac{\partial F_k}{\partial x_j} (\check{x} + t(x-\check{x})) (x_j-\check{x}_j).
$$

Juntando las dos expresiones se obtiene
$$
\frac{\partial f}{\partial x_k}(x) = \int_0^1 \varphi'(t) dt = \varphi(1) - \varphi(0) = F_k(x). \quad \spadesuit
$$
