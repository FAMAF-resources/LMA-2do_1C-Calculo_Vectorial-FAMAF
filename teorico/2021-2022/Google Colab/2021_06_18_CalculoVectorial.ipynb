# Optimización y ecuaciones no lineales


## Polinomio de Taylor

Como hemos visto en reiterados ejemplos, es de mucha utilidad aproximar una función continuamente diferenciable $f:\mathbb{R}^n \to \mathbb{R}^m$ por su aproximación afín $x \mapsto f(\bar{x}) + f'(\bar{x})(x-\bar{x})$. No obstante, si el fenómeno a estudiar involucra variaciones de segundo orden, tales como aceleraciones o curvaturas, necesitaremos una aproximación con más información.



Cuando $f:\mathbb{R} \to \mathbb{R}$ es $k$ veces diferenciable en $\bar{x}$ entonces un polinomio de Taylor de orden $k$ alrededor de $\bar{x}$ nos provee una buena aproximación de $f$. El procedimiento para armar el polinomio de Taylor en una variable nos servirá para construirlo en varias variables.

Tomando $\varphi:\mathbb{R} \to \mathbb{R}$ tal que $\varphi(t) = f(\bar{x} + th)$, se tiene que $\varphi^{(j)}(t) = f^{(j)}(\bar{x} + th) h^j$. Como $f$ es $k$ veces diferenciable en $\bar{x}$, dado $\varepsilon>0$ y $h$ suficientemente pequeño
$$
|f^{(k-1)}(\bar{x} + h) - f^{(k-1)}(\bar{x}) - f^{(k)}(\bar{x}) h | \leq \varepsilon |h|.
$$
Por lo tanto para $t \in [0,1]$,
\begin{align*}
|\varphi^{(k-1)}(t) - \varphi^{(k-1)}(0) - \varphi^{(k)}(0)t|
&=|f^{(k-1)}(\bar{x} + th)h^{k-1} - f^{(k-1)}(\bar{x})h^{k-1} - f^{(k)}(\bar{x}) h^{k}t |\\
&=|f^{(k-1)}(\bar{x} + th) - f^{(k-1)}(\bar{x}) - f^{(k)}(\bar{x}) th | |h|^{k-1} \\
&\leq \varepsilon t |h|^k.
\end{align*}

Defina
$$
R_k(h) = \int_0^1 \int_0^{t_1} \ldots \int_0^{t_{k-2}} \bigl( \varphi^{(k-1)}(t_{k-1}) - \varphi^{(k-1)}(0) - \varphi^{(k)}(0) t_{k-1} \bigr) dt_{k-1} \ldots dt_2 dt_1.
$$

Como
$$
\int_0^1 \int_0^{t_1} \ldots \int_0^{t_{i-2}} t_{i-1} dt_{i-1} \ldots dt_2 dt_1 = \frac{1}{i!},
$$
entonces
$$
| R_k(h) |
\leq \int_0^1 \int_0^{t_1} \ldots  \int_0^{t_{k-2}} \varepsilon t_{k-1} |h|^k  dt_{k-1}  \ldots dt_2 dt_1 = \frac{\varepsilon}{k!} |h|^k,
$$
y por lo tanto $\lim_{h \to 0}\frac{|R_k(h)|}{|h|^k} \leq  \frac{\varepsilon}{k!}$. Como esto vale para todo $\varepsilon$, concluímos que
$$
    R_k(h) = o(|h|^k).
$$

Veamos que
$$
R_k(h) = \varphi(1) - \sum_{j=0}^{k} \frac{1}{j!} \varphi^{(j)}(0).
$$
Claramente, si $k=2$,
\begin{align*}
    R_2(h) &= \int_0^1 \bigl( \varphi^{(1)}(t_{1}) - \varphi^{(1)}(0) - \varphi^{(2)}(0) t_{1} \bigr)  dt_1 \\
    &= \int_0^1 \bigl( \varphi^{(1)}(t_{1}) - \varphi^{(1)}(0) \bigr)  dt_1 - \int_0^1 t_{1} \varphi^{(2)}(0)   dt_1 \\
    &= \varphi^{(0)}(1) - \varphi^{(0)}(0) - \varphi^{(1)}(0)  - {\textstyle \frac{1}{2}} \varphi^{(2)}(0) .
\end{align*}
Con un argumento inductivo, supongamos que la relación vale para $k-1$ y veamos que vale para $k$. Así,
\begin{align*}
    R_k(h) =& \int_0^1 \int_0^{t_1} \ldots \int_0^{t_{k-2}} \bigl( \varphi^{(k-1)}(t_{k-1}) - \varphi^{(k-1)}(0) - \varphi^{(k)}(0) t_{k-1} \bigr) dt_{k-1} \ldots dt_2 dt_1 \\
    =& \int_0^1 \int_0^{t_1} \ldots \int_0^{t_{k-2}} \bigl( \varphi^{(k-1)}(t_{k-1}) - \varphi^{(k-1)}(0)  \bigr) dt_{k-1} \ldots dt_2 dt_1 \\
    &- \int_0^1 \int_0^{t_1} \ldots \int_0^{t_{k-2}}  t_{k-1} \varphi^{(k)}(0)  dt_{k-1} \ldots dt_2 dt_1 \\
    =& \int_0^1 \int_0^{t_1} \ldots \int_0^{t_{k-3}} \bigl( \varphi^{(k-2)}(t_{k-2}) - \varphi^{(k-2)}(0) - \varphi^{(k-1)}(0) t_{k-2} \bigr) dt_{k-2} \ldots dt_2 dt_1 \\
    &-{\textstyle \frac{1}{k!}}\varphi^{(k)}(0) \\
    =& R_{k-1}(h) -{\textstyle \frac{1}{k!}}\varphi^{(k)}(0) \\
    =& \varphi(1) - \sum_{j=0}^{k-1} {\textstyle \frac{1}{j!}} \varphi^{(j)}(0) -{\textstyle \frac{1}{k!}}\varphi^{(k)}(0).
\end{align*}


Concluyendo que
$$
f(\bar{x}+ h) = \sum_{j=0}^{k} \frac{1}{j!} f^{(j)}(\bar{x}) h^j + o(|h|^k).
$$

Note que para este resultado no fue necesario pedir la existencia de la derivada de orden $k+1$ de $f$, y tampoco fue necesario pedir que $f^{(k)}$ sea una función continua en $\bar{x}$.



Suponga ahora que $f:\mathbb{R}^n \to \mathbb{R}^m$ es $k$ veces diferenciable en $\bar{x}$. Para poder escribir el polinomio de Taylor de $f$ centrado en $\bar{x}$ tendremos que encontrar una expresión para las derivadas de orden superior.


Recuerde que la matriz Jacobiana $f'(x)\in \mathbb{R}^{m \times n}$ se construyó a partir de la función lineal $Df(x): \mathbb{R}^n \to \mathbb{R}^m$ dotando a los vectores de una estructura de columna y tomando $\frac{\partial f}{\partial x_j}(x) = Df(x)(e^j)$ como la $j$-ésima columna de $f'(x)$, por lo tanto
$$
Df(x)(v) = \sum_{j=1}^n v_j Df(x)(e^j) = \sum_{j=1}^n v_j \frac{\partial f}{\partial x_j}(x)
 = f'(x) v.
$$
Como $f$ es diferenciable en $\bar{x}$ entonces
$$
\left\| f(\bar{x}+h) - f(\bar{x}) - \sum_{j=1}^n h_j \frac{\partial f}{\partial x_j}(\bar{x}) \right\|
= \|f(\bar{x}+h) - f(\bar{x}) - Df(\bar{x})(h) \| = o(\|h\|).
$$

Si fijamos $v\in \mathbb{R}^n$ y tomamos $\tilde{f}:\mathbb{R}^n \to \mathbb{R}^m$ tal que $\tilde{f}(x) = Df(x)(v)$, entonces $\tilde{f}$ es diferenciable en $\bar{x}$ y como
$$
\tilde{f}(\bar{x}+th) = Df(\bar{x}+t h)(v) = \sum_{j=1}^n v_j \frac{\partial f}{\partial x_j}(\bar{x}+t h),
$$
entonces podemos definir $D^2 f(\bar{x})$ tal que
$$
D^2 f(\bar{x})(v,h) = D\tilde{f}(\bar{x})(h)= \sum_{j=1}^n \sum_{i=1}^n v_j h_i \frac{\partial^2 f}{\partial x_i \partial x_j}(\bar{x}).
$$
Como puede verse, $D^2 f(x): \mathbb{R}^n \times \mathbb{R}^n \to \mathbb{R}^m$ es una función bilineal (lineal en cada argumento). De la definición de diferenciabilidad en $\bar{x}$, tenemos que
$$
Df(\bar{x}+h)(v) - Df(\bar{x})(v) - D^2 f(\bar{x})(v,h) = \sum_{j=1}^n v_j\left( \frac{\partial f}{\partial x_j}(\bar{x}+h) - \frac{\partial f}{\partial x_j}(\bar{x}) -  \sum_{i=1}^n h_i \frac{\partial^2 f}{\partial x_i \partial x_j}(\bar{x}) \right),
$$
por lo tanto
$$
\left\| Df(\bar{x}+h)(v) - Df(\bar{x})(v) - D^2 f(\bar{x})(v,h)\right\| \leq \|v\| o(\|h\|).
$$

Repitiendo reiteradas veces el argumento utilizado para definir $D^2 f(x)$, podremos definir la función $k$-lineal $D^k f(x): \mathbb{R}^n \times \ldots \times \mathbb{R}^n \to \mathbb{R}^m$ tal que
$$
D^k f(x)(v^1, \ldots, v^k) = \sum_{i_1=1}^n \ldots \sum_{i_k=1}^n v^1_{i_1} \ldots v^k_{i_k} \frac{\partial^k f}{\partial x_{i_k} \ldots \partial x_{i_1}}(x),
$$
y como $f$ es $k$ veces diferenciable en $\bar{x}$, se tiene
$$
\frac{\|D^{k-1}\! f(\bar{x}+h)(v^1,\ldots,v^{k-1}) - D^{k-1}\! f(\bar{x})(v^1, \ldots,v^{k-1}) - D^k\! f(\bar{x})(v^1,\ldots,v^{k-1}, h)\|}{\|v^1\| \ldots \|v^{k-1}\|}
 \leq  o(\|h\|).
$$



Ahora podemos obtener el siguiente resultado.

>**Teorema**
>
> Suponga que $f:\mathbb{R}^n \to \mathbb{R}^m$ es $k$ veces diferenciable en $\bar{x}$. El polinomio de Taylor de grado $k$ para $f$ alrededor de $\bar{x}$ es $T_k:\mathbb{R}^n \to \mathbb{R}^m$ tal que
 $$
 T_k(h) = f(\bar{x}) + Df(\bar{x})(h) + \ldots + \frac{1}{k!} D^k f(\bar{x}) (h, \ldots, h),
 $$
 y satisface
 $$
    f(\bar{x} +h) = T_k(h) + o(\|h\|^k).
 $$
 Además, es el único polinomio de grado $k$ con tal propiedad.


**Demostración:**

Como este resultado se puede ver coordenada a coordenada, supongamos que $f:\mathbb{R}^n \to \mathbb{R}$. Luego, defina $\varphi:\mathbb{R} \to \mathbb{R}$ tal que $\varphi(t) = f(\bar{x} + t h)$. Veamos que $\varphi^{(k)}(t) = D^k f(\bar{x}+th)(h, \ldots, h)$. Note que
$$
\varphi'(t) = \sum_{j=1}^n \frac{\partial f}{\partial x_j}(\bar{x} + th) h_j = Df(\bar{x} + th)(h).
$$
Por inducción, suponga que la afirmación vale para $k-1$ y veamos que vale para $k$. Entonces
\begin{align*}
    \varphi^{(k)}(t) &= \frac{d}{dt}\bigl(\varphi^{(k-1)}(t) \bigr)  
    = \frac{d}{dt}\left( D^{k-1} f(\bar{x}+th)(h, \ldots, h) \right)\\
    &= \frac{d}{dt}\left( \sum_{i_1=1}^n \ldots \sum_{i_{k-1}=1}^n h_{i_1} \ldots h_{i_{k-1}} \frac{\partial^{k-1} f}{\partial x_{i_{k-1}} \ldots \partial x_{i_1}}(\bar{x}+th)\right) \\
    &=\sum_{i_1=1}^n \ldots \sum_{i_{k-1}=1}^n \sum_{i_{k}=1}^n h_{i_1} \ldots h_{i_{k-1}} h_{i_{k}} \frac{\partial^k f}{\partial x_{i_{k}} \partial x_{i_{k-1}} \ldots \partial x_{i_1}}(\bar{x}+th) \\
    &=D^k f(\bar{x}+th)(h, \ldots, h).
\end{align*}

Como $f$ es $k$ veces diferenciable, dado $\varepsilon>0$ y para $\|h\|$ pequeño se tiene que
$$
|D^{k-1}f(x+h)(h,\ldots, h) - D^{k-1}f(x)(h, \ldots,h) - D^k f(x)(h,\ldots,h, h)| \leq \varepsilon \|h\|^k.
$$
Por lo tanto para $t \in [0,1]$,
$$
|\varphi^{(k-1)}(t) - \varphi^{(k-1)}(0) - \varphi^{(k)}(0)t|
\leq \varepsilon t \|h\|^k.
$$
Con esto, definiendo
$$
R_k(h) = \int_0^1 \int_0^{t_1} \ldots \int_0^{t_{k-2}} \bigl( \varphi^{(k-1)}(t_{k-1}) - \varphi^{(k-1)}(0) - \varphi^{(k)}(0) t_{k-1} \bigr) dt_{k-1} \ldots dt_2 dt_1,
$$
se obtiene que $R_k(h) = o(\|h\|^k)$.

Concluyendo que
$$
f(\bar{x}+ h) = \sum_{j=0}^{k} \frac{1}{j!} D^j f(\bar{x}) (h, \ldots, h) + o(\|h\|^k).
$$


Si existiera otro polinomio de grado menor o igual a $k$, $\hat{T}_k:\mathbb{R}^n \to \mathbb{R}$ tal que $f(\bar{x} + h) = \hat{T}_k(h) + o(\|h\|^k)$ entonces $T_k(h) - \hat{T}_k(h) = o(\|h\|^k)$. Para $h$ fijo, sea $p:\mathbb{R} \to \mathbb{R}$ tal que $p(t) = T_k(t h) - \hat{T}_k(t h)$, luego $p$ es un polinomio de grado menor o igual a $k$, o sea, $p(t) = a_0 + a_1 t + \ldots + a_l t^l$ para $l\leq k$. Por lo tanto si $j\leq l$,
$$
\lim_{t\to 0} \frac{p(t)}{t^j} = \lim_{t\to 0} \frac{o(\|t h\|^k)}{t^j}
= \lim_{t\to 0} t^{k-j} \frac{o(t^k)}{t^k} = 0.
$$
Obteniendo que $a_j=0$ para todo $j$. Concluyendo que $0=p(1)=T_k(h) - \hat{T}_k(h)$ para $h$ arbitrario. $\spadesuit$




Al momento de realizar los cálculos del polinomio de Taylor en varias variables será de mucha utilidad el hecho que todas las derivadas parciales de orden $q$ para $2\leq q \leq k-1 $ conmutan. Esto es consecuencia directa de tener una función $k$ veces diferenciable.

Note que si las derivadas parciales de segundo orden son continuas, obtenemos que
$$
D^2 f(x)(h,h)
= \sum_{j=1}^n \sum_{i=1}^n h_j h_i \frac{\partial^2 f}{\partial x_i \partial x_j}(x)
=\sum_{j=1}^n h_j^2 \frac{\partial^2 f}{\partial x_j^2}(x)
+2 \sum_{j=1}^{n-1} \sum_{i=j+1}^n h_j h_i \frac{\partial^2 f}{\partial x_i \partial x_j}(x),
$$
donde la primera sumatoria tiene $n^2$ términos y la segunda $n(n+1)/2$.
Como puede apreciar, al incrementar el orden de la derivada tendremos factores asociados al número de repeticiones de ese término. Para tener una notación clara, denotaremos por $i_1 + \ldots +i_n = q$ a todas las posibles combinaciones de $n$ enteros no negativos $i_j$ tales que su suma sea $q$. Con esto, tenemos la expansión multinomial
$$
(x_1 + x_2 + \ldots +x_n)^q = \sum_{i_1 + \ldots + i_n = q} \frac{q!}{i_1!  \ldots  i_n!} x_1^{i_1} \ldots x_n^{i_n}.
$$

Con esta notación, obtenemos que
$$
\frac{1}{2!} D^2 f(x)(h,h)
= \sum_{i_1 + \ldots + i_n = 2} \frac{1}{i_1!  \ldots  i_n!} h_1^{i_1} \ldots h_n^{i_n} \frac{\partial^2 f}{\partial x_1^{i_1} \ldots \partial x_n^{i_n}}(x),
$$
y en general
$$
\frac{1}{q!} D^q f(x)(h, \ldots, h)
= \sum_{i_1 + \ldots + i_n = q} \frac{1}{i_1!  \ldots  i_n!} h_1^{i_1} \ldots h_n^{i_n} \frac{\partial^q f}{\partial x_1^{i_1} \ldots \partial x_n^{i_n}}(x),
$$



Claramente, tomando $h=x - \bar{x}$ obtenemos que
 $$
    f(x) = T_k(x-\bar{x}) + o(\|x-\bar{x}\|^k),
 $$
para $x$ cerca de $\bar{x}$.



Con este resultado podremos realizar aproximaciones como la siguiente.

#### Ejemplo

Sea $f:\mathbb{R}^3 \to \mathbb{R}$ tal que $f(x) = (x_1^2 + x_1 x_2^2) e^{x_3}$, calculemos el polinomio de Taylor de orden $2$ centrado en $\bar{x}=(1,2,0)$.

Las derivadas parciales hasta segundo orden son
\begin{align*}
\frac{\partial f}{\partial x_1}(x) &= (2 x_1 + x_2^2) e^{x_3},  &
\frac{\partial^2 f}{\partial x_1^2}(x) &= 2 e^{x_3},  &
\frac{\partial^2 f}{\partial x_1 \partial x_2}(x) &=  2 x_2 e^{x_3}, \\
\frac{\partial f}{\partial x_2}(x) &= 2 x_1 x_2 e^{x_3},  &
\frac{\partial^2 f}{\partial x_2^2}(x) &= 2 x_1 e^{x_3},  &
\frac{\partial^2 f}{\partial x_1 \partial x_3}(x) &= (2 x_1 + x_2^2) e^{x_3}, \\
\frac{\partial f}{\partial x_3}(x) &= (x_1^2 + x_1 x_2^2) e^{x_3},  &
\frac{\partial^2 f}{\partial x_3^2}(x) &= (x_1^2 + x_1 x_2^2) e^{x_3},  &
\frac{\partial^2 f}{\partial x_2 \partial x_3}(x) &= 2 x_1 x_2 e^{x_3}.
\end{align*}
Luego,
\begin{align*}
\frac{\partial f}{\partial x_1}(\bar{x}) &= 6 ,  &
\frac{\partial^2 f}{\partial x_1^2}(\bar{x}) &= 2 ,  &
\frac{\partial^2 f}{\partial x_1 \partial x_2}(\bar{x}) &=  4 , \\
\frac{\partial f}{\partial x_2}(\bar{x}) &= 4 ,  &
\frac{\partial^2 f}{\partial x_2^2}(\bar{x}) &= 2 ,  &
\frac{\partial^2 f}{\partial x_1 \partial x_3}(\bar{x}) &= 6 , \\
\frac{\partial f}{\partial x_3}(\bar{x}) &= 5 ,  &
\frac{\partial^2 f}{\partial x_3^2}(\bar{x}) &= 5 ,  &
\frac{\partial^2 f}{\partial x_2 \partial x_3}(\bar{x}) &= 4 .
\end{align*}

Con esto, $q(x)=T_2(x-\bar{x})=f(\bar{x}) + Df(\bar{x})(x-\bar{x}) + \frac{1}{2}D^2f(\bar{x})(x-\bar{x}, x-\bar{x}) $ es un polinomio en varias variables definido por
\begin{align*}
    q(x) =& 5 + 6(x_1-1) + 4(x_2-2) + 5x_3
    + \frac{1}{2} \left( 2(x_1-1)^2 + 2(x_2-2)^2 + 5 x_3^2 \right)\\
    &+  \left(4(x_1-1)(x_2-2) + 6(x_1-1)x_3 + 4(x_2-2)x_3\right)\\
    =& 5 + 6(x_1-1) + 4(x_2-2) + 5x_3
    + (x_1-1)^2 + (x_2-2)^2 + {\textstyle \frac{5}{2}} x_3^2\\
    &+ 4(x_1-1)(x_2-2) + 6(x_1-1)x_3 + 4(x_2-2)x_3.
\end{align*}

Cuando $f:\mathbb{R}^n \to \mathbb{R}$ podremos obtener una representación matricial para $D^2 f(x)$. Para ello, note que si $A\in \mathbb{R}^{n \times n}$ entonces
$$
    \langle A v,h\rangle = \sum_{i=1}^n (Av)_i h_i = \sum_{i=1}^n \left(\sum_{j=1}^n a_{ij} v_j \right) h_i
    =\sum_{i=1}^n \sum_{j=1}^n a_{ij}  h_i v_j.
$$
Definimos la **matriz Hessiana** de $f$ en $\bar{x}$, denotada por $\nabla^2 f(\bar{x})$, como aquella cuyo elemento $ij$ es $\frac{\partial^2 f}{\partial x_i \partial x_j}(\bar{x})$.
De la expresión obtenida anteriormente tenemos que
$$
D^2 f(x)(v,h) = \sum_{j=1}^n \sum_{i=1}^n v_j h_i \frac{\partial^2 f}{\partial x_i \partial x_j}(x) = \langle \nabla^2 f(\bar{x}) v, h \rangle.
$$



Luego, si $f:\mathbb{R}^n \to \mathbb{R}$ es dos veces diferenciable en $\bar{x}$, tenemos que
$$
f(x) = f(\bar{x}) + \langle \nabla f(\bar{x}), x-\bar{x} \rangle
+ {\textstyle \frac{1}{2} } \langle \nabla^2 f(\bar{x}) (x-\bar{x}), x-\bar{x} \rangle
+ o(\|x-\bar{x}\|^2).
$$



Como hemos visto, cuando las derivadas segundas no son continuas la matriz $\nabla^2 f(x)$ puede ser no simétrica. Esto no representa ningún inconveniente para la expresión anterior, pues ésta solo toma en cuenta la parte simétrica de la matriz. En general, si $A\in \mathbb{R}^{n \times n}$, la parte simétrica de $A$ es $A_s = \frac{1}{2}(A + A^{\mathrm{T}})$ y
$$
\langle A_s h, h \rangle
= \frac{1}{2} \langle A h, h \rangle + \frac{1}{2} \langle A^{\mathrm{T}} h, h \rangle
= \frac{1}{2} \langle A h, h \rangle + \frac{1}{2} \langle  h, A h \rangle
= \langle A h, h \rangle.
$$


## Optimización

Encontrar máximos y mínimos de funciones es de mucha utilidad al momento de resolver una gran diversidad de problemas.

En esta sección daremos una introducción a los problemas de minimización no lineal. Considere una función $f:\mathbb{R}^n \to \mathbb{R}$ y un conjunto de restricciones $D \subset \mathbb{R}^n$. Con esto, se desea hallar $\bar{x}$ solución de
\begin{equation*}
 \mathop{\mathrm{minimizar}}_{x\in D} \ f(x).
\end{equation*}

Note que maximizar $f$ en $D$ es equivalente a minimizar $-f$ en $D$. Con esto, todo resultado sobre minimización tiene su análogo para maximización. A fin de evitar redundancias, en esta sección trataremos solamente el problema de minimización.



Para poder analizar este problema será conveniente introducir una terminología adecuada.

>**Definición**
>
> Para el problema de minimizar $f$ sobre $D$, diremos que:
* $f$ es la función objetivo del problema,
* $D$ es el conjunto factible del problema,
* $\mathscr{v} \in [-\infty, \infty]$ es el mínimo del problema si $\displaystyle \mathscr{v} = \inf_ {x \in D} f(x)$,
* $\bar{x} \in D$ es un
 * minimizador global si $f(\bar{x}) \leq f(x)$ para todo $x\in D$,
 * minimizador local si existe $\varepsilon>0$ tal que $f(\bar{x}) \leq f(x)$ para todo $x\in D \cap B_{\varepsilon}(\bar{x})$.



Claramente, si $\bar{x}$ es un minimizador global entonces también es un minimizador local y $\mathscr{v} = f(\bar{x})$.

No es cierto que todo problema de minimización tiene solución, considere por ejemplo $f:\mathbb{R} \to \mathbb{R}$ tal que $f(x)=e^x$ con $D=\mathbb{R}$. Se puede ver que $\mathscr{v} = 0$, pero no existe un minimizador del problema. El estudio de existencia de solución de problemas de minimización es basto en la literatura, enunciaremos aquí uno de los resultados clásicos.



>**Teorema**(Weierstrass)
>
> Sea $D \in \mathbb{R}^n$ compacto no vacío y $f:\mathbb{R}^n \to \mathbb{R}$ continua en $D$. Entonces existe $\bar{x} \in D$ minimizador global de $f$ en $D$.


**Demostración:**

Como $D\neq \emptyset$ entonces $\mathscr{v} = \inf_ {x \in D} f(x) <  \infty$. Por definición de ínfimo, dado $k\in \mathbb{N}$ existe $x^k \in D$ tal que
$$
    f(x^k) <  \inf_ {x \in D} f(x) + \frac{1}{k}.
$$
Por lo visto para sucesiones, como $\{x^k\} \subset D$ y $D$ es compacto, entonces existe $\bar{x}$ tal que $x^{k_j} \to \bar{x}$. Luego, como $f$ es continua en $D$
$$
f(\bar{x}) = \lim_{j \to \infty} f(x^{k_j}) \leq \lim_{j \to \infty} \mathscr{v} + \frac{1}{k_j}
= \mathscr{v}.
$$
Concluyendo que $\mathscr{v}= f(\bar{x})$ y por lo tanto $\bar{x}$ es un minimizador global. $\spadesuit$




Para poder trabajar con minimización con restricciones, será de mucha utilidad ampliar un concepto ya conocido.

>**Definición**
>
> Sea $A\in \mathbb{R}^{n \times n}$ y $V\subset \mathbb{R}^n$ un subespacio. Diremos que $A$ es **definida positiva** en $V$ si $\langle Ad,d\rangle >0$ para todo $d \in V \setminus\{0\}$ y es **semidefinida positiva** si $\langle Ad,d\rangle \geq 0$ para todo $d \in V$. Diremos que $A$ es definida (semidefinida) negativa si $-A$ es definida (semidefinida) positiva en $V$. Si $A$ no es semidefinida positiva ni negativa en $V$, diremos que $A$ es indefinida en $V$.


En particular, si $V=\mathbb{R}^n$ uno tiene la definición estándar. Note además que si $A$ es indefinida en $V$ entonces $\langle A\hat{d},\hat{d}\rangle >0$ para alguna $\hat{d} \in V$ y $\langle A\check{d},\check{d}\rangle <0$ para alguna $\check{d} \in V$.
Una consecuencia directa del Teorema de Weierstrass es el siguiente.



>**Proposición**
>
> Sea $A\in \mathbb{R}^{n \times n}$ y $V\subset \mathbb{R}^n$ un subespacio, entonces $A$ es definida positiva en $V$ si y sólo si existe $\beta >0$ tal que
$$
\langle Ad,d\rangle \geq \beta \|d\|^2 \quad \forall \, d\in V.
$$

**Demostración:**

($\Leftarrow$). Si $d \in V\setminus\{0\}$ entonces $\langle Ad,d\rangle \geq \beta \|d\|^2>0$. Por lo tanto $A$ es definida positiva en $V$.

($\Rightarrow$). Suponga que $A$ es definida positiva en $V$. Defina $f:\mathbb{R}^n \to \mathbb{R}$ tal que $f(x) = \langle Ax,x\rangle$ y $S=\{ x\in V \mid \|x\|=1\}$. Como $f$ es continua y $S$ es compacto, entonces existe $\bar{x}$ minimizador global de $f$ en $S$. Si $\beta=f(\bar{x})$ entonces $\beta>0$ por ser $A$ definida positiva en $V$. Como la desigualdad vale si $d=0$, considere $d \in V \setminus \{0\}$. Como $V$ es un subespacio se tiene que $d/\|d\| \in V$ y por lo tanto $d/\|d\| \in S$, luego
$$
\langle Ad,d\rangle = \|d\|^2 \left\langle A {\textstyle\frac{d}{\|d\|}},{\textstyle\frac{d}{\|d\|}}\right\rangle
= \|d\|^2 f\bigl({\textstyle\frac{d}{\|d\|}}\bigr) \geq  \|d\|^2 \beta. \quad \spadesuit
$$





Cuando el conjunto donde se desea minimizar no es compacto, el Teorema de Weierstrass puede usarse como se muestra a continuación.

#### Ejemplo

Se desea resolver el problema de
\begin{align*}
    \textrm{minimizar}\quad & x_1^2 + x_2 + \frac{1}{x_1 + x_2} \\
    \textrm{sujeto a }\quad & x_1+x_2>0.
\end{align*}
Por lo tanto, la función objetivo es $f:\mathbb{R}^2 \to \mathbb{R}$ tal que $f(x) = x_1^2 + x_2 + \frac{1}{x_1 + x_2}$ y el conjunto factible es $D =\{ x\in \mathbb{R}^2 \mid x_1 + x_2 >0\}$.


```
#@title Curvas de nivel { display-mode: "form" }
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1,1)
#u, v = np.meshgrid(np.linspace(-5,5,100), np.linspace(0.1,10,100))
#c = 1/np.sqrt(2)
#s = -c
#x = c*u - s*v
#y = s*u + c*v
x, y = np.meshgrid(np.linspace(-5,5,100), np.linspace(-2,8,100))
Ind = (x+y<= 0)
f = x**2 + y + 1/(x+y)
f[Ind] = np.nan
ax.contour(x,y,f,np.arange(0.8,20,1))
fig.set_size_inches(6, 6)
ax.axis('equal')
ax.grid();
#plt.savefig('ejm_weierstrass.pdf')
```


Claramente $D$ no es compacto. Como $f(0,1)=2$, entonces si existiera un minimizador éste debería pertenecer a $\tilde{D} = \{x \in D \mid f(x) \leq 2\}$. Veamos que $f$ tiene un minimizador global en $\tilde{D}$. Como $f$ es continua en $\tilde{D}$, sólo basta demostrar que $\tilde{D}$ es compacto.

Note que si $x\in \tilde{D}$ entonces $x_1 + x_2 >0$ y
$$
2 \geq x_1^2 + x_2 + \frac{1}{x_1 + x_2} > x_1^2  -x_1 + \frac{1}{x_1 + x_2}
\geq -\frac{1}{4} + \frac{1}{x_1 + x_2},
$$
donde usamos que $x_1^2 - x_1 \geq -1/4$. Despejando, obtenemos que $\frac{9}{4}>\frac{1}{x_1 +x_2}$. Además, se obtiene que $0>x_1^2 - x_1 -2$, y como este polinomio se anula en $-1$ y $2$, concluímos que $x_1 \in (-1,2)$. Usando que $ x_1+x_2 > \frac{4}{9}$, se obtiene
$$
2 \geq x_1^2 + x_2 + \frac{1}{x_1 + x_2} > x_2 > \frac{4}{9} - x_1 > \frac{4}{9}-2=-\frac{14}{9},
$$
o sea, $x_2\in (-14/9, 2)$. Concluyendo que $\tilde{D}$ es acotado. Ahora, para cualquier sucesión $\{x^k\} \subset \tilde{D}$ con $x^k \to \hat{x}$ se tiene que $\hat{x}_1 + \hat{x}_2 \geq \frac{4}{9}$ y como $f$ es continua $f(\hat{x})\leq 2$. Obteniendo que $\hat{x} \in \tilde{D}$ y por lo tanto $\tilde{D}$ es cerrado.

Como $f$ es continua en $\tilde{D}$ compacto, tenemos que existe $\bar{x}$ minimizador global de $f$ en $\tilde{D}$. Luego, si $x \in D \setminus \tilde{D}$ entonces $f(\bar{x}) \leq 2 < f(x)$. Por lo tanto $\bar{x}$ es minimizador global de $f$ en $D$.


---


Los resultados de existencia nos garantizan que el problema puede resolverse, pero no nos dá ninguna información útil para hallar la solución del problema.

Cuando estudiamos el gradiente de una función, vimos que la dirección de menos el gradiente apunta en la dirección de máximo descenso. Tomando esto en cuenta, un mecanismo para minimizar $f$ en $D$ sería tomar $x\in D$ y avanzar en la dirección $-\nabla f(x)$. Infelizmente esta dirección puede estar apuntando hacia afuera del conjunto $D$.

Para mantenernos dentro de $D$ defina $d\in \mathcal{T}_D(x)$ como la proyección ortogonal de $-\nabla f(x)$ sobre $\mathcal{T}_D(x)$. Con esto $w = -\nabla f(x) - d$ es ortogonal a $d$ y $w \in \mathcal{N}_D(x)$. De la ortogonalidad de $d$ y $w$ obtenemos que
$$
 \langle \nabla f(x) , d \rangle = \langle -d - w , d \rangle=  -\|d\|^2.
$$

Luego, como $d\in \mathcal{T}_D(x)$, existe $\gamma:[0,\varepsilon] \to D$ tal que $\gamma(0)=x$ y $\gamma'(0) =d$ obteniendo que
\begin{align*}
    f(\gamma(t)) &= \varphi(t) = \varphi(0) + \varphi'(0) t + o(t) \\
    &= f(x) + t \langle \nabla f(x) , d\rangle + o(t) \\
    & = f(x) - t \|d\|^2 + o(t) = f(x) + t \left( \frac{o(t)}{t} - \|d\|^2\right) \\
    &< f(x),
\end{align*}
donde la desigualdad vale para $t>0$ pequeño y siempre que $\|d\| \neq 0$. Por lo tanto, siempre que $d\neq 0$ podremos encontrar $x^+ \in D$ tal que $f(x^+)<f(x)$. Este razonamiento nos conduce al siguiente resultado.



>**Teorema**
>
> Sea $f: D \subset \mathbb{R}^n \to \mathbb{R}$ diferenciable en $\bar{x} \in D$. Si $\bar{x}$ es un minimizador local de $f$ en $D$ entonces
 $$
  \langle \nabla f(\bar{x}), d \rangle \geq 0, \quad \forall\, d\in \mathcal{T}_D(\bar{x}).
 $$
 Equivalentemente, $-\nabla f(\bar{x}) \in \mathcal{N}_D(\bar{x})$.

**Demostración:**

Sea $d\in \mathcal{T}_D(\bar{x})$, entonces existe $\gamma:[0,\varepsilon] \to D$ con $\gamma(0)=\bar{x}$ y $\gamma'(0)=d$. Como $\bar{x}$ es minimizador local, para $t>0$ pequeño
$$
    f(\bar{x}) \leq f(\gamma(t)) = f(\bar{x}) + t \langle \nabla f(\bar{x}), d \rangle + o(t),
$$
luego, dividiendo por $t>0$,
$$
 \langle \nabla f(\bar{x}), d \rangle
 = \lim_{t \to 0^+} \langle \nabla f(\bar{x}), d \rangle + \frac{o(t)}{t} \geq 0. \quad \spadesuit
$$




Este resultado nos provee una condición necesaria que involucra las derivadas parciales $f$, por ello se conoce como condición necesaria de optimalidad de primer orden.

>**Definición**
>
> Diremos que $\bar{x}\in D$ es un **punto estacionario** (o crítico) del problema de minimización, si $-\nabla f(\bar{x}) \in \mathcal{N}_D(\bar{x})$.




Hay que remarcar que un punto estacionario puede no ser un minimizador. Por ejemplo, si el conjunto de direcciones normales es un subespacio entonces $ -\nabla (-f)(\bar{x}) = \nabla f(\bar{x}) \in \mathcal{N}_D(\bar{x})$, obteniendo que $\bar{x}$ también puede ser un maximizador local.
En esos casos suele ser conveniente utilizar además una condición necesaria de segundo orden. Para ello, usando la función auxiliar $\varphi(t) = f(\gamma(t))$, tenemos que $\varphi'(t) = \langle \nabla f(\gamma(t)), \gamma'(t) \rangle$ y
$$
\varphi''(t) = \langle \nabla^2 f (\gamma(t)) \gamma'(t), \gamma'(t) \rangle + \langle \nabla f(\gamma(t)), \gamma''(t) \rangle.
$$
Por lo tanto, si $\bar{x}$ es un minimizador de $f$ en $D$,
$$
f(\bar{x}) \leq f(\gamma(t))
= f(\bar{x}) + t \langle \nabla f(\bar{x}), d \rangle + \frac{t^2}{2} \left( \langle \nabla^2 f(\bar{x}) d, d \rangle + \langle \nabla f(\bar{x}), \gamma''(0) \rangle \right) + o(t^2).
$$
Note que la expresión anterior involucra la derivada segunda de $\gamma$, que nos provee información de la curvatura, en este caso del conjunto $D$. Por este motivo la condición necesaria de segundo orden tendrá expresiones distintas dependiendo de la estructura de $D$.




Por ejemplo, si $\bar{x} \in \operatorname{int}(D)$, que para la geometría local es equivalente a tomar $D=\mathbb{R}^n$, obtenemos las siguientes condiciones necesarias y suficientes para optimización sin restricciones.

>**Teorema**
>
> Sea $f:\mathbb{R}^n \to \mathbb{R}$.
>
> CN. Si $\bar{x}$ un minimizador local de $f$ en $\mathbb{R}^n$ y
1. $f$ es diferenciable en $\bar{x}$, entonces
$$
     \nabla f(\bar{x})=0.
$$
2. $f$ es dos veces diferenciable en $\bar{x}$, vale además que
$$
     \langle \nabla^2 f(\bar{x}) d, d \rangle \geq 0 \quad \forall\, d\in \mathbb{R}^n.
$$
>
> CS. Si $f$ es dos veces diferenciable en $\bar{x}$, $\nabla f(\bar{x})=0$ y
$$
     \langle \nabla^2 f(\bar{x}) d, d \rangle > 0 \quad \forall\, d\in \mathbb{R}^n,
$$
entonces $\bar{x}$ es un minimizador local de $f$ en $D$ y además existen $\varepsilon>0$ y $\alpha>0$ tal que
$$
     f(x) \geq f(\bar{x}) + \alpha \|x-\bar{x}\|^2 \quad \forall\, x\in B_\varepsilon(\bar{x}).
$$

**Demostración:**

(**CN**).
Como $D=\mathbb{R}^n$, entonces $\mathcal{T}_D(\bar{x})=\mathbb{R}^n$ y $\mathcal{N}_D(\bar{x})=\{0\}$.
Como $\bar{x}$ es un minimizador local, entonces $-\nabla f(\bar{x}) \in \mathcal{N}_D(\bar{x})=\{0\}$. Obteniendo la condición de primer orden.

Si $f$ es dos veces diferenciable en $\bar{x}$, para $d\in \mathcal{T}_D(\bar{x})=\mathbb{R}^n$ sea $\gamma$ dos veces diferenciable en $0$ con $\gamma'(0)=\bar{x}$ y $\gamma'(0)=d$. Luego, como $\nabla f(\bar{x})=0$,
\begin{align*}
f(\bar{x}) &\leq f(\gamma(t))
= f(\bar{x}) + t \langle \nabla f(\bar{x}), d \rangle + \frac{t^2}{2} \left( \langle \nabla^2 f(\bar{x}) d, d \rangle + \langle \nabla f(\bar{x}), \gamma''(0) \rangle \right) + o(t^2) \\
&= f(\bar{x})+ \frac{t^2}{2} \langle \nabla^2 f(\bar{x}) d, d \rangle + o(t^2),
\end{align*}
por lo tanto
$$
\langle \nabla^2 f(\bar{x}) d, d \rangle = \lim_{t\to 0^+} \langle \nabla^2 f(\bar{x}) d, d \rangle + 2 \frac{o(t^2)}{t^2} \geq 0.
$$
Como $d\in \mathbb{R}^n$ era arbitrario, obtenemos que $\nabla^2 f(\bar{x})$ es una matriz semidefinida positiva.

(**CS**). Como $\nabla^2 f(\bar{x})$ es una matriz definida positiva, existe $\beta>0$ tal que $\langle \nabla^2 f(\bar{x}) d, d \rangle \geq \beta \|d\|^2$. Luego, si $\nabla f(\bar{x})=0$ entonces
\begin{align*}
    f(x) &= f(\bar{x}) + \langle \nabla f(\bar{x}), x-\bar{x} \rangle + { \textstyle \frac{1}{2} } \langle \nabla^2 f(\bar{x}) (x-\bar{x}), x-\bar{x} \rangle + o(\| x-\bar{x} \|^2) \\
    & \geq f(\bar{x}) + \left( \frac{\beta}{2} + \frac{o(\| x-\bar{x} \|^2)}{\| x-\bar{x} \|^2}\right)\| x-\bar{x} \|^2.
\end{align*}
Por lo tanto, dado $0 <\alpha < \beta/2$ existe $\varepsilon>0$ tal que $\frac{\beta}{2} + \frac{o(\| x-\bar{x} \|^2)}{\| x-\bar{x} \|^2} > \alpha$ si $x \in B_\varepsilon(\bar{x})$.
Claramente esta desigualdad implica que $\bar{x}$ es un minimizador local. $\spadesuit$





#### Ejemplo

Se desea resolver el problema de
\begin{align*}
    \textrm{minimizar}\quad & x_1^2 + x_2 + \frac{1}{x_1 + x_2} \\
    \textrm{sujeto a }\quad & x_1+x_2>0.
\end{align*}
Como el conjunto factible es abierto, si existiera un minimizador $\bar{x}$ éste tiene que ser un punto interior, por lo tanto si denotamos por $f$ a la función objetivo tendremos que $\nabla f(\bar{x})=0$.
Como
$$
\nabla f(x) = \left[ \begin{array}{c} 2x_1 - \frac{1}{(x_1 + x_2)^2}  \\ 1 - \frac{1}{(x_1 + x_2)^2} \end{array}\right],
$$
si $\nabla f(\bar{x})=0$, entonces $(\bar{x}_1 + \bar{x}_2)^2 = 1$ y $1=2\bar{x}_1 (\bar{x}_1 + \bar{x}_2)^2 = 2\bar{x}_1$. Por lo tanto $\bar{x}_1 = 1/2$ y como $\bar{x}_1 + \bar{x}_2 >0$ obtenemos que $\bar{x}_2 = 1 - \bar{x}_1 = 1/2$. Concluyendo que $\bar{x}=(1/2, 1/2)$ es el único punto estacionario del problema.


Veamos si es un minimizador, maximizador o punto de ensilladura. Como
$$
\nabla^2 f(x) = \left[ \begin{array}{cc} 2 + \frac{2}{(x_1 + x_2)^3} & \frac{2}{(x_1 + x_2)^3}  \\ \frac{2}{(x_1 + x_2)^3} & \frac{2}{(x_1 + x_2)^3} \end{array}\right],
$$
entonces
$$
\nabla^2 f(\bar{x}) = \left[ \begin{array}{cc} 4 & 2  \\ 2 & 2 \end{array}\right].
$$
Esta matriz es definida positiva ya que
$$
\langle \nabla^2 f(\bar{x}) d, d\rangle = 4d_1^2 + 4 d_1 d_2 + 2 d_2^2 = 2 d_1^2 + 2(d_1 + d_2)^2
$$
es positivo siempre que $d\neq 0$. Concluyendo que $\bar{x}$ es un minimizador.

Ya que la aproximación afín es importante para la geometría de conjuntos generales, veamos las condiciones de optimalidad para este tipo de conjuntos.

>**Teorema**
>
> Sea $f:D \subset \mathbb{R}^n \to \mathbb{R}$ con $D= \{ x\in \mathbb{R}^n \mid Ax+b=0\}$ para $A\in\mathbb{R}^{m \times n}$ y $b\in \mathbb{R}^m$.
>
> CN. Si $\bar{x}$ un minimizador local de $f$ en $D$ y
1. $f$ es diferenciable en $\bar{x}$, entonces existe $\bar{\lambda} \in \mathbb{R}^m$ tal que
$$
     0=\nabla f(\bar{x}) + A^{\mathrm{T}} \bar{\lambda}.
$$
2. Si $f$ es dos veces diferenciable en $\bar{x}$, vale además que
$$
     \langle \nabla^2 f(\bar{x}) d, d \rangle \geq 0 \quad \forall\, d \in \operatorname{nu}(A).
$$
>
> CS. Si $f$ es dos veces diferenciable en $\bar{x}$ con $\bar{x}\in D$, $\nabla f(\bar{x}) + A^{\mathrm{T}} \bar{\lambda}=0$ para algún $\bar{\lambda} \in \mathbb{R}^m$ y
$$
     \langle \nabla^2 f(\bar{x}) d, d \rangle > 0 \quad \forall\, d\in \operatorname{nu}(A) \setminus\{0\},
$$
entonces $\bar{x}$ es un minimizador local de $f$ en $D$ y además existen $\varepsilon>0$ y $\alpha>0$ tal que
$$
     f(x) \geq f(\bar{x}) + \alpha \|x-\bar{x}\|^2 \quad \forall\, x\in D \cap B_\varepsilon(\bar{x}).
$$

**Demostración:**

(**CN**). Si $\bar{x} \in D=\{ x\in \mathbb{R}^n \mid Ax+b=0\}$, entonces $\mathcal{T}_D(\bar{x})=\operatorname{nu}(A)$ y $\mathcal{N}_D(\bar{x}) = \operatorname{img}(A^{\mathrm{T}})$.
Como $\bar{x}$ es un minimizador local obtenemos que $-\nabla f(\bar{x})\in \mathcal{N}_D(\bar{x}) = \operatorname{img}(A^{\mathrm{T}})$, o sea, existe $\bar{\lambda} \in \mathbb{R}^m$ tal que $-\nabla f(\bar{x}) = A^{\mathrm{T}} \bar{\lambda}$.

Si $f$ es dos veces diferenciable en $\bar{x}$, para $d\in \mathcal{T}_D(\bar{x})=\operatorname{nu}(A)$ sea $\gamma$ dos veces diferenciable en $0$ con $\gamma'(0)=\bar{x}$, $\gamma'(0)=d$ y $A\gamma(t) + b=0$. Como $A\gamma'(t) = 0$, entonces $A \gamma''(t) = 0$.
Luego, como $\nabla f(\bar{x}) + A^{\mathrm{T}} \bar{\lambda}=0$,
$$
0 = \langle \nabla f(\bar{x})  +  A^{\mathrm{T}} \bar{\lambda}, d \rangle
    = \langle \nabla f(\bar{x}), d \rangle  + \langle  \bar{\lambda}, Ad \rangle
    = \langle \nabla f(\bar{x}), d \rangle.
$$
De manera similar se obtiene que $\langle \nabla f(\bar{x}), \gamma''(0) \rangle = 0$. Con esto,
\begin{align*}
f(\bar{x}) &\leq f(\gamma(t))
= f(\bar{x}) + t \langle \nabla f(\bar{x}), d \rangle + \frac{t^2}{2} \left( \langle \nabla^2 f(\bar{x}) d, d \rangle + \langle \nabla f(\bar{x}), \gamma''(0) \rangle \right) + o(t^2) \\
&= f(\bar{x}) + \frac{t^2}{2} \langle \nabla^2 f(\bar{x}) d, d \rangle + o(t^2),
\end{align*}
por lo tanto
$$
\langle \nabla^2 f(\bar{x}) d, d \rangle = \lim_{t\to 0^+} \langle \nabla^2 f(\bar{x}) d, d \rangle + 2 \frac{o(t^2)}{t^2} \geq 0,
$$
para todo $d \in \operatorname{nu}(A)$.

(**CS**). Como $\nabla^2 f(\bar{x})$ es una matriz definida positiva en el subespacio $\operatorname{nu}(A)$, existe $\beta>0$ tal que $\langle \nabla^2 f(\bar{x}) d, d \rangle \geq \beta \|d\|^2$ para todo $d \in \operatorname{nu}(A)$. Ahora, por un lado, si $\nabla f(\bar{x}) + A^{\mathrm{T}} \bar{\lambda}=0$ entonces $\langle \nabla f(\bar{x}), d \rangle = 0$ para todo $d\in \operatorname{nu}(A)$. Por otro lado, si $x\in D$ entonces $0=Ax+b = Ax + b - (A\bar{x}+b) = A(x-\bar{x})$, o sea, $x-\bar{x} \in \operatorname{nu}(A)$. Juntando todo, obtenemos que
\begin{align*}
    f(x) &= f(\bar{x}) + \langle \nabla f(\bar{x}), x-\bar{x} \rangle + { \textstyle \frac{1}{2} } \langle \nabla^2 f(\bar{x}) (x-\bar{x}), x-\bar{x} \rangle + o(\| x-\bar{x} \|^2) \\
    & \geq f(\bar{x}) + \left( \frac{\beta}{2} + \frac{o(\| x-\bar{x} \|^2)}{\| x-\bar{x} \|^2}\right)\| x-\bar{x} \|^2.
\end{align*}
Por lo tanto, dado $0 <\alpha < \beta/2$ existe $\varepsilon>0$ tal que $\frac{\beta}{2} + \frac{o(\| x-\bar{x} \|^2)}{\| x-\bar{x} \|^2} > \alpha$ si $x \in B_\varepsilon(\bar{x})$.
Claramente esta desigualdad implica que $\bar{x}$ es un minimizador local. $\spadesuit$




Cuando $D= \{ x\in \mathbb{R}^n \mid Ax+b=0\}$ entonces $\bar{x}\in D$ es un punto estacionario si y sólo si existe $\bar{\lambda}$ tal que $0=\nabla f(\bar{x}) + A^{\mathrm{T}} \bar{\lambda}$.

Note que la condiciones anteriores pueden escribirse en términos de una función auxiliar.

>**Definición**
>
> Diremos que $L:\mathbb{R}^n \times \mathbb{R}^m \to \mathbb{R}$ tal que
$$
L(x, \lambda) = f(x) + \langle \lambda, h(x) \rangle,
$$
es la **función Lagrangiana** asociada al problema de minimizar $f$ en $D=\{x\in \mathbb{R}^n \mid h(x) = 0\}$ con $h:\mathbb{R}^n \to \mathbb{R}^m$.
>
> Diremos que $\bar{\lambda} \in \mathbb{R}^m$ es un vector de **multiplicadores de Lagrange** asociado a $\bar{x}$ si
$$
0 = \left[ \begin{array}{c} \nabla_x L(\bar{x}, \bar{\lambda})  \\  h(\bar{x}) \end{array}\right].
$$





Como $L(x, \lambda) = f(x) + \langle \lambda, h(x) \rangle = f(x) + \sum_{j=1}^m \lambda_j h_j(x)$, se tiene que
$$
\nabla_x L(x,\lambda) = \nabla f(x) + \sum_{j=1}^m \lambda_j \nabla h_j(x), \quad
\nabla_{xx}^2 L(x,\lambda) = \nabla^2 f(x) + \sum_{j=1}^m \lambda_j \nabla^2 h_j(x).
$$

En particular, si $h(x) = Ax+b$ obtenemos que $h_j(x) = \langle a^j, x\rangle + b_j$, donde $(a^j)^{\mathrm{T}}$ es la $j$-ésima fila de $A$. Luego,
$$
\nabla_x L(x,\lambda) = \nabla f(x) + \sum_{j=1}^m \lambda_j a^j =\nabla f(x) + A^{\mathrm{T}} \lambda, \quad
\nabla_{xx}^2 L(x,\lambda) = \nabla^2 f(x).
$$
Con esto, el resultado anterior nos dice que $\bar{\lambda}$ es un vector de multiplicadores de Lagrange y que la matriz $\nabla_{xx}^2 L(\bar{x},\bar{\lambda})$ es semidefinida positiva en $\operatorname{nu}(A)$.


Ahora podemos esperar un resultado similar para conjuntos generales.

>**Teorema**
>
> Sea $f:D \subset \mathbb{R}^n \to \mathbb{R}$ con $D= \{ x\in \mathbb{R}^n \mid h(x)=0\}$ para $h:\mathbb{R}^n \to \mathbb{R}^m$.
>
> CN. Si $\bar{x}$ un minimizador local de $f$ en $D$ y
1. $f$ es diferenciable en $\bar{x}$ con $h$ es continuamente diferenciable en un entorno de $\bar{x}$ tal que $\nabla h_1(\bar{x}), \ldots, \nabla h_m(\bar{x})$ son linealmente independientes, entonces existe $\bar{\lambda} \in \mathbb{R}^m$ tal que
$$
  0=\nabla_x L(\bar{x},\bar{\lambda}) =\nabla f(\bar{x}) + \sum_{j=1}^m \bar{\lambda}_j \nabla h_j(\bar{x}).
$$
2. Si $f$ es dos veces diferenciable en $\bar{x}$, vale además que
$$
 \left\langle  \nabla_{xx}^2 L(\bar{x},\bar{\lambda})  d, d \right\rangle \geq 0 \quad
 \forall\, d \in \operatorname{nu}(h'(\bar{x})).
$$
>
> CS. Si $f$ es dos veces diferenciable en $\bar{x}$ con $h(\bar{x})=0$, $\nabla_x L(\bar{x},\bar{\lambda})=0$ para algún $\bar{\lambda} \in \mathbb{R}^m$ y
$$
  \langle \nabla_{xx}^2 L(\bar{x},\bar{\lambda}) d, d \rangle > 0 \quad \forall\, d\in \operatorname{nu}(h'(\bar{x})) \setminus\{0\},
$$
entonces $\bar{x}$ es un minimizador local de $f$ en $D$ y además existen $\varepsilon>0$ y $\alpha>0$ tal que
$$
  f(x) \geq f(\bar{x}) + \alpha \|x-\bar{x}\|^2 \quad \forall\, x\in D \cap B_\varepsilon(\bar{x}).
$$


**Demostración:**

(**CN**). Note que la $j$-ésima fila de $h'(\bar{x})$ es $\nabla h_j(\bar{x})$.
Como $D=\{ x\in \mathbb{R}^n \mid h(x)=0\}$, $h(\bar{x})=0$ y $h'(\bar{x})$ tiene filas linealmente independientes, entonces $\mathcal{T}_D(\bar{x})=\operatorname{nu}(h'(\bar{x}))$ y $\mathcal{N}_D(\bar{x}) = \operatorname{img}(h'(\bar{x})^{\mathrm{T}})$.
Como $\bar{x}$ es un minimizador local obtenemos que $-\nabla f(\bar{x})\in \mathcal{N}_D(\bar{x}) = \operatorname{img}(h'(\bar{x})^{\mathrm{T}})$, o sea, existe $\bar{\lambda} \in \mathbb{R}^m$ tal que $-\nabla f(\bar{x}) = h'(\bar{x})^{\mathrm{T}} \bar{\lambda} = \sum_{j=1}^m \bar{\lambda}_j \nabla h_j(\bar{x})$.

Si $f$ y $h$ son dos veces diferenciable en $\bar{x}$, para $d\in \operatorname{nu}(h'(\bar{x}))$ sea $\gamma$ dos veces diferenciable en $0$ con $\gamma'(0)=\bar{x}$, $\gamma'(0)=d$ y $h(\gamma(t))=0$. Como $0=\langle \bar{\lambda}, h(\gamma(t)) \rangle$, entonces al derivar obtenemos
\begin{align*}
    0&=\langle \bar{\lambda}, h'(\gamma(t))\gamma'(t) \rangle
    = \sum_{j=1}^m \bar{\lambda}_j \langle \nabla h_j(\gamma(t)), \gamma'(t) \rangle \\
    0&=\sum_{j=1}^m \bar{\lambda}_j \langle \nabla^2 h_j(\gamma(t)) \gamma'(t), \gamma'(t) \rangle
    + \sum_{j=1}^m \bar{\lambda}_j \langle \nabla h_j(\gamma(t)), \gamma''(t) \rangle.
\end{align*}
Evaluando en $t=0$ y usando que $0=\nabla f(\bar{x}) + \sum_{j=1}^m \bar{\lambda}_j \nabla h_j(\bar{x})$, obtenemos
\begin{align*}
\langle \nabla f(\bar{x}) , d \rangle
&= - \left\langle {\textstyle \sum_{j=1}^m} \bar{\lambda}_j \nabla h_j(\bar{x}) , d \right\rangle
= 0,\\
\langle \nabla f(\bar{x}) , \gamma''(0) \rangle
&= - \left\langle {\textstyle \sum_{j=1}^m} \bar{\lambda}_j \nabla h_j(\bar{x}) , \gamma''(0) \right\rangle
= \sum_{j=1}^m \bar{\lambda}_j \langle \nabla^2 h_j(\bar{x}) d, d \rangle.    
\end{align*}
Con esto,
\begin{align*}
f(\bar{x}) &\leq f(\gamma(t))
= f(\bar{x}) + t \langle \nabla f(\bar{x}), d \rangle + \frac{t^2}{2} \left( \langle \nabla^2 f(\bar{x}) d, d \rangle + \langle \nabla f(\bar{x}), \gamma''(0) \rangle \right) + o(t^2) \\
&= f(\bar{x}) + \frac{t^2}{2} \left( \langle \nabla^2 f(\bar{x}) d, d \rangle + {\textstyle\sum_{j=1}^m} \bar{\lambda}_j \langle \nabla^2 h_j(\bar{x}) d, d \rangle \right) + o(t^2) \\
&= f(\bar{x}) + \frac{t^2}{2} \left\langle \nabla_{xx}^2 L(\bar{x},\bar{\lambda}) d, d \right\rangle + o(t^2),
\end{align*}
por lo tanto,
$$
\left\langle \nabla_{xx}^2 L(\bar{x},\bar{\lambda}) d, d \right\rangle
= \lim_{t \to 0^+} \left\langle \nabla_{xx}^2 L(\bar{x},\bar{\lambda}) d, d \right\rangle  + 2 \frac{o(t^2)}{t^2}\geq 0,
$$
para todo $d \in \operatorname{nu}(h'(\bar{x}))$.


(**CS**). Como $\nabla_{xx}^2 L(\bar{x},\bar{\lambda})$ es una matriz definida positiva en $\operatorname{nu}(h'(\bar{x}))$, existe $\beta>0$ tal que $\langle \nabla_{xx}^2 L(\bar{x},\bar{\lambda}) d, d \rangle \geq \beta \|d\|^2$ para todo $d \in \operatorname{nu}(h'(\bar{x}))$.
Si la desigualdad no es cierta, entonces para cada $k\in \mathbb{N}$ existe $x^k \in D \cap B_{1/k}(\bar{x})$ tal que
$$
f(x^k) < f(\bar{x}) + \frac{1}{k} \|x^k-\bar{x}\|^2.
$$
Como $h(\bar{x})=0$ y $h(x^k)=0$ entonces $L(\bar{x},\bar{\lambda})=f(\bar{x})$ y $L(x^k,\bar{\lambda}) = f(x^k)$, luego
\begin{align*}
    0 &= h(x^k) - h(\bar{x}) =   h'(\bar{x})(x^k - \bar{x}) + o(\| x^k - \bar{x}\|),\\
    {\textstyle \frac{1}{k}} \|x^k-\bar{x}\|^2 &> L(x^k,\bar{\lambda}) - L(\bar{x},\bar{\lambda})\\
    &= \langle \nabla_x L(\bar{x}, \bar{\lambda}), x^k-\bar{x} \rangle + { \textstyle \frac{1}{2} } \langle \nabla_{xx}^2 L(\bar{x}, \bar{\lambda}) (x^k-\bar{x}), x^k-\bar{x} \rangle + o(\| x^k-\bar{x} \|^2) \\
    &= { \textstyle \frac{1}{2} } \langle \nabla_{xx}^2 L(\bar{x}, \bar{\lambda}) (x^k-\bar{x}), x^k-\bar{x} \rangle + o(\| x^k-\bar{x} \|^2).
\end{align*}
Si $d^k = (x^k-\bar{x})/\| x^k -\bar{x} \|$, entonces $\{d^k\}$ es acotado y existe $d^{k_j} \to \bar{d}$ con $\|\bar{d}\| = 1$. Como $x^k \to \bar{x}$, dividiendo la primera ecuación por $\| x^{k_j}-\bar{x} \|$ y tomando límite obtenemos que
$$
0 = h'(\bar{x}) \bar{d},
$$
por lo tanto $\bar{d} \in \operatorname{nu}(h'(\bar{x}))$. Dividiendo la desigualdad anterior por $ \|x^{k_j} - \bar{x}\|^2$ y tomando límite tenemos que
$$
0 \geq { \textstyle \frac{1}{2} } \langle \nabla_{xx}^2 L(\bar{x}, \bar{\lambda}) \bar{d}, \bar{d} \rangle
    \geq { \textstyle \frac{1}{2} }\beta \|\bar{d}\|^2 = { \textstyle \frac{1}{2} }\beta,
$$
contradiciendo que $\beta>0$. Por lo tanto deben existir $\varepsilon>0$ y $\alpha>0$ tal que si $x\in D \cap B_\varepsilon(\bar{x})$ entonces
$$
 f(x) \geq f(\bar{x}) + \alpha \|x-\bar{x}\|^2.
$$
Esta desigualdad implica que $\bar{x}$ es un minimizador local. $\spadesuit$





En este caso, cuando $D= \{ x\in \mathbb{R}^n \mid h(x)=0\}$ entonces $\bar{x}\in D$, con $h'(\bar{x})$ de rango $m$, es un punto estacionario si y sólo si existe $\bar{\lambda}$ tal que $0=\nabla f(\bar{x}) + h'(\bar{x})^{\mathrm{T}} \bar{\lambda}$.


#### Ejemplo

Se desea hallar el paralelepípedo de lados paralelos a los ejes con mayor volumen circunscripto en el elipsoide $\frac{x_1^2}{a_1^2} + \frac{x_2^2}{a_2^2} + \frac{x_3^2}{a_3^2} = 1$, con $a_i>0$.

Como los vértices del paralelepípedo son de la forma $(\pm x_1, \pm x_2, \pm x_3)$ para $x$ en el elipsoide, el problema a resolver es
\begin{align*}
    \textrm{minimizar}\quad & - x_1 x_2 x_3 \\
    \textrm{sujeto a }\quad & \frac{x_1^2}{a_1^2} + \frac{x_2^2}{a_2^2} + \frac{x_3^2}{a_3^2} - 1= 0.
\end{align*}
Denotemos por $f$ a la función objetivo y por $h$ a la restricción. Los gradientes de las mismas son
$$
\nabla f (x) = \left[\begin{array}{c} - x_2 x_3 \\ - x_1 x_3  \\ - x_1 x_2 \end{array} \right], \qquad
\nabla h(x) = \left[\begin{array}{c} \frac{2}{a_1^2}x_1 \\ \frac{2}{a_2^2} x_2\\ \frac{2}{a_3^2} x_3\end{array} \right].
$$
Como $\nabla h(\bar{x}) \neq 0$ si $h(\bar{x})=0$, entonces existe $\bar\lambda$ tal que
$$
0 = \nabla f(\bar{x}) + \bar{\lambda} \nabla h(\bar{x})
= \left[\begin{array}{c} - \bar{x}_2 \bar{x}_3 \\ - \bar{x}_1 \bar{x}_3  \\ - \bar{x}_1 \bar{x}_2 \end{array} \right] + \bar{\lambda} \left[\begin{array}{c} \frac{2}{a_1^2} \bar{x}_1 \\ \frac{2}{a_2^2} \bar{x}_2 \\ \frac{2}{a_3^2} \bar{x}_3\end{array} \right]
=  \left[\begin{array}{c} - \bar{x}_2 \bar{x}_3 + \frac{2\bar{\lambda}}{a_1^2} \bar{x}_1  \\ - \bar{x}_1 \bar{x}_3 + \frac{2\bar{\lambda}}{a_2^2} \bar{x}_2  \\ - \bar{x}_1 \bar{x}_2 + \frac{2\bar{\lambda}}{a_3^2} \bar{x}_3 \end{array} \right].
$$
Multiplicando por $\bar{x}$ tenemos que $0 = \langle \bar{x}, \nabla f(\bar{x}) + \bar{\lambda} \nabla h(\bar{x}) \rangle  = - 3 \bar{x}_1 \bar{x}_2 \bar{x}_3 + 2 \bar{\lambda}$.
Del sistema de ecuaciones anterior obtenemos que
$$
0 = \bar{x}_2 \bar{x}_3 \left( \frac{3}{a_1^2} \bar{x}_1^2 -1 \right), \quad
0 = \bar{x}_1 \bar{x}_3 \left( \frac{3}{a_2^2} \bar{x}_2^2 -1 \right), \quad
0 = \bar{x}_1 \bar{x}_2 \left( \frac{3}{a_3^2} \bar{x}_3^2 -1 \right).
$$
Tomando en cuenta que $h(\bar{x})=0$, obtenemos los pares
$(\bar{x}^0, \bar{\lambda}^0)=\left( \bigl( \frac{a_1}{\sqrt{3}}, \frac{a_2}{\sqrt{3}}, \frac{a_3}{\sqrt{3}}\bigr), \frac{a_1 a_2 a_3}{2\sqrt{3}} \right)$, $(\bar{x}^1, \bar{\lambda}^1)=\bigl( (a_1, 0,0), 0 \bigr)$, $(\bar{x}^2, \bar{\lambda}^2)=\bigl( (0, a_2,0), 0 \bigr)$, $(\bar{x}^3, \bar{\lambda}^3)=\bigl( (0, 0,a_3), 0 \bigr)$.

Claramente $f$ alcanza su mínimo en el punto estacionario $\bar{x}^0$, donde $f(\bar{x}^0)=-\frac{a_1 a_2 a_3}{3\sqrt{3}}$. Concluyendo que el volumen máximo del paralelepípedo es $8\frac{a_1 a_2 a_3}{3\sqrt{3}}$ y los vértices de éste se obtienen reflejando $\bigl( \frac{a_1}{\sqrt{3}}, \frac{a_2}{\sqrt{3}}, \frac{a_3}{\sqrt{3}}\bigr)$.

Aunque el problema ya ha sido resuelto, realicemos el análisis de segundo orden del mismo. Note que
$$
\nabla_{xx}^2 L (x,\lambda) = \left[\begin{array}{ccc}
     \frac{2 \lambda}{a_1^2}& -x_3                      & -x_2                  \\
     -x_3                   &  \frac{2 \lambda}{a_2^2}  & -x_1                  \\
     -x_2                   & -x_1                      &\frac{2 \lambda}{a_3^2}
\end{array} \right].
$$
Luego,
$$
\nabla_{xx}^2 L (\bar{x}^0, \bar{\lambda}^0) = \left[\begin{array}{ccc}
     \frac{a_2 a_3}{a_1\sqrt{3}}& -\frac{a_3}{\sqrt{3}}         & -\frac{a_2}{\sqrt{3}}         \\
     -\frac{a_3}{\sqrt{3}}      & \frac{a_1 a_3}{a_2\sqrt{3}}   & -\frac{a_1}{\sqrt{3}}         \\
     -\frac{a_2}{\sqrt{3}}      & -\frac{a_1}{\sqrt{3}}         & \frac{a_1 a_2}{a_3\sqrt{3}}
\end{array} \right], \quad
\operatorname{nu}(h'(\bar{x}^0)) = \operatorname{img}\left( B \right), \
B = \left[\begin{array}{cc} -a_1 & -a_1  \\ a_2 & 0 \\ 0 & a_3  \end{array} \right],
$$
ya que $h'(\bar{x}^0) = \bigl[ \frac{2}{\sqrt{3} a_1} \ \frac{2}{\sqrt{3} a_2} \ \frac{2}{\sqrt{3} a_3} \bigr]$. Con esto
\begin{align*}
    \langle \nabla_{xx}^2 L (\bar{x}^0, \bar{\lambda}^0) d, d \rangle
    &=\langle \nabla_{xx}^2 L (\bar{x}^0, \bar{\lambda}^0) Bv, Bv \rangle
    =\langle B^{\mathrm{T}} \nabla_{xx}^2  L (\bar{x}^0, \bar{\lambda}^0) Bv, v \rangle\\
    &= \frac{2a_1 a_2 a_3}{\sqrt{3}} \left\langle \left[\begin{array}{cc} 2 & 1  \\ 1 & 2 \end{array} \right] v, v \right\rangle = \frac{2a_1 a_2 a_3}{\sqrt{3}} \left( 2v_1^2 + 2 v_1 v_2 +2 v_2^2\right) \\
    &= \frac{2a_1 a_2 a_3}{\sqrt{3}} \left( v_1^2 + (v_1 + v_2)^2 + v_2^2\right)
    \geq \frac{2a_1 a_2 a_3}{\sqrt{3}} \|v\|^2.
\end{align*}
Por lo tanto $\bar{x}^0$ es un minimizador.

Para los otros puntos, por ejemplo $(\bar{x}^1, \bar{\lambda}^1)$, se tiene que
$$
\nabla_{xx}^2 L (\bar{x}^1, \bar{\lambda}^1) = \left[\begin{array}{ccc}
     0  & 0     & 0    \\
     0  & 0     & -a_1 \\
     0  & -a_1  & 0
\end{array} \right], \quad
\operatorname{nu}(h'(\bar{x}^1)) = \operatorname{img}\left( B \right), \
B = \left[\begin{array}{cc} 0 & 0  \\ 1 & 0 \\ 0 & 1  \end{array} \right],
$$
ya que $h'(\bar{x}^1) = \bigl[ \frac{2}{a_1} \ 0 \ 0 \bigr]$. Con esto,
\begin{align*}
    \langle \nabla_{xx}^2 L (\bar{x}^1, \bar{\lambda}^1) d, d \rangle
    &=\langle \nabla_{xx}^2 L (\bar{x}^1, \bar{\lambda}^1) Bv, Bv \rangle
    =\langle B^{\mathrm{T}} \nabla_{xx}^2  L (\bar{x}^1, \bar{\lambda}^1) Bv, v \rangle\\
    &=  \left\langle \left[\begin{array}{cc} 0 & -a_1  \\ -a_1 & 0 \end{array} \right] v, v \right\rangle = -2a_1 v_1 v_2.
\end{align*}
Por lo tanto $\bar{x}^1$ es un punto de ensilladura. Lo mismo puede probarse para $\bar{x}^2$ y $\bar{x}^3$.

Note que para hallar un punto estacionario se debe hallar un cero de $F:\mathbb{R}^n \times \mathbb{R}^m \to \mathbb{R}^n \times \mathbb{R}^m $ tal que
$$
F(x,\lambda) = \left[ \begin{array}{c} \nabla_x L(x,\lambda) \\ h(x) \end{array}\right].
$$
Como hemos visto, si $F(\bar{x}, \bar{\lambda})=0$ podemos garantizar la estabilidad de $(\bar{x}, \bar{\lambda})$ si  $F'(\bar{x}, \bar{\lambda})$ es invertible. Como
$$
F'(x,\lambda) = \left[ \begin{array}{cc} \nabla_{xx}^2 L(x,\lambda) &  h'(x)^{\mathrm{T}} \\ h'(x) & 0  \end{array}\right],
$$
entonces $F'(\bar{x}, \bar{\lambda}) \left[ \begin{array}{c} d \\ \nu \end{array}\right]=0$ si y sólo si
$$
0 = \nabla_{xx}^2 L(\bar{x},\bar{\lambda}) d + h'(\bar{x})^{\mathrm{T}} \nu, \quad 0 = h'(\bar{x}) d.
$$
La segunda ecuación implica que $d \in \operatorname{nu}(h'(\bar{x}))$ y de la primera se obtiene
$$
 0 = \langle \nabla_{xx}^2 L(\bar{x},\bar{\lambda}) d + h'(\bar{x})^{\mathrm{T}} \nu, d \rangle
   = \langle \nabla_{xx}^2 L(\bar{x},\bar{\lambda}) d , d \rangle,
$$
Obteniendo que $d=0$ si $\nabla_{xx}^2 L(\bar{x},\bar{\lambda})$ es definida positiva en $\operatorname{nu}(h'(\bar{x}))$. Además, si $h'(\bar{x})$ tiene rango $m$ entonces
$$
0 = \nabla_{xx}^2 L(\bar{x},\bar{\lambda}) d + h'(\bar{x})^{\mathrm{T}} \nu
  = h'(\bar{x})^{\mathrm{T}} \nu,
$$
obteniendo que $\nu=0$. Concluyendo que $F'(\bar{x}, \bar{\lambda})$ es invertible si $\nabla_{xx}^2 L(\bar{x},\bar{\lambda})$ es definida positiva en $\operatorname{nu}(h'(\bar{x}))$ con $h'(\bar{x})$ de rango $m$.
