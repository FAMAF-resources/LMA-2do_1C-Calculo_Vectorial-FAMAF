# Geometría del espacio Euclídeo


## Longitud y perpendicularidad

Para poder obtener nociones de longitud, ángulo y perpendicularidad deberemos introducir otras operaciones entre elementos de $\mathbb{R}^n$. Dados $x = (x_1, x_2, \ldots, x_n)$ e $y = (y_1, y_2, \ldots, y_n)$ definimos el producto interno Euclídeo (o producto punto) como el número
$$
\langle x, y \rangle = x_1 y_1 + x_2 y_2 + \ldots + x_n y_n.
$$
En la literatura, este producto también suele denotarse por $x \cdot y$. Se puede verificar que para $x, y, z\in \mathbb{R}^n$ y $\alpha \in \mathbb{R}$ valen:

1. $\langle x, y \rangle = \langle y , x\rangle$,
2. $\langle x, y + z \rangle = \langle x, y \rangle + \langle x, z \rangle$,
3. $\langle \alpha x, y \rangle = \alpha \langle x, y \rangle$,
4. $\langle x, x \rangle > 0$ para $x\neq 0$.

Note que de estas propiedades se deduce que $\langle x + y, z \rangle = \langle x, z \rangle + \langle y, z \rangle$, $\langle  x, \alpha y \rangle = \alpha \langle x, y \rangle$ y $\langle  x, 0 \rangle = 0$.


Este producto interno nos provee la noción de distancia Euclídea. Diremos que la norma (o longitud) del vector $x$ es el número
$$
\| x \| = \sqrt{\langle x,x \rangle} = \sqrt{x_1^2 + x_2^2 + \ldots + x_n^2}.
$$
Note que para $x\in \mathbb{R}^1$, $\| x \| = |x|$. Diremos que $x$ es un vector es unitario si $\| x\| = 1$. Con esto, todo vector puede escribirse como un producto entre su magnitud y su dirección
$$
x = \| x \| \frac{x}{\|x\|},
$$
donde la dirección $x/\|x\|$ es un vector unitario. Pues, si $v=x/\|x\|$, entonces
$$
\|v\|^2 = \langle v, v \rangle = \frac{1}{\|x\|^2} \langle x, x\rangle = 1
$$


---
#### Ejemplo.

Una velocidad se compone de una dirección y una magnitud, por ello puede expresarse como un vector. La magnitud de una velocidad es llamada rapidez.

Suponga que un avión vuela en dirección N$60^\circ$E ($60$ grados hacia el Este a partir de la dirección Norte) a $250$ km/h y que el viento sopla en dirección N$45^\circ$O a $50$ km/h. El curso real del avión se obtiene tomando en cuenta las velocidades del avión y del viento. Vectorialmente la velocidad del avión es el vector
$$
v = 250 \left(\operatorname{sen}\left(\frac{\pi}{3}\right), \cos\left(\frac{\pi}{3}\right) \right)
 = 250 \left( \frac{\sqrt{3}}{2}, \frac{1}{2} \right)
 \approx ( 216.51, 125),
$$
y la velocidad del viento es
$$
w = 50 \left(-\operatorname{sen}\left(\frac{\pi}{4}\right), \cos\left(\frac{\pi}{4}\right) \right)
 = 50 \left( -\frac{\sqrt{2}}{2}, \frac{\sqrt{2}}{2} \right)
 \approx ( -35.36, 35.36).
$$
Por lo tanto la velocidad real del avión es
$$
v + w = \bigl( 125\sqrt{3} - 25\sqrt{2}, 125 + 25\sqrt{2} \bigr) \approx (181.15, 160.36).
$$

Dos vectores $x, y \in \mathbb{R}^n$ son ortogonales, denotado por $x \perp y$, si $\langle x, y \rangle =0$. Por ejemplo, $\langle e^i, e^j \rangle =0$ si $i\neq j$.
Un ejemplo menos trivial de ortogonalidad es el siguiente.


>**Proposicion**
>
>Si $x\in\mathbb{R}^n$ es no nulo, entonces para todo $y\in\mathbb{R}^n$ el vector $z= y - \frac{\langle x, y \rangle}{\| x \|^2} x$ es ortogonal a $x$.

**Demostración:**
El producto interno entre $z$ y $x$ es
\begin{align*}
\langle z, x \rangle &= \left\langle y - \frac{\langle x, y \rangle}{\| x \|^2} x , x\right\rangle \\
        &= \langle y , x \rangle - \frac{\langle x, y \rangle}{\| x \|^2} \langle x, x \rangle \\
        &= \langle y , x \rangle - \langle x , y \rangle = 0. \qquad\spadesuit
\end{align*}



Del resultado anterior, tenemos que $y = \frac{\langle x, y \rangle}{\| x \|^2} x + z$, o sea que dado $x\in\mathbb{R}^n$ no nulo, todo vector $y\in\mathbb{R}^n$ se escribe como la suma de un múltiplo de $x$ con un vector ortogonal a $x$. Esto puede apreciarse en la Figura de abajo. Además, esta descomposición es única, pues si $y = \alpha x + z$ con $z \perp x$, entonces $\langle y, x\rangle = \alpha \langle x, x \rangle$. Por lo tanto $\alpha = \langle y, x\rangle / \| x\|^2$. De aquí concluímos que el vector
$$
\frac{\langle x, y \rangle}{\| x \|^2} x,
$$
es la proyección ortogonal de $y$ sobre la recta que contiene a $x$.
Además, note que
$$
\frac{\langle x, y \rangle}{\| x \|^2} x = \frac{\langle x, y \rangle}{\| x \|} \frac{x}{\|x\|},
$$
de esta manera, decimos que $\frac{\langle x, y \rangle}{\| x \|}$ es la componente (coordenada) de $y$ en la dirección de $x$.

Note que los vectores canónicos $e_i$ son los generadores de los ejes y, por lo tanto, si $x\in\mathbb{R}^n$ entonces $x_i$ es la proyección de $x$ sobre el eje generado por $e_i$.


```
#@title Proyección { display-mode: "form" }

import numpy as np
import matplotlib.pyplot as plt
u = 5
x = u*np.array([0.5, 0.1])
y = u*np.array([-0.2, 0.3])
w = np.inner(x, y)/np.inner(x, x) * x
z = y - w
# x
plt.plot([0, x[0]], [0, x[1]], 'k')
plt.arrow(0, 0, x[0], x[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='k')
plt.text(x[0], x[1],'$x$',color='k',fontsize=14, verticalalignment='top',horizontalalignment='right')
# y
plt.plot([0, y[0]], [0, y[1]], 'k')
plt.arrow(0, 0, y[0], y[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='k')
plt.text(y[0], y[1],'$y$',color='k',fontsize=14, verticalalignment='top',horizontalalignment='left')
# w
plt.plot([0, w[0]], [0, w[1]], 'g')
plt.plot([y[0], w[0]], [y[1], w[1]], 'g:')
plt.arrow(0, 0, w[0], w[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='g')
plt.text(w[0], w[1],r'$\frac{\langle x, y \rangle}{|| x ||^2} x$',color='g',fontsize=14, verticalalignment='top',horizontalalignment='left')
# z
plt.plot([0, z[0]], [0, z[1]], 'b')
plt.arrow(0, 0, z[0], z[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='b')
plt.text(z[0], z[1],r'$y - \frac{\langle x, y \rangle}{|| x ||^2} x$',color='b',fontsize=14, verticalalignment='bottom',horizontalalignment='left')

xx = x/np.linalg.norm(x)
yy = y/np.linalg.norm(y)
zz = z/np.linalg.norm(z)
uu = 0.3
rr = 0.6
th1 = np.arccos(xx[0])
th2 = np.arccos(yy[0])
tt = np.linspace(th1, th2, 200)
plt.plot(rr*np.cos(tt), rr*np.sin(tt), 'g')
plt.text(rr*np.cos(tt[100]), rr*np.sin(tt[100]),r'$\theta$',color='g',fontsize=14, verticalalignment='top',horizontalalignment='right')
plt.plot([uu*xx[0], uu*(xx[0]+zz[0]), uu*zz[0]], [uu*xx[1], uu*(xx[1]+zz[1]), uu*zz[1]], 'b')
plt.axis('equal')
plt.axis('off');
#plt.savefig('proyeccion.pdf')
```

>**Teorema (Pitágoras)**
>
>Si $x \perp y$ entonces $\| x+y\|^2 = \|x\|^2 + \|y\|^2$.

**Demostración:**
De la definición de norma, tenemos que
\begin{align*}
    \| x+y \|^2 &= \langle x+y, x+y\rangle \\
    &= \langle x, x\rangle + \langle x, y\rangle + \langle y, x\rangle + \langle y, y\rangle \\
    &= \|x\|^2 + 2 \langle x, y\rangle +  \| y\|^2 \\
    &= \|x\|^2 + \|y\|^2. \quad \spadesuit
\end{align*}



Por ejemplo, considere $x=(x_1, x_2, x_3)$ como suma de $w=(x_1, x_2, 0)$ y $z=(0,0,x_3)$, luego
$$
    \left(\sqrt{x_1^2 + x_2^2 + x_3^2}\right)^2 = \left(\sqrt{x_1^2 + x_2^2}\right)^2 + x_3^2,
$$


```
#@title Pitágoras { run: "auto", form-width: "30%", display-mode: "form" }
x1 = 1 #@param {type:"slider", min:-10, max:10, step:1}
x2 = -2 #@param {type:"slider", min:-10, max:10, step:1}
x3 = 3 #@param {type:"slider", min:-10, max:10, step:1}

import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits import mplot3d
plt.figure(figsize=(12,8))
ax = plt.axes(projection='3d')
x = np.array([x1, x2, x3])
# ejes
vmin = np.minimum(-1, np.min(x))
vmax = np.maximum(1, np.max(x))
ax.plot3D([vmin,vmax], [0,0], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([0,0], [vmin,vmax], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([0,0], [0,0], [vmin,vmax], color=(0.5,0.5,0.5))
# x
ax.quiver(0, 0, 0, x[0], x[1], x[2], color='k', arrow_length_ratio=0.1)
ax.plot3D([0,x[0]], [0,x[1]], [0,x[2]], 'k')
ax.text3D(x[0], x[1], x[2],f'$x=({x[0]},{x[1]},{x[2]})$')
#
ax.plot3D([x[0],x[0]], [0,x[1]], [0,0], 'b:')
ax.text3D(x[0], 0.5*x[1], 0,'$x_2$', color='b', verticalalignment='top',horizontalalignment='left')
ax.plot3D([0,x[0]], [x[1],x[1]], [0,0], 'b:')
ax.text3D(0.5*x[0], x[1], 0,'$x_1$', color='b', verticalalignment='top',horizontalalignment='left')
ax.plot3D([0,x[0]], [0,x[1]], [0,0], 'k:')
ax.text3D(0.5*x[0], 0.5*x[1], 0,r'$\sqrt{x_1^2 + x_2^2}$', color='k', verticalalignment='center',horizontalalignment='center')
ax.plot3D([x[0],x[0]], [x[1],x[1]], [0,x[2]], 'b:')
ax.text3D(x[0], x[1], 0.5*x[2],'$x_3$', color='b')
ax.view_init(30, 30)
ax.axis('off');
```

Cuando los vectores no son perpendiculares, la longitud de la suma de vectores se verá afectada por el ángulo entre los vectores. Para este desarrollo será de mucha utilidad la siguiente relación.


>**Teorema (Desigualdad de Cauchy-Schwarz)**
>
>Para cualquier $x,y\in\mathbb{R}^n$ se tiene que
\begin{equation*}
| \langle x, y \rangle | \leq \|x\| \|y\|,     
\end{equation*}
donde la igualdad vale si y sólo si un vector es múltiplo del otro.

**Demostración:**
La desigualdad es obvia si $x=0$. Suponiendo que $x\neq 0$ podemos escribir $ y = \alpha x + z$ con $z \perp x$ y $\alpha = \langle y, x\rangle / \| x\|^2$. Usando el Teorema de Pitágoras obtenemos
$$
\| y\|^2 = \alpha^2 \|x\|^2 + \| z\|^2 \geq \alpha^2 \|x\|^2,
$$
donde la igualdad vale si y sólo si $y = \alpha x$. Utilizando la expresión para $\alpha$ obtenemos que $\|y\|^2 \geq \langle y, x\rangle^2 / \| x\|^2 $, o sea $\langle y, x\rangle^2 \leq \| x\|^2 \| y\|^2$, que es equivalente a $| \langle y, x\rangle | \leq \| x\| \| y\|$ y donde la igualdad vale si y sólo si $y = \alpha x$. $\spadesuit$



La desigualdad de Cauchy-Schwarz nos garantiza que $\frac{\langle x, y \rangle}{\|x\| \|y\|} = \left\langle \frac{x}{\|x\|}, \frac{y}{\|y\|} \right\rangle$ es un número en el intervalo $[-1,1]$. Veamos
que este número coincide con el coseno del ángulo formado entre $x$ e $y$.


>**Corolario**
>
>Si $\theta\in [0, \pi]$ es el ángulo entre los vectores $x$ e $y$, entonces
$$
\langle x, y \rangle = \|x\| \|y\| \cos(\theta).
$$

**Demostración:**
Como puede apreciarse en la [Figura de la proyección](#scrollTo=Ou5RTIiWTWMO), el valor $\|y\| \cos(\theta)$ es igual a la longitud de la proyección ortogonal de $y$ sobre la recta que contiene a $x$, esto es
$$
\|y\| \cos(\theta) = \frac{\langle x, y \rangle}{\| x \|}. \qquad \spadesuit
$$



Veamos una simple aplicación de este resultado.

---

#### Ejemplo

Veamos que todo ángulo inscrito en un semicírculo es un ángulo recto. Para ello, considere $x$ e $y$ en la circunferencia y el triángulo de vértices $x$, $-x$ e $y$ como se muestra en el gráfico.


```
#@title Ángulo inscrito { form-width: "30%", display-mode: "form" }
t = 151 #@param {type:"slider", min:0, max:180, step:1}

import matplotlib.pyplot as plt
import numpy as np

th = np.linspace(0, np.pi, 181)
cc = np.cos(th)
cs = np.sin(th)
x = np.array([1, 0])
y = np.array([cc[t], cs[t]])
plt.plot(cc, cs, 'k')
plt.plot([x[0], -x[0]], [x[1], -x[1]], 'g')
plt.plot([0, y[0]], [0, y[1]], 'g')
plt.plot([-x[0], y[0], x[0]], [-x[1], y[1], x[1]], 'b')
plt.plot([x[0], y[0], -x[0], 0], [x[1], y[1], -x[1], 0], 'k.')
plt.text(x[0], x[1], '$x$', verticalalignment='top', horizontalalignment='left', fontsize=14)
plt.text(-x[0], -x[1], '$-x$', verticalalignment='top', horizontalalignment='right', fontsize=14)
plt.text(y[0], y[1], '$y$', verticalalignment='bottom', horizontalalignment='center', fontsize=14)
plt.text(0.5*(y[0]-x[0]), 0.5*(y[1]-x[1]), '$-x-y$', color='b', verticalalignment='top', horizontalalignment='left', fontsize=14)
plt.text(0.5*(y[0]+x[0]), 0.5*(y[1]+x[1]), '$x-y$', color='b', verticalalignment='top', horizontalalignment='right', fontsize=14)
plt.axis('equal')
plt.axis('off');
#plt.savefig('ejm_anguloinscrito.pdf')
```

Como $\|x\| = \|y\|$,
$$
    \langle -x-y, x-y \rangle = (-1) \langle x + y, x-y \rangle
                            = (-1) \bigl( \|x\|^2 - \|y\|^2 \bigr) = 0.
$$
Entonces el ángulo entre $-x-y$ y $x-y$ es $\pi/2$.

---

Del Corolario anterior, podemos afirmar que el producto interno entre dos vectores es positivo si y sólo si el ángulo que los separa es agudo, o sea $\langle x, y \rangle > 0$ si y sólo si $\theta \in [0, \frac{\pi}{2})$. De manera análoga, el producto interno entre dos vectores es negativo si y sólo si el ángulo que los separa es obtuso, o sea $\langle x, y \rangle < 0$ si y sólo si $\theta \in (\frac{\pi}{2}, \pi]$.

Con esto, cada vector $w\in\mathbb{R}^n$ nos separa el espacio en dos semiespacios,
\begin{gather*}
    \mathcal{H}^+_{w,0} = \{ x \in\mathbb{R}^n \mid \langle w, x \rangle \geq 0 \}, \\
    \mathcal{H}^-_{w.0} = \{ x \in\mathbb{R}^n \mid \langle w, x \rangle \leq 0 \},
\end{gather*}
cuya intersección es el hiperplano de vectores ortogonales a $w$,
$$
\mathcal{H}_{w,0} = \{ x \in\mathbb{R}^n \mid \langle w, x \rangle = 0 \} = w^\perp.
$$


Dado $w\in\mathbb{R}^n \setminus \{0\}$ y $\bar{x}\in\mathbb{R}^n$, el semiespacio
$$
    \mathcal{H}^{-}_{w,0}(\bar{x}) = \{ x \in\mathbb{R}^n \mid \langle w, x - \bar{x}\rangle \leq 0 \},
$$
es simplemente tomar $\mathcal{H}^{-}_{w,0}$ y desplazar el origen a $\bar{x}$.

Para ver esto, note que $x\in \mathcal{H}^{-}_{w,0}(\bar{x})$ si y sólo si $\langle w, x - \bar{x}\rangle \leq 0$, si y sólo si $x-\bar{x} \in \mathcal{H}^-_{w,0}$, si y sólo si $x\in \{\bar{x} + v \mid v\in \mathcal{H}^-_{w,0}\}$. Lo que es más, si $\tilde{x}$ es cualquier elemento en el hiperplano paralelo a $\mathcal{H}_{w,0}$ que pasa por $\bar{x}$, o sea $\tilde{x} = \bar{x} + u$ con $u\in \mathcal{H}_{w,0}$, tendremos que $\mathcal{H}^{-}_{w,0}(\bar{x}) = \mathcal{H}^{-}_{w,0}(\tilde{x})$ ya que $\langle w, \tilde{x} \rangle = \langle w, \bar{x} \rangle + \langle w, u \rangle = \langle w, \bar{x} \rangle$.

Por otro lado, note que $\langle w, x - \bar{x}\rangle \leq 0$ si y sólo si
$$
\frac{\langle w, x \rangle}{\|w\|} \leq \frac{\langle w, \bar{x}\rangle}{\|w\|} = \frac{\langle w, \tilde{x}\rangle}{\|w\|},
$$
o sea, los elementos de $\mathcal{H}^{-}_{w,0}(\bar{x})$ son aquellos cuya componente a lo largo de $w$ es menor o igual a $\langle w, \bar{x}\rangle / \|w\|$, donde $\bar{x}$ puede reemplazarse por cualquier punto como $\tilde{x}$. Pudiendo obtener así que para $b=\langle w, \bar{x}\rangle$, el semiespacio es independiente de $\bar{x}$ y luego todo semiespacio desplazado del origen se define como
$$
    \mathcal{H}^{-}_{w,b} = \{ x \in\mathbb{R}^n \mid \langle w, x \rangle \leq b \}.
$$
De manera análoga se definen $\mathcal{H}^{+}_{w,b}$ y $\mathcal{H}_{w,b}$.


```
#@title Semiplano { display-mode: "form" }
import numpy as np
import matplotlib.pyplot as plt

w = 0.3*np.array([5, 8])
b = 3
wp = np.array([-w[1], w[0]])
wb = b/np.inner(w, w)*w
vi = wb + wp
vd = wb - wp
# w
plt.plot([0, w[0]], [0, w[1]], 'k')
plt.arrow(0, 0, w[0], w[1], head_width=0.1, head_length=0.2, length_includes_head=True, color='k')
plt.text(w[0], w[1],'$w$',color='k',fontsize=14, verticalalignment='top',horizontalalignment='left')
plt.plot([vi[0],vd[0]], [vi[1],vd[1]],'g:')
plt.text(wb[0]-0.2*wp[0], wb[1]-0.2*wp[1],r'$\frac{b}{||w||}$',color='k',fontsize=14, verticalalignment='bottom',horizontalalignment='left')

for i in range(500):
  x = 1.5*np.random.randn(2)
  tw = np.minimum(b,np.inner(w, x))/np.inner(w, w)
  twp = np.inner(wp, x)/np.inner(wp, wp)
  Px = tw*w + twp*wp
  plt.plot(Px[0], Px[1], 'g')
  plt.arrow(0, 0, Px[0], Px[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='g')

plt.axis('equal')
plt.axis('off');
plt.savefig('semiplano.pdf')
```
