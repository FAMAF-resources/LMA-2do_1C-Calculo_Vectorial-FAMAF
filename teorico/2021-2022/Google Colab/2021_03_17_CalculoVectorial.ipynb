# Geometría del espacio Euclídeo

## Interpretando el espacio Vectorial

En estas notas se usará $\mathbb{R}$, $\mathbb{Z}$ y $\mathbb{N}$ para denotar el conjunto de números reales, enteros y naturales, respectivamente.

Sea $n$ un número natural. El espacio Euclídeo $n$-dimensional $\mathbb{R}^n$ es el producto cartesiano de $n$ factores iguales a $\mathbb{R}$, o sea, $\mathbb{R}^n = \mathbb{R} \times \mathbb{R} \times \ldots \times \mathbb{R}$.
Los elementos de $\mathbb{R}^n$ serán llamados puntos o vectores y un elemento $x\in \mathbb{R}^n$ será de la forma $x = (x_1, x_2, \ldots, x_n)$ donde $x_i\in\mathbb{R}$ es la $i$-ésima coordenada de $x$.

De esta manera, $\mathbb{R}^1 = \mathbb{R}$ es el conjunto de números reales, $\mathbb{R}^2$ es el modelo numérico del plano y $\mathbb{R}^3$ es el modelo numérico del espacio tridimensional.

En la literatura suele denotarse a los puntos en $\mathbb{R}^2$ por $(x,y)$ y a los puntos en $\mathbb{R}^3$ por $(x,y,z)$, en vez de $(x_1, x_2)$ y $(x_1, x_2, x_3)$, respectivamente. Ambas notaciones se utilizarán en estas notas dependiendo del contexto.


Para dos elementos $x = (x_1, x_2, \ldots, x_n)$ e $y = (y_1, y_2, \ldots, y_n)$ en $\mathbb{R}^n$, la suma se define como
$$
x + y = (x_1 + y_1, x_2 + y_2, \ldots, x_n + y_n).
$$

Veamos la interpretación gráfica de la suma en $\mathbb{R}^2$.


---
#### Ejemplo.
Considere los vectores $x=(-0.2, 0.3)$, $y=(0.6, 0.2)$ y su suma $x+y = (0.4, 0.5)$. Los mismos pueden representarse gráficamente de la siguiente manera.


```
#@title Ley del paralelogramo 2D { run: "auto", form-width: "30%", display-mode: "form" }
x1 = -0.2 #@param {type:"slider", min:-2, max:2, step:0.1}
x2 = 0.3 #@param {type:"slider", min:-2, max:2, step:0.1}
y1 = 0.4 #@param {type:"slider", min:-2, max:2, step:0.1}
y2 = 0.1 #@param {type:"slider", min:-2, max:2, step:0.1}
import matplotlib.pyplot as plt
import numpy as np

x = np.array([x1, x2])
y = np.array([y1, y2])
z = x + y

plt.figure()
# ejes
plt.axhline(color=(.5,.5,.5))
plt.axvline(color=(.5,.5,.5))
# x
plt.plot([0, x[0]], [0, x[1]], 'k')
plt.arrow(0, 0, x[0], x[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='k')
plt.text(x[0], x[1],f'$x=({x[0]:.1f},{x[1]:.1f})$',color='k',fontsize=12)
# y
plt.plot([0, y[0]], [0, y[1]], 'k')
plt.arrow(0, 0, y[0], y[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='k')
plt.text(y[0], y[1],f'$y=({y[0]:.1f},{y[1]:.1f})$',color='k',fontsize=12)
# x+y
plt.plot([0, z[0]], [0, z[1]], 'b')
plt.arrow(0, 0, z[0], z[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='b')
plt.text(z[0], z[1],f'$x+y=({z[0]:.1f},{z[1]:.1f})$',color='b',fontsize=12)
# desplazados
plt.plot([x[0], z[0]], [x[1], z[1]], 'k:')
plt.plot([y[0], z[0]], [y[1], z[1]], 'k:')
#plt.plot([-0.18, 0.71], [0.1, 0.51], 'w.')
# escala y grilla
plt.axis('equal')
plt.grid();
#plt.savefig('paralelogramo2D.pdf')
```

Como puede verse en en esta figura, el vector suma puede obtenerse gráficamente desplazando un vector sobre el otro. Obteniendo que el vector $x+y$ es la diagonal del paralelogramo en la dirección de los vectores $x$ e $y$.

Para cualquier $\alpha \in \mathbb{R}$ y $x \in \mathbb{R}^n$, la multiplicación escalar (''escalar'' se utilizará como sinónimo de ''número real'') se define como
$$
\alpha x = (\alpha x_1, \alpha x_2, \ldots, \alpha x_n).
$$

Escribiremos $-x$ para el producto $(-1) x$ y $x - y$ como una abreviación de $x + (-y)$.
Para la interpretación gráfica de la resta de vectores, como $x = (x-y) + y$, obtenemos que $x-y$ es un vector que al desplazarlo desde el origen hasta $y$ debería coincidir con $x$.

```
#@title Resta
import matplotlib.pyplot as plt
import numpy as np

x = np.array([-0.2, 0.3])
y = np.array([0.6, 0.2])
# x
plt.plot([0, x[0]], [0, x[1]], 'k')
plt.arrow(0, 0, x[0], x[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='k')
plt.text(x[0], x[1],'$x$',color='k',fontsize=14, verticalalignment='top',horizontalalignment='right')
# y
plt.plot([0, y[0]], [0, y[1]], 'k')
plt.arrow(0, 0, y[0], y[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='k')
plt.text(y[0], y[1],'$y$',color='k',fontsize=14, verticalalignment='top',horizontalalignment='left')
#plt.plot(0.4, 0.5, 'w') # suma
# x-y
plt.plot([0, x[0]-y[0]], [0, x[1]-y[1]], 'b')
plt.arrow(0, 0, x[0]-y[0], x[1]-y[1], head_width=0.05, head_length=0.1, length_includes_head=True, color='b')
plt.text(x[0]-y[0], x[1]-y[1],'$x-y$',color='b',fontsize=14, verticalalignment='bottom',horizontalalignment='right')
#
plt.plot([x[0]-y[0], x[0]], [x[1]-y[1], x[1]], 'k:')
plt.plot([y[0], x[0]], [y[1], x[1]], 'k:')
plt.axis('equal')
plt.axis('off');
#plt.savefig('resta_vec.pdf')
```

Como puede apreciarse gráficamente, el vector $x-y$ desplazado hasta $y$ es justamente la otra diagonal del paralelogramo formado al realizar la suma de $x$ e $y$.


Usaremos $0$ para denotar el vector $n$-dimensional $(0, 0, \ldots, 0)$, o sea, el cero (u origen) de $\mathbb{R}^n$. Por más que esta notación pueda resultar ambigua, ya que $0$ puede ser $(0,0)$ o $(0,0,0,0)$, la dimensión del vector $0$ quedará claro por el contexto y sólo una interpretación tendrá sentido.



De lo anterior, se tiene que para $x, y, z \in \mathbb{R}^n$ y $\alpha, \beta \in  \mathbb{R}$ valen
1. $\alpha x + \beta x = (\alpha + \beta) x$,
2. $\alpha x + \alpha y = \alpha (x + y)$,
3. $\alpha (\beta x) = (\alpha \beta) x$,
4. $x + y = y + x$,
5. $(x + y) + z = x + (y + z)$,
6. $x + 0 = x$,
7. $-x + x = 0$.



Las propiedades anteriores muestran que $\mathbb{R}^n$ es un espacio vectorial real. Por lo tanto un elemento $x\in\mathbb{R}^n$ es un vector. Los vectores $e^1=(1, 0, \ldots, 0)$, $e^2=(0, 1, \ldots, 0)$, ... , $e^n=(0, 0, \ldots, 1)$ tienen la propiedad que si $x \in\mathbb{R}^n$ entonces
$$
x = x_1 e^1 + x_2 e^2 + \ldots + x_n e^n.
$$
Por este motivo, el conjunto $\{e^1, e^2, \ldots, e^n \}$ es llamado la base canónica de $\mathbb{R}^n$. Note que utilizaremos supraíndices para numerar vectores y reservaremos los subíndices para numerar las coordenadas. Esta notación no producirá ambigüedades pues la potencia de un vector no puede definirse con las operaciones que se verán en este curso. Cuando se trabaja en $\mathbb{R}^3$ algunos autores denotan a los vectores canónicos $e^1$, $e^2$ y $e^3$ por $\mathbf{i}$, $\mathbf{j}$ y $\mathbf{k}$, respectivamente.


La interpretación gráfica es fundamental para afianzar la comprensión de las operaciones vistas.
Por ejemplo, dado $v\in\mathbb{R}^n\setminus\{0\}$, el conjunto
$$
\mathcal{R}_0 = \{ \alpha v \in \mathbb{R}^n \mid \alpha \in \mathbb{R}\},
$$
es la recta generada por $v$ que pasa por el origen. Ahora, dado $x\in\mathbb{R}^n$, el conjunto
$$
\mathcal{R} = \{ x+ \alpha v \in \mathbb{R}^n \mid \alpha \in \mathbb{R}\} = x + \mathcal{R}_0,
$$
es la recta paralela a $\mathcal{R}_0$ que pasa por $x$. Podemos apreciar estos conjuntos en $\mathbb{R}^3$.


```
#@title Rectas paralelas { run: "auto", form-width: "30%", display-mode: "form" }
x1 = 0.5 #@param {type:"slider", min:-2, max:2, step:0.1}
x2 = 0.5 #@param {type:"slider", min:-2, max:2, step:0.1}
x3 = 2 #@param {type:"slider", min:-2, max:2, step:0.1}

import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits import mplot3d

ax = plt.axes(projection='3d')
v = np.array([1., 1, 1])
x = np.array([x1, x2, x3])
tmin = -3
tmax = 3
smin = -3-np.min(x)
smax = 3-np.max(x)

# ejes
ax.plot3D([np.minimum(tmin*v[0],tmax*v[0]), np.maximum(tmin*v[0],tmax*v[0])], [0,0], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([0,0], [np.minimum(tmin*v[1],tmax*v[1]), np.maximum(tmin*v[1],tmax*v[1])], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([0,0], [0,0], [np.minimum(tmin*v[2],tmax*v[2]), np.maximum(tmin*v[2],tmax*v[2])], color=(0.5,0.5,0.5))
# v
ax.quiver(0, 0, 0, v[0], v[1], v[2], color='k', arrow_length_ratio=0.1)
ax.text3D(v[0], v[1], v[2],f'$v=({v[0]:.0f},{v[1]:.0f},{v[2]:.0f})$', verticalalignment='top', horizontalalignment='left')
ax.plot3D([tmin*v[0], tmax*v[0]], [tmin*v[1], tmax*v[1]], [tmin*v[2], tmax*v[2]], 'b')
ax.text3D(tmin*v[0], tmin*v[1], tmin*v[2],r'$\mathcal{R}_0$', color='b', fontsize=18, verticalalignment='top', horizontalalignment='left')
# x
ax.quiver(0, 0, 0, x[0], x[1], x[2], color='k', arrow_length_ratio=0.1)
ax.text3D(x[0], x[1], x[2],f'$x=({x[0]:.1f},{x[1]:.1f},{x[2]:.1f})$', verticalalignment='top', horizontalalignment='left')
ax.plot3D([x[0]+smin*v[0], x[0]+smax*v[0]], [x[1]+smin*v[1], x[1]+smax*v[1]], [x[2]+smin*v[2], x[2]+smax*v[2]], 'g')
ax.text3D(x[0]+smin*v[0], x[1]+smin*v[1], x[2]+smin*v[2],r'$\mathcal{R}$', color='g', fontsize=18, verticalalignment='top', horizontalalignment='left')

ax.view_init(20,-80);
plt.savefig('recta.pdf')
```

#### Ejemplo
Una golondrina volando en trayectoria recta pasa por la posición $(5,0,30)$, medida en metros, y luego de $5$ segundos por la posición $(26, -42, 2.5)$.
1. ¿Cuál es la velocidad de la golondrina?
2. Si agregamos el hecho de que en ese momento el viento soplaba con una velocidad $(1,-1,0)$ metros por segundo, ¿cuál es la velocidad de la golondrina en este caso?


Si denotamos por $x$ a la primera posición e $y$ a la segunda, obtenemos que la recta que describe el movimiento de la golondrina es
$$
x + \frac{t}{5}(y-x),
$$
donde $t$ es el tiempo en segundos. Note que para $t=0$ se obtiene el punto $x$ y para $t=5$ se obtiene el punto $y$. Por lo tanto la velocidad de la golondrina es
$$
 v = \frac{1}{5} (y - x) = (21, -42, -27.5)/5 = (4.2, -8.4, -5.5).
$$

Si adicionamos la velocidad del viento $w$, la recta del movimiento de la golondrina sería
$$
x + t (\tilde{v} + w ),
$$
y como en $t=5$ debe pasar por el punto $y$, se obtiene que
$$
\tilde{v} = \frac{1}{5}(y-x) - w = (4.2, -8.4, -5.5) - (1, -1, 0) = (3.2, -7.4, -5.5).
$$

#### Ejemplo

Suponga que se tiene el siguiente esquema de flujos (agua, vehículos, información, etc.): dos flujos entrantes de 10 unidades, dos flujos salientes de 15 y 5 unidades, y conexiones entre nodos que supondremos con flujo positivo al moverse en sentido antihorario.


```
#@title Flujos { display-mode: "form" }
import numpy as np
import matplotlib.pyplot as plt

th = np.arange(0, 2*np.pi, np.pi/100)
circulo = np.array([np.cos(th), np.sin(th)])
plt.plot(circulo[0], circulo[1], 'k')
for i in range(4):
  plt.plot(circulo[0,i*50],circulo[1,i*50],'co',markersize=30)
  plt.text(circulo[0,i*50],circulo[1,i*50],f'{i+1}',fontsize=18,verticalalignment='center',horizontalalignment='center')
  plt.arrow(circulo[0,i*50-15],circulo[1,i*50-15],circulo[0,i*50-8]-circulo[0,i*50-15],circulo[1,i*50-8]-circulo[1,i*50-15],length_includes_head=True,head_width=0.08, head_length=0.2,color='k')
  plt.text(1.2*circulo[0,i*50-15],1.2*circulo[1,i*50-15],f'$f_{i+1}$',fontsize=18,verticalalignment='center',horizontalalignment='center')

tmp = 0.3
plt.arrow(-1.5,0,tmp,0,length_includes_head=True,head_width=0.08, head_length=0.2,color='c')
plt.text(-1.5,-0.1,'10',fontsize=18,verticalalignment='top',horizontalalignment='center', color='c')
plt.arrow(1.5-tmp,0,tmp,0,length_includes_head=True,head_width=0.08, head_length=0.2,color='c')
plt.text(1.5,-0.1,'15',fontsize=18,verticalalignment='top',horizontalalignment='center', color='c')
plt.arrow(0,-1.5,0,tmp,length_includes_head=True,head_width=0.08, head_length=0.2,color='c')
plt.text(0.1,-1.5,'10',fontsize=18,verticalalignment='bottom',horizontalalignment='left', color='c')
plt.arrow(0,1.5-tmp,0,tmp,length_includes_head=True,head_width=0.08, head_length=0.2,color='c')
plt.text(-0.1,1.5,'5',fontsize=18,verticalalignment='top',horizontalalignment='right', color='c')
plt.plot([-1.5,1.5], [-1.5,1.5],'w.')

plt.axis('equal')
plt.axis('off');
#plt.savefig('ejm_flujo.pdf')
```

Las ecuaciones que surgen de igualar cantidades en cada nodo son
\begin{align*}
    f_1 &= 15 + f_2 && \Rightarrow & f_1 - f_2 &= 15,\\
    f_2 &= 5 + f_3 && \Rightarrow &  f_2 - f_3 &= 5,\\
    f_3 + 10 &= f_4 && \Rightarrow &    f_3 - f_4 &= -10,\\
    f_4 + 10 &= f_1 && \Rightarrow & f_4 - f_1 &= -10.
\end{align*}
Note que un flujo admisible es $a = (20, 5, 0,10)$. Si tenemos otro flujo admisible $x$ tenemos que $v = x-a$ satisface las ecuaciones $v_1 - v_2 = 0$, $v_2 - v_3 =0$, $v_3 - v_4 =0$ y $v_4 - v_1 =0$. De esto se obtiene que $v_2 = v_1$, $v_4=v_1$ y $v_3 = v_2 = v_1$, $v_4 = v_3 = v_1$. Por lo tanto $v = t(1,1,1,1)$ para cualquier $t$ real. Entonces todo flujo admisible está en la recta
$$
a + t (1,1,1,1),
$$
para $t\in \mathbb{R}$. Dependiendo del problema a resolver se deberá hallar uno o algunos elementos de este conjunto.

Por otro lado, si uno tiene $v^1, v^2 \in\mathbb{R}^n\setminus\{0\}$ linealmente independientes, el conjunto
$$
\mathcal{P}_0 = \{\alpha_1 v^1 + \alpha_2 v^2 \mid \alpha \in\mathbb{R}^2\},
$$
es el plano generado por $v^1$ y $v^2$ que pasa por el origen. Dado $x\in\mathbb{R}^n$, el conjunto
$$
\mathcal{P} = \{x + \alpha_1 v^1 + \alpha_2 v^2 \mid \alpha \in\mathbb{R}^2\},
$$
es el plano paralelo a $\mathcal{P}_0$ que pasa por $x$. Grafiquemos estos conjuntos en $\mathbb{R}^3$.

```
#@title Planos { form-width: "50%" }
x1 = 3 #@param {type:"slider", min:-10, max:10, step:1}
x2 = -7 #@param {type:"slider", min:-10, max:10, step:1}
x3 = -7 #@param {type:"slider", min:-10, max:10, step:1}

import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits import mplot3d

v1 = np.array([1.,1,1])
v2 = np.array([2., -1, 3])
x = np.array([x1, x2, x3])

alpha, beta = np.meshgrid(np.linspace(-3,3,100), np.linspace(-3,3,100))
ax = plt.axes(projection='3d')
# ejes
ax.plot3D([-5,5], [0,0], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([0,0], [-5,5], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([0,0], [0,0], [-5,5], color=(0.5,0.5,0.5))

ax.plot_surface(alpha*v1[0]+beta*v2[0], alpha*v1[1]+beta*v2[1], alpha*v1[2]+beta*v2[2], cmap='Purples')
ax.text3D(alpha[0,0]*v1[0]+beta[0,0]*v2[0], alpha[0,0]*v1[1]+beta[0,0]*v2[1], alpha[0,0]*v1[2]+beta[0,0]*v2[2],r'$\mathcal{P}_0$', color='purple', fontsize=18, verticalalignment='top', horizontalalignment='left')
ax.plot_surface(x[0]+alpha*v1[0]+beta*v2[0], x[1]+alpha*v1[1]+beta*v2[1], x[2]+alpha*v1[2]+beta*v2[2], cmap='Greens')
ax.text3D(x[0]+alpha[0,0]*v1[0]+beta[0,0]*v2[0], x[1]+alpha[0,0]*v1[1]+beta[0,0]*v2[1], x[2]+alpha[0,0]*v1[2]+beta[0,0]*v2[2],r'$\mathcal{P}$', color='green', fontsize=18, verticalalignment='top', horizontalalignment='left')
# x
ax.quiver(0, 0, 0, x[0], x[1], x[2], color='k', arrow_length_ratio=0.1)
ax.text3D(x[0], x[1], x[2],f'$x=({x[0]:.0f},{x[1]:.0f},{x[2]:.0f})$', verticalalignment='top', horizontalalignment='left')
#ax.view_init(elevacion, acimut);
ax.view_init(20, -40);
plt.savefig('planos.pdf')
```

Cuando tenemos $v^1, v^2, \ldots, v^{n-1} \in\mathbb{R}^n\setminus\{0\}$ linealmente independientes, el conjunto
$$
\mathcal{H}_0 = \{\alpha_1 v^1 + \alpha_2 v^2 + \ldots + \alpha_{n-1} v^{n-1} \mid \alpha \in\mathbb{R}^{n-1}\},
$$
tiene dimensión $n-1$ y se conoce como el hiperplano generado por $v^1, \ldots, v^{n-1}$ que pasa por el origen. Dado $x\in\mathbb{R}^n$, el conjunto
$$
\mathcal{H} = \{x + \alpha_1 v^1 + \alpha_2 v^2 + \ldots + \alpha_{n-1} v^{n-1} \mid \alpha \in\mathbb{R}^{n-1}\},
$$
es el hiperplano paralelo a $\mathcal{H}_0$ que pasa por $x$. Note que con esta definición, una recta es un hiperplano en $\mathbb{R}^2$ y un plano es un hiperplano en $\mathbb{R}^3$.
