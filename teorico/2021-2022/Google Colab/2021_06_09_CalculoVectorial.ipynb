# Cálculo Vectorial


## Teorema de Gauss

Hemos visto que el Teorema de Green relaciona una integral sobre una región en $\mathbb{R}^2$ con una integral sobre la curva que la limita. De manera similar, el Teorema de Stokes relaciona una integral sobre una superficie en $\mathbb{R}^3$ con la integral sobre la curva que la limita. En el siguiente resultado relacionaremos la integral de una región en $\mathbb{R}^3$ con la integral de la superficie que la limita. Para ello la región de integración debe tener ciertas propiedades.






> **Definición**
>
> Diremos que un conjunto $D\subset \mathbb{R}^3$ es una **región simple** si es acotado y su frontera es cortada a lo sumo dos veces, o en un segmento, por cualquier recta paralela a alguno de los ejes.



Cuando tenemos una región simple $D \subset \mathbb{R}^3$, podremos definir funciones $\eta_3, \xi_3 : W_3 \subset \mathbb{R}^2 \to \mathbb{R}$ con $\eta_3 \leq \xi_3$ en $W_3$, $\eta_2, \xi_2 : W_2 \subset \mathbb{R}^2 \to \mathbb{R}$ con $\eta_2 \leq \xi_2$ en $W_2$ y $\eta_1, \xi_1 : W_1 \subset \mathbb{R}^2 \to \mathbb{R}$ con $\eta_1 \leq \xi_1$ en $W_1$ tales que
\begin{align*}
    D &= \{ x\in \mathbb{R}^3 \mid  (x_1,x_2) \in W_3, \, \eta_3(x_1,x_2) \leq x_3 \leq \xi_3(x_1,x_2)\} \\
    &= \{ x\in \mathbb{R}^3 \mid  (x_1,x_3) \in W_2, \, \eta_2(x_1,x_3) \leq x_2 \leq \xi_2(x_1,x_3)\} \\
    &= \{ x\in \mathbb{R}^3 \mid  (x_2,x_3) \in W_1, \, \eta_1(x_2,x_3) \leq x_1 \leq \xi_1(x_2,x_3)\} .
\end{align*}



Cuando una superficie es determinada por la frontera de una región $D \subset \mathbb{R}^3$, diremos que su orientación es positiva cuando los vectores normales a la misma apuntan hacia afuera de $D$, en ese caso tal superficie será denotada por $\partial D$.

>**Teorema**(Gauss)
>
> Sea $D\subset \mathbb{R}^3$ acotado y unión finita de regiones simples, cuya frontera $\partial D$ es una superficie regular a trozos orientada positivamente. Si $F:\mathbb{R}^3 \to \mathbb{R}^3$ es continuamente diferenciable en $\operatorname{cl}(D)$, entonces
 $$
    \int_D \operatorname{div}(F)(x) dx = \int_{\partial D} F(x) dS.
 $$
 Esta ecuación también puede escribirse como
 $$
 \int_D \! \left( \frac{\partial F_1}{\partial x_1}(x) + \frac{\partial F_2}{\partial x_2}(x) +\frac{\partial F_3}{\partial x_3}(x) \right) \! dx_1 dx_2 dx_3
 = \int_{\partial D} \!\! F_1(x) dx_2 dx_3 + F_2(x) dx_3 dx_1 + F_3(x) dx_1 dx_2.
 $$


**Demostración:**

La ecuación se obtiene probando que
\begin{align*}
\int_D  \frac{\partial F_1}{\partial x_1}(x) dx_1 dx_2 dx_3 &= \int_{\partial D} F_1(x) dx_2 dx_3, \\
\int_D  \frac{\partial F_2}{\partial x_2}(x) dx_1 dx_2 dx_3 &= \int_{\partial D} F_2(x) dx_3 dx_1, \\
\int_D  \frac{\partial F_3}{\partial x_3}(x) dx_1 dx_2 dx_3 &= \int_{\partial D} F_3(x) dx_1 dx_3.
\end{align*}
Como las demostraciones son análogas, veremos solamente la segunda ecuación.

Supongamos primero que $D$ es una región simple, luego existe $W_2 \subset \mathbb{R}^2$ y funciones $\eta_2, \xi_2 : W_2 \to \mathbb{R}$ tales que
$$
D = \{ x\in \mathbb{R}^3 \mid  (x_1,x_3) \in W_2, \, \eta_2(x_1,x_3) \leq x_2 \leq \xi_2(x_1,x_3)\}.
$$

Luego, la superficie $\partial D$ puede cubrirse por una cantidad finita de parametrizaciones que serán de la forma
$$
\phi(u) = \bigl( u_1, \eta_2(u_1,u_2), u_2 \bigr), \quad
\psi(u) = \bigl( u_2, \xi_2(u_2,u_1), u_1 \bigr), \quad
\mu(u) = \bigl( \gamma_1(u_1), u_2, \gamma_2(u_1)\bigr),
$$
con dominio $U=W_2$ para $\phi$, $V=\{u \mid (u_2,u_1)\in W_2\}$ para $\psi$, y donde la última corresponde a las parametrizaciones con secciones paralelas al eje $x_2$ para alguna curva $\gamma:\mathbb{R} \to \mathbb{R}^2$.
Note que
\begin{align*}
    \phi'(u)
    &= \left[ \begin{array}{cc}
        1 & 0 \\
        \frac{\partial \eta_2}{\partial x_1} & \frac{\partial \eta_2}{\partial x_3} \\
        0 & 1
        \end{array}\right], &
    \frac{\partial \phi}{\partial u_1} \times \frac{\partial \phi}{\partial u_2}
        &=\left(\frac{\partial \eta_2}{\partial x_1}, -1,  \frac{\partial \eta_2}{\partial x_3} \right), \\
    \psi'(u)
    &= \left[ \begin{array}{cc}
        0 & 1 \\
        \frac{\partial \xi_2}{\partial x_3} & \frac{\partial \xi_2}{\partial x_1} \\
        1 & 0
        \end{array}\right], &
    \frac{\partial \psi}{\partial u_1} \times \frac{\partial \psi}{\partial u_2}
        &=\left(-\frac{\partial \xi_2}{\partial x_1}, 1,  -\frac{\partial \xi_2}{\partial x_3} \right), \\
    \mu'(u)
    &= \left[ \begin{array}{cc}
        \gamma_1' & 0 \\
        0 & 1 \\
        \gamma_2' & 0
        \end{array}\right], &
    \frac{\partial \mu}{\partial u_1} \times \frac{\partial \mu}{\partial u_2}
        &=\left(-\gamma_2', 0, \gamma_1'\right).
\end{align*}
Por lo tanto en cada parte el vector normal es saliente a $D$, escogiendo $\gamma$ apropiado en el último caso.


Por otro lado, si la imagen de $\phi$ en $U$ es la superficie $\mathcal{S}$ entonces
$$
\int_{\mathcal{S}} F_2(x) dx_3 dx_1 = \int_U F_2(\phi(u)) \det\left(\left[ \begin{array}{cc}
     \frac{\partial \phi_3}{\partial u_1} & \frac{\partial \phi_3}{\partial u_2} \\
     \frac{\partial \phi_1}{\partial u_1} & \frac{\partial \phi_1}{\partial u_2}
\end{array}\right]\right) du_1 du_2,
$$
para el cambio de variable $(x_3,x_1) = \bigl( \phi_3(u), \phi_1(u)\bigr)$. El determinante de la matriz Jacobiana del cambio de variables asociado a $\phi$ es  $-1$, para $\psi$ es $1$ y para $\mu$ es $0$. Con esto, separando $\partial D$ en sus piezas, obtenemos que
\begin{align*}
    \int_{\partial D} F_2(x) dx_3 dx_1
    &= \int_{U} F_2(\phi(u))(-1) du_1 du_2 + \int_{V} F_2(\psi(u))1 du_1 du_2 \\
    &= \int_{U} -F_2\bigl( u_1, \eta_2(u_1,u_2), u_2 \bigr) du_1 du_2
    + \int_{V} F_2\bigl( u_2, \xi_2(u_2,u_1), u_1 \bigr) du_1 du_2 \\
    &= \int_{W_2} -F_2\bigl( x_1, \eta_2(x_1,x_3), x_3 \bigr) dx_1 dx_3
    + \int_{W_2} F_2\bigl( x_1, \xi_2(x_1,x_3), x_3 \bigr) dx_1 dx_3 \\
    &= \int_{W_2} \left( \int_{\eta_2(x_1,x_3)}^{\xi_2(x_1,x_3)} \frac{\partial F_2}{\partial x_2}(x_1,x_2,x_3) dx_2 \right) dx_1 dx_3 \\
    &= \int_D \frac{\partial F_2}{\partial x_2}(x) dx_1 dx_2 dx_3,
\end{align*}
donde para transformar $V$ en $W_2$ se usó $T:W_2 \to V$ tal que $T(u_1,u_2) = (u_2,u_1)$.

Obteniendo de manera similas las ecuaciones para $F_1$ y $F_3$, al sumar estas ecuaciones obtenemos
$$
 \int_D \left( \frac{\partial F_1}{\partial x_1} + \frac{\partial F_2}{\partial x_2} +\frac{\partial F_3}{\partial x_3} \right)  dx_1 dx_2 dx_3
 = \int_{\partial D}  F_1 dx_2 dx_3 + F_2 dx_3 dx_1 + F_3 dx_1 dx_2.
 $$

Cuando $D$ es unión de regiones simples, la ecuación se obtiene sumando la anterior para cada región simple. $\spadesuit$





Veamos cómo utilizar este resultado.

#### Ejemplo

Se desea calcular la integral de superficie $\int_{\mathcal{S}} F(x) dS$ donde $F:\mathbb{R}^3 \to \mathbb{R}^3$ tal que $F(x) = \bigl( 2x_1, x_2^2, x_3^2\bigr)$ y $\mathcal{S}$ es la esfera unitaria orientada positivamente.

Por el Teorema de Gauss, tenemos que
$$
    \int_{\mathcal{S}} F(x) dS = \int_B \operatorname{div}(F)(x) dx,
$$
donde $B$ es la bola unitaria. Como $\operatorname{div}(F)(x) = 2+2x_2+2x_3$ y
$$
    \int_B x_2 dx = \int_{T(B)} x_2 dx = \int_B -y_2 dy,
$$
donde $T(y) = -y$. Luego
$$
\int_B \operatorname{div}(F)(x) dx = \int_B 2+2x_2+2x_3 dx = \int_B 2 dx = 2 |B| = \frac{8}{3}\pi,
$$
ya que $\int_B x_2 dx = \int_B x_3 dx = 0$ y $|B| = 4\pi /3$.

---



Usando el Teorema de Gauss podemos dar una interpretación para $\operatorname{div}(F)$. Suponga que $F:\mathbb{R}^3 \to \mathbb{R}^3$ es el campo vectorial de velocidad de un fluído. Sea $\nu(x)$ el vector normal unitario saliente para cada $x\in B_r(\bar{x})$. Usando el Teorema de Gauss y el de punto medio para integrales, tenemos
$$
 \int_{\partial B_r(\bar{x})} \langle F(x), \nu(x) \rangle dS
 = \int_{B_r(\bar{x})} \operatorname{div}(F)(x) dx =  \operatorname{div}(F)(\tilde{x}_r) |B_r(\bar{x})|,
$$
donde $\tilde{x}_r \in B_r(\bar{x})$. De esto se obtiene que
$$
    \operatorname{div}(F)(\bar{x}) = \lim_{r \to 0} \frac{1}{|B_r(\bar{x})|} \int_{\partial B_r(\bar{x})} \langle F(x), \nu(x) \rangle dS.
$$
Así, si $\operatorname{div}(F)(\bar{x})>0$ entonces hay un flujo neto saliendo de $\bar{x}$, en este caso se dice que $\bar{x}$ es una fuente. Si $\operatorname{div}(F)(\bar{x})<0$ hay un flujo neto entrando a $\bar{x}$, y se dice que $\bar{x}$ es un sumidero. Concluyendo que $\operatorname{div}(F)(\bar{x})$ es la tasa de flujo neto hacia el exterior de $\bar{x}$.


# Optimización y ecuaciones no lineales


## Ecuaciones no lineales

Las ecuaciones no lineales surgen de manera natural en muchas aplicaciones, este tipo de problema consiste en hallar $\bar{x}$ tal que
$$
F(\bar{x})=0,
$$
donde $F:\mathbb{R}^n \to \mathbb{R}^m$. Veamos un ejemplo práctico donde aparece este problema.




#### Ejemplo

Un intercambiador de calor a partir de fluidos puede modelarse considerando un fluido caliente y uno frío, como lo indica la siguiente figura.

El fluido caliente ingresa al intercambiador con una temperatura $t_{ce}$ y un flujo $q_c$, saliendo con una temperatura $t_{cs}$. El fluido frío ingresa al intercambiador con una temperatura $t_{fe}$ y un flujo $q_f$, saliendo con una temperatura $t_{fs}$. Considerando una constante de transferencia de calor $\kappa$ y constantes $\alpha_c$, $\alpha_f$ de proporcionalidad del flujo del fluido caliente y frío, se obtiene el siguiente sistema de ecuaciones no lineales:
\begin{align*}
    \alpha_c q_c (t_{ce} - t_{cs}) &= \alpha_f q_f (t_{fs} - t_{fe})\\
    \alpha_c q_c (t_{ce} - t_{cs}) &= \kappa \frac{(t_{ce} - t_{fs}) - (t_{cs} - t_{fe})}{\log(t_{ce} - t_{fs}) - \log(t_{cs} - t_{fe})}
\end{align*}


Suponga que tanto $\kappa$ como los flujos y temperaturas de los fluidos entrantes son conocidos. Luego de regular los flujos entrantes escogiendo $\alpha_c$ y $\alpha_f$, se desea establecer las temperaturas de salida $t_{cs}$ y $t_{fs}$.

En este caso, tomando $x=(t_{cs}, t_{fs})$ hay que encontrar un cero de $F:\mathbb{R}^2 \mapsto \mathbb{R}^2$ tal que
\begin{equation*}
    F(x) = \left[\begin{array}{c}
        \alpha_c q_c (t_{ce} - x_1) - \alpha_f q_f (x_2 - t_{fe}) \\[1em]
        \alpha_c q_c (t_{ce} - x_1) - \kappa \frac{(t_{ce} - x_2) - (x_1 - t_{fe})}{\log(t_{ce} - x_2) - \log(x_1 - t_{fe})}
    \end{array} \right]
\end{equation*}

---

Cuando $F$ es una función diferenciable en $\bar{x}$, tenemos que
$$
    F(x) = F(\bar{x}) + F'(\bar{x})(x-\bar{x}) + o(\|x-\bar{x}\|).
$$
Por lo tanto, hallar un cero de la función afín $A:\mathbb{R}^n \to \mathbb{R}^m$ tal que $A(x) = F(\bar{x}) + F'(\bar{x})(x-\bar{x})$, o sea resolver un sistema de ecuaciones lineales, es equivalente a resolver $F(x)=p$ cerca de $\bar{x}$ para una pequeña perturbación $p\in\mathbb{R}^m$.



En la práctica lo más común es resolver un sistema perturbado $F(x)-p=0$, ya que el modelo matemático viene dado por $F$ y la perturbación $p$ corresponde a los errores provenientes de la precisión de los instrumentos de medición o del redondeo numérico. Claramente, lo deseable es que mientras más pequeña sea la perturbación, más próximo se esté de la solución del modelo exacto. Los sistemas que cumplen esa propiedad son llamados estables y es sumamente importante poder determinar la estabilidad o inestabilidad de un sistema.

Puede notar que el análisis del sistema no lineal perturbado está fuertemente relacionado con el análisis de la aproximación afín de la función.




Analizaremos primero el caso para sistemas con igual número de ecuaciones que de incógnitas.
Para ello, será de mucha utilidad contar con un resultado que nos garantice la existencia de un punto fijo.


>**Teorema** (Punto fijo)
>
> Sea $\Phi:\mathbb{R}^n \to \mathbb{R}^n$ y suponga que existen $r>0$, $\hat{x} \in \mathbb{R}^n$ y $0 \leq \kappa < 1$ tales que
 $\| \Phi(\hat{x}) - \hat{x}\| < r (1-\kappa)$ y
 $\|\Phi(y) - \Phi(x) \| \leq \kappa \|y-x\|$ para todo $x,y\in B_r(\hat{x})$.
 Entonces existe un único $\bar{x}\in B_r(\hat{x})$ tal que $\Phi(\bar{x}) = \bar{x}$.


**Demostración:**

Sea $\hat{r} = \| \Phi(\hat{x}) - \hat{x}\|/(1-\kappa)$ y $B = \{x\in \mathbb{R}^n \mid \|x-\hat{x}\| \leq \hat{r}\}$ es la bola cerrada de centro $\hat{x}$ y radio $\hat{r}$. Claramente $B\subset B_r(\hat{x})$.
Veamos primero que $\Phi$ mapea elementos de $B$ en $B$. Sea $x\in B$ arbitrario, entonces
$$
\| \Phi(x) - \hat{x} \| \leq \| \Phi(x) - \Phi(\hat{x}) \| + \| \Phi(\hat{x}) - \hat{x}\|
    \leq \kappa \|x - \hat{x} \| + \hat{r}(1-\kappa)
    \leq \kappa \hat{r} + \hat{r}(1-\kappa) = \hat{r}.
$$
Por lo tanto $\Phi(x) \in B$.
Veamos ahora la existencia de un punto fijo. Dado $x^0 \in B$, genere la sucesión $\{x^k\}$ tal que $x^{k+1} = \Phi(x^k)$. Por lo anterior, podemos garantizar que $\{x^k\} \subset B$.
Tomo $k$, $j$ arbitrarios con $k > j$, entonces
\begin{align*}
    \| x^k - x^j\| &= \| \Phi(x^{k-1}) - \Phi(x^{j-1}) \| \leq \kappa \|x^{k-1} - x^{j-1}\| \\
    &\leq \kappa^2 \| x^{k-2} - x^{j-2}\| \leq \ldots \leq \kappa^j \| x^{k-j}  - x^0 \| \\
    &\leq \kappa^j \left( \|x^{k-j} - \hat{x}\| + \|\hat{x} - x^0\| \right) \leq 2\hat{r} \kappa^j.
\end{align*}
Usando que $\kappa < 1$, obtenemos que $\lim_{k,j \to \infty} \|x^k - x^j\| = 0$, concluyendo que $\{x^k\}$ es una sucesión de Cauchy. Luego, como $B$ es cerrado, existe $\bar{x} \in B \subset B_r(\hat{x})$ tal que $x^k \to \bar{x}$.
Ahora, como $\Phi$ es continua en $B_r(\hat{x})$, tenemos que
$$
    \bar{x} = \lim_{k \to \infty} x^{k+1} = \lim_{k \to \infty} \Phi(x^k) = \Phi(\bar{x}).
$$

Para ver la unicidad, supongamos que existe $y \in B_r(\hat{x})$ tal que $\Phi(y) = y$. Entonces
$$
\|y-\bar{x} \| (1-\kappa) = \|\Phi(y) - \Phi(\bar{x})\| - \kappa \|y-\bar{x}\|  \leq 0.
$$
Como $1 - \kappa > 0$, concluímos que $y = \bar{x}$. $\spadesuit$
