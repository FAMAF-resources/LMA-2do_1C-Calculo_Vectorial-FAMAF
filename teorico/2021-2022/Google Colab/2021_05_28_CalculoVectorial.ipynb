# Cálculo Vectorial
Recuerde que cuando se trabaja en la recta, si $f'$ es continua en $[a,b]$ entonces
$$
    \int_a^b f'(t) dt = f(b) - f(a).
$$
Además, utilizando el concepto de integral de línea, sabemos que si $\mathcal{C}$ es una curva simple con extremos $y_a$, $y_b$ entonces
$$
\int_{\mathcal{C}} \frac{\partial f}{\partial x_1}(x) dx_1 + \frac{\partial f}{\partial x_2}(x) dx_2  = \int_{\mathcal{C}} \nabla f(x) ds = f(y_b) - f(y_a).
$$
Veremos ahora variaciones de esta idea, donde la integral de una expresión que involucra las derivadas de la función puede calcularse a través de los valores de la función en un conjunto de menor dimensión.





## Teorema de Green


Para los siguientes resultados necesitaremos regiones de integración con ciertas propiedades.

>**Definición**
>
> Diremos que un conjunto $D\subset \mathbb{R}^2$ es una **región simple** si es acotado y su frontera es cortada a lo sumo dos veces, o en un segmento, por cualquier recta paralela a alguno de los ejes.


Cuando tenemos una región simple $D \subset [a,b] \times [c,d]$, podremos definir funciones $\eta_1, \xi_1 : [a,b] \to \mathbb{R}$ con $\eta_1(t) \leq \xi_1(t)$, y $\eta_2, \xi_2 : [c,d] \to \mathbb{R}$ con $\eta_2(t) \leq \xi_2(t)$ tales que
\begin{align*}
    D &= \{ x\in \mathbb{R}^2 \mid a \leq x_1 \leq b, \, \eta_1(x_1) \leq x_2 \leq \xi_1(x_1)\} \\
      &= \{ x\in \mathbb{R}^2 \mid c \leq x_2 \leq d, \, \eta_2(x_2) \leq x_1 \leq \xi_2(x_2)\}.
\end{align*}


Por otro lado, note que si $F:\mathbb{R}^2 \to \mathbb{R}^2$ y $\mathcal{C}$ es una curva en el plano parametrizada por $\gamma$, entonces
\begin{align*}
\int_{\mathcal{C}} F_1(x) dx_1 + F_2(x) dx_2 &= \int_a^b \langle F(\gamma(t)), \gamma'(t) \rangle dt
        =\int_a^b F_1(\gamma(t)) \gamma_1'(t) dt + \int_a^b F_2(\gamma(t)) \gamma_2'(t)  dt \\
        &= \int_a^b \langle \bigl(F_1(\gamma(t)), 0 \bigr), \gamma'(t) \rangle dt + \int_a^b \langle \bigl( 0 ,F_2(\gamma(t)) \bigr), \gamma'(t) \rangle dt \\
        &= \int_{\mathcal{C}} F_1(x) dx_1 + \int_{\mathcal{C}} F_2(x) dx_2.
\end{align*}



>**Teorema** (Green)
>
> Sea $D\subset \mathbb{R}^2$ acotado y unión finita de regiones simples. Si $F: \mathbb{R}^2 \to \mathbb{R}^2$ es continuamente diferenciable en $\operatorname{cl}(D)$, entonces
 $$
    \int_D \left( \frac{\partial F_2}{\partial x_1}(x) - \frac{\partial F_1}{\partial x_2}(x) \right)dx = \int_{\partial D} F_1(x) dx_1 + F_2(x) dx_2.
 $$


**Demostración**

Supongamos primero que $D$ es una región simple. Considere por un lado que $D = \{ x\in \mathbb{R}^2 \mid a \leq x_1 \leq b, \, \eta_1(x_1) \leq x_2 \leq \xi_1(x_1)\}$, luego podemos escribir $\partial D = \mathcal{C}_1 \cup \mathcal{C}_2 \cup \mathcal{C}_3 \cup \mathcal{C}_4$ donde $\mathcal{C}_i$ es parametrizada por $\gamma^i$ con
\begin{align*}
    \gamma^1(t) &= \bigl( t, \eta_1(t) \bigr), & a &\leq t \leq b,\\
    \gamma^2(t) &= \bigl( b, t \bigr), & \eta_1(b) & \leq t \leq \xi_1(b),\\
    \gamma^3(t) &= \bigl( a+b-t, \xi_1(a+b-t) \bigr), &  a &\leq t \leq b,\\
    \gamma^4(t) &= \bigl( a, \eta_1(a)+\xi_1(a) -t \bigr),  & \eta_1(a) &\leq t \leq \xi_1(a).
\end{align*}
Por lo tanto, como $(\gamma^2_1)'(t) = (\gamma^4_1)'(t) = 0$,
\begin{align*}
    \int_{\partial D} F_1(x_1,x_2) dx_1
    =& \int_{\mathcal{C}_1} F_1(x_1,x_2) dx_1 + \int_{\mathcal{C}_2} F_1(x_1,x_2) dx_1 \\
     &+ \int_{\mathcal{C}_3} F_1(x_1,x_2) dx_1 + \int_{\mathcal{C}_4} F_1(x_1,x_2) dx_1 \\
    =& \int_a^b F_1(t, \eta_1(t)) dt + \int_a^b F_1(a+b-t, \xi_1(a+b-t))(-1) dt \\
    =& \int_a^b F_1(x_1, \eta_1(x_1)) dx_1 + \int_b^a F_1(x_1, \xi_1(x_1)) dx_1 \\
    =& \int_a^b \bigl( F_1(x_1, \eta_1(x_1)) - F_1(x_1, \xi_1(x_1)) \bigr) dx_1 \\
    =& \int_a^b \left( - \int_{\eta_1(x_1)}^{\xi_1(x_1)} \frac{\partial F_1}{\partial x_2}(x_1,x_2) dx_2 \right) dx_1 \\
    =& \int_D  -\frac{\partial F_1}{\partial x_2}(x_1,x_2) dx_1 dx_2.
\end{align*}

Considere ahora $D = \{ x\in \mathbb{R}^2 \mid c \leq x_2 \leq d, \, \eta_2(x_2) \leq x_1 \leq \xi_2(x_2)\}$, entonces $\partial D = \mathcal{C}_1 \cup \mathcal{C}_2 \cup \mathcal{C}_3 \cup \mathcal{C}_4$ donde $\mathcal{C}_i$ es parametrizada por $\gamma^i$ con
\begin{align*}
    \gamma^1(t) &= \bigl( \eta_2(c+d-t), c+d-t \bigr), & c&\leq t \leq d,\\
    \gamma^2(t) &= \bigl( t , c \bigr), & \eta_2(c)&\leq t \leq \xi_2(c),\\
    \gamma^3(t) &= \bigl( \xi_2(t), t  \bigr), & c&\leq t \leq d,\\
    \gamma^4(t) &= \bigl( \eta_2(d)+\xi_2(d)-t, d \bigr),  &\eta_2(d)&\leq t \leq \xi_2(d).
\end{align*}
Como $(\gamma^2_2)'(t) = (\gamma^4_2)'(t) = 0$,
\begin{align*}
    \int_{\partial D} F_2(x_1,x_2) dx_2
    =& \int_{\mathcal{C}_1} F_2(x_1,x_2) dx_2 + \int_{\mathcal{C}_2} F_2(x_1,x_2) dx_2 \\
     &+ \int_{\mathcal{C}_3} F_2(x_1,x_2) dx_2 + \int_{\mathcal{C}_4} F_2(x_1,x_2) dx_2 \\
    =& \int_c^d F_2(\eta_2(c+d-t), c+d-t )(-1) dt + \int_c^d F_2(\xi_2(t), t ) dt \\
    =& \int_d^c F_2(\eta_2(x_2), x_2 ) dx_2 + \int_c^d F_2(\xi_2(x_2), x_2 ) dx_2 \\
    =& \int_c^d \bigl(  F_2(\xi_2(x_2), x_2 ) - F_2(\eta_2(x_2), x_2 ) \bigr) dx_2 \\
    =& \int_c^d \left(  \int_{\eta_2(x_2)}^{\xi_2(x_2)} \frac{\partial F_2}{\partial x_1}(x_1,x_2) dx_1 \right) dx_2 \\
    =& \int_D  \frac{\partial F_2}{\partial x_1}(x_1,x_2) dx_1 dx_2.
\end{align*}
Concluyendo que
$$
 \int_{\partial D} F_1(x_1,x_2) dx_1 + F_2(x_1,x_2) dx_2
 = \int_D  \left( \frac{\partial F_2}{\partial x_1}(x_1,x_2) - \frac{\partial F_1}{\partial x_2}(x_1,x_2)\right) dx_1 dx_2.
$$

Supongamos ahora que $D = \cup_{i=1}^k D_i$ para regiones simples $D_i$. Note que las partes de $\partial D_i$ que no pertenecen a $\partial D$ son recorridas en las dos posibles orientaciones, y como para una curva orientada $\mathcal{C}$,
$$
\int_{\mathcal{C}} F_1(x) dx_1 + F_2(x) dx_2 + \int_{\mathcal{C}^-} F_1(x) dx_1 + F_2(x) dx_2 = 0,
$$
podemos concluír que
\begin{align*}
    \int_D  \left( \frac{\partial F_2}{\partial x_1}(x) - \frac{\partial F_1}{\partial x_2}(x)\right) dx
    &= \sum_{i=1}^k \int_{D_i}  \left( \frac{\partial F_2}{\partial x_1}(x) - \frac{\partial F_1}{\partial x_2}(x)\right) dx \\
    &=\sum_{i=1}^k  \int_{\partial D_i} F_1(x) dx_1 + F_2(x) dx_2 \\
    &=\int_{\partial D} F_1(x) dx_1 + F_2(x) dx_2. \quad \spadesuit
\end{align*}




Entre las distintas consecuencias que podemos obtener, veamos la siguiente.

>**Corolario**
>
> Sea $\mathcal{C}$ una curva simple en $\mathbb{R}^2$ orientada positivamente y sea $D$ la región acotada por esta curva, entonces $\mathcal{C} = \partial D$ y
$$
    |D| = \frac{1}{2} \int_{\partial D} x_1 dx_2 - x_2 dx_1.
$$

**Demostración**

Tomando $F(x_1,x_2) = (-x_2, x_1)$, tenemos que
\begin{align*}
    \frac{1}{2} \int_{\partial D} x_1 dx_2 - x_2 dx_1
    &= \frac{1}{2} \int_{\partial D} F_1(x) dx_1 + F_2(x) dx_2 \\
    &= \frac{1}{2} \int_D  \left( \frac{\partial F_2}{\partial x_1}(x) - \frac{\partial F_1}{\partial x_2}(x)\right) dx \\
    &= \frac{1}{2} \int_D  (1 -(-1)) dx = |D|.\quad \spadesuit
\end{align*}





Para los próximos resultados necesitaremos definir el siguiente operador.

>**Definición**
>
> Si $F:\mathbb{R}^3 \to \mathbb{R}^3$ es continuamente diferenciable, el **rotacional** de $F$ es
$$
 \operatorname{rot}(F)(x)
  = \left( \frac{\partial F_3}{\partial x_2}(x) - \frac{\partial F_2}{\partial x_3}(x),
             \frac{\partial F_1}{\partial x_3}(x) - \frac{\partial F_3}{\partial x_1}(x),
             \frac{\partial F_2}{\partial x_1}(x) - \frac{\partial F_1}{\partial x_2}(x) \right).
$$


Cuando se tiene un campo vectorial $F:\mathbb{R}^2 \to \mathbb{R}^2$ se considera $\operatorname{rot}(F)$ como $\operatorname{rot}(\bar{F})$ donde $\bar{F}:\mathbb{R}^3 \to \mathbb{R}^3$ con $\bar{F}(x_1,x_2,x_3) = \bigl( F_1(x_1,x_2), F_2(x_1,x_2), 0 \bigr)$.



>**Corolario**
>
> Sea $D\subset \mathbb{R}^2$ acotado y unión finita de regiones simples. Si $F: \mathbb{R}^2 \to \mathbb{R}^2$ es continuamente diferenciable en $\operatorname{cl}(D)$, entonces
$$
 \int_{\partial D} F(x) ds = \int_D \langle \operatorname{rot}(F)(x), e^3 \rangle dx.
$$


**Demostración**

Es inmediato al notar que
$$
\langle \operatorname{rot}(F)(x), e^3 \rangle = \frac{\partial F_2}{\partial x_1}(x) - \frac{\partial F_1}{\partial x_2}(x). \quad \spadesuit
$$


#### Ejemplo

Se desea conocer la rotación de un campo bidimensional $F(x)=(x_1 x_2^2, x_1 + x_2)$ sobre la región del plano entre las curvas $x_2=x_1^2$ y $x_2 = x_1$.

Como
$$
\operatorname{rot}(F)(x) = \left( 0, 0, \frac{\partial F_2}{\partial x_1}(x) - \frac{\partial F_1}{\partial x_2}(x) \right)
= \bigl( 0, 0, 1 - 2x_1 x_2 \bigr),
$$
entonces
\begin{align*}
\int_D \langle \operatorname{rot}(F)(x), e^3 \rangle dx
&= \int_0^1 \int_{x_1^2}^{x_1} 1-2x_1x_2 dx_2 dx_1 \\
&= \int_0^1 \left[ x_2- x_1 x_2^2 \right]_{x_2=x_1^2}^{x_2=x_1} dx_1 \\
&= \int_0^1 \left( x_1- x_1^3 - x_1^2 + x_1^5 \right) dx_1 \\
&= \textstyle \frac{1}{2} - \frac{1}{4} - \frac{1}{3} + \frac{1}{6} = \frac{1}{12}.
\end{align*}

Por otro lado, la frontera de la región puede recorrerse de manera positiva por las curvas $\mathcal{C}_1$ y $\mathcal{C}_2$, parametrizadas por
\begin{align*}
    \gamma^1(t) &= \bigl( t, t^2  \bigr), & 0 &\leq t \leq 1,\\
    \gamma^2(t) &= \bigl( 1-t, 1-t \bigr), & 0 & \leq t \leq 1.
\end{align*}
Obteniendo que
\begin{align*}
 \int_{\partial D} F(x) ds
 &=  \int_{\mathcal{C}_1} F(x) ds +  \int_{\mathcal{C}_2} F(x) ds \\
 &=  \int_0^1 t^5 + 2t(t+t^2) dt +  \int_0^1 -(1-t)^3 - 2(1-t) dt \\
 &= \textstyle \left[\frac{t^6}{6} + \frac{2t^3}{3} + \frac{t^4}{2} \right]_0^1 + \left[ \frac{(1-t)^4}{4} + (1-t)^2 \right]_0^1 \\
 &= \textstyle \frac{1}{6} + \frac{2}{3} + \frac{1}{2} - \frac{1}{4} - 1= \frac{1}{12}
\end{align*}

Veamos ahora otro operador importante para los siguientes resultados.

>**Definición**
>
> Si $F:\mathbb{R}^n \to \mathbb{R}^n$ es continuamente diferenciable, la **divergencia** de $F$ es
$$
    \operatorname{div}(F)(x) = \frac{\partial F_1}{\partial x_1}(x)
    + \ldots + \frac{\partial F_n}{\partial x_n}(x).
$$




>**Corolario**
>
> Sea $D\subset \mathbb{R}^2$ acotado y unión finita de regiones simples, y sea $\nu$ el vector normal unitario saliente a $D$. Si $F: \mathbb{R}^2 \to \mathbb{R}^2$ es continuamente diferenciable en $\operatorname{cl}(D)$, entonces
$$
 \int_{\partial D} \langle F(x), \nu(x) \rangle ds = \int_D \operatorname{div}(F)(x) dx.
$$


**Demostración**

Si $\gamma:[a,b] \to \mathbb{R}^2$ es la parametrización de $\partial D$ y denotamos por $\nu(t)$ a $\nu(\gamma(t))$, tenemos que
$$
    \langle \gamma'(t), \nu(t) \rangle = 0 \quad \textrm{ y } \quad \|\nu(t) \|=1.
$$
Sumando que el vector debe ser saliente, obtenemos que
$$
    \nu(t) = \frac{1}{\|\gamma'(t)\|} \bigl( \gamma_2'(t), - \gamma_1'(t) \bigr).
$$
Por lo tanto
\begin{align*}
     \int_{\partial D} \langle F(x), \nu(x) \rangle ds
     &= \int_a^b \langle F(\gamma(t)), \nu(t) \rangle \|\gamma'(t)\| dt \\
     &= \int_a^b F_1(\gamma(t)) \gamma_2'(t) - F_2(\gamma(t)) \gamma_1'(t) dt \\
     &= \int_{\partial D} (- F_2)(x) dx_1 + F_1(x) dx_2  \\
     &= \int_D  \left( \frac{\partial F_1}{\partial x_1}(x) - \frac{\partial (- F_2)}{\partial x_2}(x)\right) dx \\
     &= \int_D \operatorname{div}(F)(x) dx. \qquad \spadesuit
\end{align*}


Note que si usamos el resultado anterior para el caso unidimensional, tendríamos que $D=[a,b]$, $\partial D=\{a, b\}$ y el vector normal unitario saliente a $D$ sería $\nu(a) = -1$ y $\nu(b) = 1$. Por lo tanto, si $f:\mathbb{R} \to \mathbb{R}$,
$$
\int_a^b f'(x) dx = \int_D \operatorname{div}(f) dx
= \int_{\partial D} f(x)\nu(x) ds = f(a)\nu(a) + f(b)\nu(b) = f(b) - f(a).
$$

Por otro lado, si $F:\mathbb{R}^2 \to \mathbb{R}^2$ y $g:\mathbb{R}^2 \to \mathbb{R}$ se tiene que
$$
\operatorname{div}(gF)(x) = \langle \nabla g(x) , F(x) \rangle + g(x) \operatorname{div}(F)(x).
$$
Obteniendo así
$$
\int_D g(x) \operatorname{div}(F)(x) dx = \int_{\partial D} g(x) \langle F(x), \nu(x) \rangle ds - \int_D \langle \nabla g(x) , F(x) \rangle dx.
$$

Con esto, si $D=[a,b]$, $\partial D=\{a, b\}$, $\nu(a) = -1$ y $\nu(b) = 1$. Por lo tanto, si $f:\mathbb{R} \to \mathbb{R}$ y $g:\mathbb{R} \to \mathbb{R}$,
\begin{align*}
\int_a^b  g(x) f'(x) dx
&= \int_D g(x) \operatorname{div}(f)(x) dx \\
&= \int_{\partial D} g(x) f(x) \nu(x)  ds - \int_D g'(x) f(x) dx\\
&=  g(a) f(a) \nu(a) + g(b) f(b) \nu(b) - \int_a^b g'(x) f(x) dx\\
&=  g(x) f(x)\bigm|_a^b  - \int_a^b g'(x) f(x) dx
\end{align*}
