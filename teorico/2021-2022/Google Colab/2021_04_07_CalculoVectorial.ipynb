## Definiendo funciones

Recordemos que una función es una regla de asignación que asocia a cada elemento de un conjunto de partida un único elemento de un conjunto de llegada.

El **dominio** de una función es el conjunto donde la función está definida y la **imagen** (o rango) de una una función es el conjunto de valores que puede asumir. El dominio de la función es un subconjunto del **espacio dominio** y la imagen un subconjunto del **espacio imagen**. En este curso los espacios dominio e imagen serán espacios Euclídeos.
Por lo tanto, si denotamos con $f$ la función a estudiar, siempre tendremos
$$
f:\mathbb{R}^n \to \mathbb{R}^m,
$$
para espacios de dimensión $n$ y $m$. Otra notación usada para una función es $x \mapsto f(x)$, acentuando que $f$ lleva $x$ a $f(x)$. O sea, para cada $x=(x_1, x_2, \ldots, x_n)$ la acción de la función $f$ nos retorna
$$
f(x) = (f_1(x), f_2(x), \ldots, f_m(x)),
$$
donde $f_i$ es llamada una función coordenada. En algunos casos, cuando sea necesario enfatizar el dominio $D$ de la función $f$, utilizaremos $f: D\subset \mathbb{R}^n \to \mathbb{R}^m$. Cuando el espacio dominio $\mathbb{R}^n$ tiene $n>1$, diremos que $f$ es una función de varias variables. Si el espacio imagen $\mathbb{R}^m$ tiene $m=1$ diremos que $f$ es una función a valores escalares, si $m>1$ diremos que $f$ es una función a valores vectoriales.
Se suele llamar *campo vectorial* a una función cuyo espacio de dominio y de imagen son iguales.

El uso de $f$ para denotar una función y de $x$ para denotar su variable, corresponde a una convención matemática. La notación universal no existe y dependiendo del problema en estudio se pueden encontrar funciones y variables denotadas con otras letras o incluso palabras.

Formalmente, el dominio e imagen de una función $f:\mathbb{R}^n \to \mathbb{R}^m$ son los conjuntos
\begin{gather*}
    \operatorname{dom}(f) = \{x\in\mathbb{R}^n \mid f(x) = y \textrm{ para algún } y \in \mathbb{R}^m\}, \\
    \operatorname{img}(f) = \{y\in\mathbb{R}^m \mid y = f(x) \textrm{ para algún } x \in \mathbb{R}^n\}.
\end{gather*}

Estos conjuntos juegan un rol importante en el estudio de una función.


>**Definición**
>
>Diremos que $f:\mathbb{R}^n \to \mathbb{R}^m$ es **sobreyectiva** (o sobre) si cada elemento de $\mathbb{R}^m$ está en la imagen de $f$, o sea, $\operatorname{img}(f) = \mathbb{R}^m$.
>
>Diremos que $f$ es **inyectiva** (o uno a uno) si dos elementos distintos del dominio de $f$ no son llevados a un mismo elemento de la imagen, o sea, si $x^1, x^2 \in\operatorname{dom}(f)$ y $x^1 \neq x^2$ entonces $f(x^1) \neq f(x^2) $.

#### Ejemplo (funciones didácticas)

Considere la función $f:\mathbb{R}^3 \to \mathbb{R}^2$ tal que
$$
f(x,y,z) = \bigl(x^2 + y^2 + z^2\, ,\, x + y  + z \bigr).
$$
Sus funciones coordenadas son $f_1(x,y,z) = x^2 + y^2 + z^2$ y $f_2(x,y,z) = x + y  + z$. Podemos ver que $\operatorname{dom}(f) = \mathbb{R}^3$ y $\operatorname{img}(f) = \{ w\in \mathbb{R}^2 \mid w_1\geq 0\}$. Por lo tanto la función no es sobreyectiva. Además no es inyectiva, pues $f(t,-t,0) = f(-t,t,0)$.

Para la función $g: \mathbb{R}^2 \to \mathbb{R}^2$ tal que
$$
g(x) = \left( \sqrt{1- x_1^2 - x_2^2}\, ,\, x_1 + x_2 \right),
$$
tenemos funciones coordenadas $g_1(x) = \sqrt{1- x_1^2 - x_2^2}$ y $g_2(x) = x_1 + x_2$. Claramente, $\operatorname{dom}(g) = \{ x\in \mathbb{R}^2 \mid x_1^2 + x_2^2 \leq 1\}$ debido a la raíz cuadrada. Por otro lado, veamos que $\operatorname{img}(g) = \left\{ y\in \mathbb{R}^2 \mid  |y_2| \leq \sqrt{2}\sqrt{1 - y_1^2}, \, y_1 \in [0,1]\right\}$. Si $x\in\operatorname{dom}(g)$,
$$
    0 \leq 1 - x_1^2 - x_2^2 \leq 1 \Rightarrow 0\leq g_1(x) \leq 1,
$$
y usando la desigualdad de Cauchy-Schwarz,
$$
|g_2(x)| = |\langle (1,1)\, , \, (x_1,x_2)\rangle| \leq \sqrt{2}\sqrt{x_1^2 + x_2^2}
= \sqrt{2} \sqrt{1 - g_1(x)^2}.
$$



```
#@title Función didáctica { display-mode: "form" }
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

from matplotlib import rc
#rc('animation', html='jshtml')
rc('animation', html='html5')


o1 = np.array([-2,0])
o2 = np.array([2,0])
pt = np.array([np.cos(np.pi/4), np.sin(np.pi/4)])
s = np.flip(np.linspace(-1,1,100))
pol1 = np.outer(pt,s)
th = np.linspace(-3*np.pi/4, np.pi/4, 100)
pol2 = np.array([np.cos(th), np.sin(th)])
x = np.concatenate((pol1,pol2), axis=1)
g = np.array([np.sqrt(np.abs(1 - x[0]**2 - x[1]**2)), x[0]+x[1]])
xr = -x
gr = np.array([np.sqrt(np.abs(1 - xr[0]**2 - xr[1]**2)), xr[0]+xr[1]])

#cm = 1/2.54
#fig, (ax1,ax2) = plt.subplots(2,1,figsize=(15*cm, 5*cm))
fig, (ax1,ax2) = plt.subplots(2,1)
plt.close()
def animacion(t):
  ax1.clear()
  ax2.clear()
  # ejes
  ax1.axis('off')
  ax1.axis('equal')
  ax1.plot((1-t)*o1[0]+t*o2[0] + np.array([-1.1,1.1]), np.array([0,0]), color=(0.5,0.5,0.5))
  ax1.plot((1-t)*o1[0]+t*o2[0] + np.array([0,0]), np.array([-1.5,1.5]), color=(0.5,0.5,0.5))
  ax1.plot([o1[0]-1.1, o2[0]+1.1], [o1[1]-1.5, o2[1]+1.5],'w.')
  # ejes
  ax2.axis('off')
  ax2.axis('equal')
  ax2.plot((1-t)*o1[0]+t*o2[0] + np.array([-1.1,1.1]), np.array([0,0]), color=(0.5,0.5,0.5))
  ax2.plot((1-t)*o1[0]+t*o2[0] + np.array([0,0]), np.array([-1.5,1.5]), color=(0.5,0.5,0.5))
  ax2.plot([o1[0]-1.1, o2[0]+1.1], [o1[1]-1.5, o2[1]+1.5],'w.')
  ax2.set_title(r'$g(x) = \left( \sqrt{1- x_1^2 - x_2^2}\, ,\, x_1 + x_2 \right)$')
  # conjunto
  ax1.fill( (1-t)*(o1[0]+x[0])+t*(o2[0]+g[0]), (1-t)*(o1[1]+x[1])+t*(o2[1]+g[1]))
  plot = ax2.fill( (1-t)*(o1[0]+xr[0])+t*(o2[0]+gr[0]), (1-t)*(o1[1]+xr[1])+t*(o2[1]+gr[1]))
  return plot

anim = animation.FuncAnimation(fig, animacion, frames=np.linspace(0,1,50), repeat = False)

anim
```

Esta función no es sobreyectiva, pues $\operatorname{img}(g) \neq \mathbb{R}^2$ y tampoco es inyectiva ya que  
\begin{align*}
g\left(r\cos({\textstyle \frac{\pi}{4}}+\theta), r\operatorname{sen}({\textstyle\frac{\pi}{4}}+\theta)\right)
    &=\left(\sqrt{1-r^2}, \sqrt{2}\cos(\theta) \right) \\
    &= g\left(r\cos({\textstyle\frac{\pi}{4}}-\theta), r\operatorname{sen}({\textstyle\frac{\pi}{4}}-\theta)\right).
\end{align*}

## Funciones lineales

Para facilitar cálculos, en algunos casos los vectores del espacio dominio y espacio imagen se asociarán con vectores columna, o sea, asociaremos $\mathbb{R}^n$ con $\mathbb{R}^{n \times 1}$. Aunque a simple vista ésto parece un detalle estético, el mismo nos permitirá obtener representaciones matriciales y realizar cálculos utilizando operaciones algebráicas matriciales.


---

#### Ejemplo

Considere la función $h:\mathbb{R}^3 \to \mathbb{R}^2$ tal que
$$
h(x,y,z) = (3x+4y, 3y+5z).
$$
Si escribimos el espacio imagen $\mathbb{R}^2$ como $\mathbb{R}^{2 \times 1}$, obtenemos que
$$
h(x,y,z) = \left[ \begin{array}{c}
     3x+4y \\
     3y+5z
\end{array}\right]
    = \left[ \begin{array}{ccc}
     3 & 4 & 0 \\
     0 & 3 & 5
    \end{array}\right]
    \left[ \begin{array}{c}
     x \\ y \\ z
     \end{array}\right].
$$

---



La función anterior es miembro de una familia muy importante de funciones.

>**Definición**
>
>Diremos que $f: V \subset \mathbb{R}^n \to \mathbb{R}^m$, con $V$ subespacio, es una **función lineal** si
\begin{align*}
    f(x+y) &=f(x) + f(y),\\
    f(\alpha x) &= \alpha f(x),
\end{align*}
para todo $x,y\in\mathbb{R}^n$ y todo $\alpha\in\mathbb{R}$.



Claramente, una función lineal debe cumplir que $f(0) = 0$.
Lo visto en el ejemplo anterior, puede formaliarse de la siguiente forma.


>**Teorema**
>
> Una función $f:\mathbb{R}^n \to \mathbb{R}^m$ es lineal, si y sólo si su representación matricial coincide con el producto de uma matriz $m\times n$ y un vector $n\times 1$.


**Demostración:**
($\Rightarrow$). Sean $e^j \in \mathbb{R}^{n\times 1}$ los vectores canónicos. Defina la matriz $A$ tal que su $j$-ésima columna es el vector $f(e^j) \in \mathbb{R}^m$, para $j=1, \ldots,n$. Por lo tanto $A\in \mathbb{R}^{m \times n}$ y para $x\in\mathbb{R}^{n \times 1}$,
\begin{align*}
    f(x)&= f(x_1 e^1 + \ldots + x_n e^n) \\
        &= x_1 f(e^1) + \ldots + x_n f(e^n) \\
        &= x_1 \left[\begin{array}{c} a_{11} \\ a_{21} \\ \vdots \\ a_{m1} \end{array} \right]
         + \ldots
         + x_n \left[\begin{array}{c} a_{1n} \\ a_{2n} \\ \vdots \\ a_{mn} \end{array} \right]\\
        &=\left[\begin{array}{c}    a_{11}x_1 + \ldots + a_{1n}x_n \\
                                    a_{21}x_1 + \ldots + a_{2n}x_n \\
                                    \vdots \\
                                    a_{m1}x_1 + \ldots + a_{mn}x_n \end{array} \right]\\
        &= Ax.
\end{align*}

($\Leftarrow$). Si $f(x) = Ax$ para $A\in\mathbb{R}^{m\times n}$ y $x\in\mathbb{R}^{n\times 1}$, entonces para todo $x,y\in\mathbb{R}^{n\times 1}$ y $\alpha\in \mathbb{R}$,
\begin{align*}
    f(x+y) &= A(x+y) = Ax + Ay = f(x) + f(y),\\
    f(\alpha x) &= A(\alpha x) = \alpha Ax = \alpha f(x). \qquad \spadesuit
\end{align*}


Por construcción, note que la matriz asociada a una función lineal $f$ debe tener su $j$-ésima columna igual a $f(e^j)$.


Otra consecuencia directa del resultado anterior es que la composición de funciones lineales es una función lineal. Pues, si $f:\mathbb{R}^n \to \mathbb{R}^m$ y $g:\mathbb{R}^m \to \mathbb{R}^p$ son lineales con matrices asociadas $A\in\mathbb{R}^{m\times n}$ y $B\in \mathbb{R}^{p \times m}$, respectivamente, entonces $g\circ f :\mathbb{R}^n \to \mathbb{R}^p$ tiene asociada la matriz $BA \in \mathbb{R}^{p \times n}$. Claramente,
$$
(g \circ f)(x) = g\bigl(f(x)\bigr) = g(Ax) = B(Ax) = (BA) x.
$$


Para funciones de una variable algunos autores llaman función lineal a una de la forma $f(x) = ax+b$. Note que este tipo de funciones no corresponden a la definición dada anteriormente. Por ejemplo, si $f(x) = 2x+1$ entonces $f(0) = 1 \neq 0$. Este tipo de funciones presenta una traslación del origen en el espacio imagen, formalmente, dado $b\in\mathbb{R}^n$ llamaremos translación inducida por $b$ a la función $x \mapsto x+b$.

>**Definición**
>
> Diremos que $f: V \subset \mathbb{R}^n \to \mathbb{R}^m$, con $V$ subespacio, es una **función afín** si es la composición de una función lineal y una traslación.



Con una demostración análoga a la vista para funciones lineales, se tiene lo siguiente.

>**Teorema**
>
> Una función $f:\mathbb{R}^n \to \mathbb{R}^m$ es afín, si y sólo si su representación matricial es de la forma $f(x) = Ax+b$ con $A\in\mathbb{R}^{m \times n}$ y $b\in\mathbb{R}^m$.


Es fácil ver que una definición equivalente de función afín es pedir que $x \mapsto f(x) - f(0)$ sea lineal.


## Graficando funciones

La comprensión de la acción de las funciones es esencial, y para ello es de mucha utilidad poder contar con una representación gráfica. Por más que no podremos representar gráficamente todas las funciones vectoriales, el estudio de su gráfico nos brinda muchas propiedades.

>**Definición**
>
> Diremos que el gráfico de una función $f: \mathbb{R}^n \to \mathbb{R}^m$ es el conjunto
$$
\operatorname{gr}(f) = \{ (x,f(x)) \in \mathbb{R}^{n+m} \mid x \in \operatorname{dom}(f)\}.
$$


Si $n=m=1$ este conjunto es justamente la curva en $\mathbb{R}^2$ que representa gráficamente a $f$. Claramente podremos visualizar el gráfico de $f$ si $n+m\leq 3$.



Para poder tener representaciones gráficas en dimensiones mayores, será de mucha utilidad los siguientes enfoques.

>**Definición**
>
> Diremos que para $f: \mathbb{R}^n \to \mathbb{R}^m$ un conjunto $S$ es definido:
*  **explícitamente** si $S\subset \mathbb{R}^{n+m}$ es el gráfico de $f$, o sea $S = \operatorname{gr}(f)$,    
*  **paramétricamente** si $S\subset \mathbb{R}^{m}$ es la imagen de $f$, o sea $S = \operatorname{img}(f)$,
* **implícitamente** si $S\subset \mathbb{R}^{n}$ es un **conjunto de nivel** de $f$, o sea, para algún $\bar{y}\in\mathbb{R}^m$, $S = \{x\in \mathbb{R}^n \mid f(x) = \bar{y}\}$.


Cuando $f:\mathbb{R}^2 \to \mathbb{R}$, el conjunto de nivel se conoce como **curva de nivel**. Cuando $f:\mathbb{R}^3 \to \mathbb{R}$, el conjunto de nivel se conoce como **superficie de nivel**.
Veamos algunos ejemplos.


#### Ejemplo

Considere la función norma Euclídea en $\mathbb{R}^2$, o sea $f:\mathbb{R}^2 \to \mathbb{R}$ tal que
$$
f(x,y) = \|(x,y)\| = \sqrt{x^2 + y^2}.
$$


```
#@title Norma Euclídea { form-width: "30%", display-mode: "form" }

import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits import mplot3d
import matplotlib.animation as animation

from matplotlib import rc
rc('animation', html='jshtml')

x = np.linspace(-5,5,200)
y = np.linspace(-5,5,200)
X, Y = np.meshgrid(x, y)
Z = np.sqrt(X**2 + Y**2)
Xp = X[[0,-1],:][:,[0,-1]]
Yp = Y[[0,-1],:][:,[0,-1]]

fig=plt.figure(figsize=plt.figaspect(0.5))
fig.suptitle(r'$f(x,y) = \sqrt{x^2 + y^2}$')
ax2=fig.add_subplot(1,2,1,projection='3d')
ax1=fig.add_subplot(1,2,2)

plt.close()

def animacion(nivel):
  ax1.clear()
  ax2.clear()
  CS = ax1.contour(X, Y, Z, levels=[nivel])
  ax1.clabel(CS, inline=True, fontsize=10)
  ax1.set_title(f'Curva de nivel $f(x,y)=${nivel:.3f}')
  ax1.axis('equal')


  Zp = nivel*np.ones((2,2))
  ax2.plot_surface(Xp, Yp, Zp, linewidth=0, alpha=0.2, color='y')
  ax2.plot_surface(X, Y, Z, linewidth=0, alpha=0.4)

  ax2.set_title('Gráfico de $f$')
  ax2.set_xlabel('$x$')
  ax2.set_ylabel('$y$')
  plot = ax2.view_init(30, 30);
  return plot


anim = animation.FuncAnimation(fig, animacion, frames=np.linspace(0.5,6,10), repeat = False)

anim

```

Claramente $\operatorname{dom}(f) = \mathbb{R}^2$ y $\operatorname{img}(f) = \{ \alpha\in\mathbb{R} \mid \alpha\geq 0\}$. Además, tenemos que $\operatorname{gr}(f) = \{(x,y,\sqrt{x^2 + y^2})\in\mathbb{R}^3 \mid x,y\in\mathbb{R}\}$ y su representación gráfica es una superficie cónica con vértice en el origen, por lo tanto esa superficie está definida explícitamente por $f$. Ahora, dado $r\geq 0$ tenemos que el conjunto de nivel de $f$ en $r$ es $\{(x,y)\in\mathbb{R}^2 \mid \sqrt{x^2 + y^2} = r\}$, su representación gráfica es una **circunferencia** con centro en el origen y radio $r$ y por lo tanto esta circunferencia esta definida implícitamente por $f$.

La circunferencia también puede describirse paramétricamente.

#### Ejemplo

Considere la función $f:\mathbb{R} \to \mathbb{R}^2$ tal que
$$
f(t) =(\cos(t), \operatorname{sen}(t)).
$$


```
#@title Hélice circular { form-width: "30%", display-mode: "form" }

import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits import mplot3d
import matplotlib.animation as animation

from matplotlib import rc
rc('animation', html='jshtml')


#t = 1.
#i = np.int((t+6.5)/13*199)
th = np.linspace(-6.5, 6.5,200)

fig=plt.figure(figsize=(10,4))
fig.suptitle(r'$f(t) = (\cos(t), \mathrm{sen}(t))}$')
ax1=fig.add_subplot(1,2,1,projection='3d')
ax2=fig.add_subplot(1,2,2)
plt.close()

def animacion(i):
  ax1.clear()
  ax2.clear()
  # fig1
  ax1.plot3D([th[0], th[-1]], [0,0], [0,0], color=(0.5,0.5,0.5))
  ax1.plot3D([0,0], [-3,3], [0,0], color=(0.5,0.5,0.5))
  ax1.plot3D([0,0], [0,0], [-3,3], color=(0.5,0.5,0.5))
  ax1.plot3D(th, np.cos(th), np.sin(th))
  ax1.plot3D([th[i]], [np.cos(th[i])], [np.sin(th[i])],'ro')
  ax1.set_title('Gráfico de $f$')
  ax1.set_xlabel('$x$')
  ax1.set_ylabel('$y$')
  ax1.set_zlabel('$z$')

  # fig2
  ax2.plot([-1,1], [0,0], color=(0.5,0.5,0.5))
  ax2.plot([0,0], [-1,1], color=(0.5,0.5,0.5))
  ax2.plot(np.cos(th), np.sin(th))
  ax2.plot(np.cos(th[i]), np.sin(th[i]),'ro')
  ax2.set_title('Imagen de $f$')
  plot = ax2.axis('equal');
  return plot

#plt.savefig('ejm_heliceCilindrica.pdf')

anim = animation.FuncAnimation(fig, animacion, frames=range(0,200,5), repeat = False)

anim
```

Analizando esta función tenemos que $\operatorname{dom}(f) = \mathbb{R}$. Como $\|f(t)\| = \sqrt{\cos^2(t) + \operatorname{sen}^2(t)} = 1$, entonces $\operatorname{img}(f) = \{ x\in\mathbb{R}^2 \mid \| x\| = 1\}$, o sea $f$ define paramétricamente la circunferencia unitaria . El gráfico $\operatorname{gr}(f) = \{ (t, \cos(t), \operatorname{sen}(t)) \in \mathbb{R}^3 \mid t\in \mathbb{R}\}$ es una curva conocida como **hélice circular**, por ello esta curva está definida explícitamente por $f$.

---

Con la hélice circular se puede generar una superficie como sigue.

#### Ejemplo

Considere la función $f:\mathbb{R}^2 \to \mathbb{R}^3$ tal que
$$
f(u,v) = \bigl( u\cos(v), u\operatorname{sen}(v), v\bigr),\ \textrm{ para } \ 0\leq u \leq 4, \ 0 \leq v \leq 2 \pi.
$$


```
#@title Helicoide circular { form-width: "30%", display-mode: "form" }

import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits import mplot3d
import matplotlib.animation as animation

from matplotlib import rc
rc('animation', html='jshtml')

u = np.linspace(0,4,100)
v = np.linspace(0, 2*np.pi,200)
U, V = np.meshgrid(u,v)
X = U*np.cos(V)
Y = U*np.sin(V)


fig=plt.figure(figsize=(10,4))
fig.suptitle(r'$f(u,v) = (u\cos(v), u\mathrm{sen}(v), v)}$')
ax1=fig.add_subplot(1,2,1)
ax2=fig.add_subplot(1,2,2,projection='3d')
plt.close()

#i = 10 # range(50)
def animacion(i):
  ax1.clear()
  ax2.clear()
  iu = 2*i
  iv = 4*i
  # fig2
  ax2.plot_surface(X, Y, V, linewidth=0, alpha=0.4)
  ax2.plot3D(X[:,iu], Y[:,iu], V[:,iu], color='r')
  ax2.plot3D(X[iv,:], Y[iv,:], V[iv,:], color='m')
  ax2.plot3D([-4, 4], [0,0], [0,0], color=(0.5,0.5,0.5))
  ax2.plot3D([0,0], [-4,4], [0,0], color=(0.5,0.5,0.5))
  ax2.plot3D([0,0], [0,0], [0,6.5], color=(0.5,0.5,0.5))
  ax2.set_title('Imagen de $f$')
  ax2.set_xlabel('$x$')
  ax2.set_ylabel('$y$')
  ax2.set_zlabel('$z$')
  #ax2.view_init(30, 30)

  # fig1
  ax1.plot([0,4], [0,0], color=(0.5,0.5,0.5))
  ax1.plot([0,0], [0,2*np.pi], color=(0.5,0.5,0.5))
  ax1.plot([u[iu], u[iu]], [v[0], v[-1]], 'r')
  ax1.plot([u[0], u[-1]], [v[iv], v[iv]], 'm')
  ax1.set_title('Dominio de $f$')
  ax1.set_xlabel('$u$')
  ax1.set_ylabel('$v$')
  plot = ax1.axis('equal');
  return plot

#plt.savefig('ejm_helicoide.pdf')

anim = animation.FuncAnimation(fig, animacion, frames=range(50), repeat = False)

anim
```

Por definición, el dominio es la caja $\operatorname{dom}(f) = [0, 4]\times [0, 2\pi]$ y claramente la imagen es $\operatorname{img}(f) = \{(u\cos(v), u\operatorname{sen}(v), v)\in\mathbb{R}^3 \mid 0\leq u \leq 4, \, 0 \leq v \leq 2 \pi  \}$. Esta última es una superficie conocida como **helicoide circular**, que en este caso $f$ define paramétricamente.
