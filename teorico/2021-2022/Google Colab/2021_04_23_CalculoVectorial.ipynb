# Funciones de varias variables

## Funciones diferenciables

Recuerde que una función $f:\mathbb{R}^n \to \mathbb{R}^m$ es diferenciable en $\bar{x}$ si
\begin{equation*}
    f(x) = f(\bar{x}) + Df(\bar{x})(x - \bar{x}) + o(\| x - \bar{x} \|),    
\end{equation*}
para $x$ suficientemente cerca de $\bar{x}$.



Una condición suficiente que nos garantiza que la función es diferenciable es la siguiente.

>**Teorema**
>
> Sea $f:\mathbb{R}^n \to \mathbb{R}^m$. Si las derivadas parciales $\frac{\partial f_i}{\partial x_j}$ son continuas en un entorno de $\bar{x}$, entonces $f$ es diferenciable en $\bar{x}$

**Demostración**

Consideremos la función lineal $Df(\bar{x})$, o sea, aquella cuya representación matricial es la matriz Jacobiana $f'(\bar{x})$. Veamos que
\begin{equation*}
    \lim_{x \to \bar{x}} \frac{f(x) - f(\bar{x}) - Df(\bar{x})(x-\bar{x})}{\|x-\bar{x}\|} = 0.    
\end{equation*}
Como el límite existe si y sólo si existe en cada coordenada, sin pérdida de generalidad podemos asumir que $f:\mathbb{R}^n \to \mathbb{R}$.

Supongamos que las derivadas parciales $\frac{\partial f}{\partial x_j}$ son continuas en $B_\varepsilon(\bar{x})$ y sea $x \in B_\varepsilon(\bar{x})$. Definamos los puntos auxiliares
$$
y^k = (x_1, \ldots, x_k, \bar{x}_{k+1}, \ldots, \bar{x}_n), \qquad k=0, \ldots, n.
$$
Con esto, las primeras $k$ coordenadas de $y^k$ coinciden con $x$ y las restantes con $\bar{x}$. De esta manera $y^0=\bar{x}$, $y^n = x$ y siempre tendremos $y^k \in B_\varepsilon(\bar{x})$. Además,
$$
    f(x) - f(\bar{x}) = \sum_{k=1}^n f(y^{k}) - f(y^{k-1}).
$$
Note que $y^k$ e $y^{k-1}$ difieren sólo en la coordenada $k$, y para la función $\varphi:\mathbb{R} \to \mathbb{R}$ tal que   $\varphi(t) = f(x_1, \ldots, x_{k-1}, t, \bar{x}_{k+1}, \ldots, \bar{x}_n)$ tenemos
$$
f(y^k) - f(y^{k-1}) = \varphi(x_k) - \varphi(\bar{x}_k)
                    = \varphi'(\bar{x}_k + \xi_k (x_k - \bar{x}_k) )(x_k - \bar{x}_k)
                    = \frac{\partial f}{\partial x_k}(z^k) (x_k - \bar{x}_k),
$$
donde $\xi_k\in [0,1]$ y $z^k = y^{k-1} + \xi_k (y^k - y^{k-1})$. Luego $z^k \in B_\varepsilon(\bar{x})$ y
$$
    f(x) - f(\bar{x}) = \sum_{k=1}^n \frac{\partial f}{\partial x_k}(z^k) (x_k - \bar{x}_k).
$$
Por otro lado,
$$
    Df(\bar{x})(x-\bar{x}) = \sum_{k=1}^n (x_k - \bar{x}_k) Df(\bar{x})(e^k)
                            = \sum_{k=1}^n \frac{\partial f}{\partial x_k}(\bar{x}) (x_k - \bar{x}_k).
$$
Entonces
\begin{align*}
    | f(x) - f(\bar{x}) - Df(\bar{x})(x-\bar{x})|
    & = \left|\sum_{k=1}^n \left(\frac{\partial f}{\partial x_k}(z^k) - \frac{\partial f}{\partial x_k}(\bar{x}) \right)(x_k - \bar{x}_k)  \right| \\
    & \leq \sum_{k=1}^n \left|\frac{\partial f}{\partial x_k}(z^k) - \frac{\partial f}{\partial x_k}(\bar{x}) \right| \|x - \bar{x}\|,
\end{align*}
donde usamos que $|x_k - \bar{x}_k| \leq \|x - \bar{x}\|$. Como $z^k$ tiende a $\bar{x}$ cuando $x$ tiende a $\bar{x}$ y $\frac{\partial f}{\partial x_k}$ es continua para todo $k$, obtenemos el límite deseado. $\spadesuit$




Note que este resultado es válido no sólo para $\bar{x}$ sino también para todos los puntos en un entorno de $\bar{x}$, obteniendo que $f$ es diferenciable en todos los puntos de un entorno de $\bar{x}$. Además, pensando a la derivada como una función $Df:\mathbb{R}^n \to \mathbb{R}^{mn}$ tal que $Df(x) = \left(\frac{\partial f_1}{\partial x_!}(x), \ldots, \frac{\partial f_m}{\partial x_n}(x) \right)$, obtenemos que la derivada es continua. Una función cuya derivada es continua es llamada **continuamente diferenciable**, o de clase $C^1$.
De esta manera, el resultado anterior simplemente nos dice que una función continuamente diferenciable debe ser diferenciable.



Veamos cómo utilizar este resultado.

#### Ejemplo
Considere la función $f:\mathbb{R}^2 \to \mathbb{R}$ tal que
$$
    f(x)= \frac{\cos(x_1) + e^{x_1 x_2}}{x_1^2 + x_2^2}.
$$
Claramente $\operatorname{dom}(f) = \mathbb{R}^2 \setminus\{ 0\}$ y para $x$ en el dominio
\begin{align*}
    \frac{\partial f}{\partial x_1}(x) &= \frac{(x_2 e^{x_1 x_2} - \operatorname{sen}(x_1))(x_1^2 + x_2^2) - 2 x_1 (\cos(x_1) + e^{x_1 x_2})}{(x_1^2 + x_2^2)^2},\\
    \frac{\partial f}{\partial x_2}(x) &= \frac{x_1 e^{x_1 x_2}(x_1^2 + x_2^2) - 2 x_2 (\cos(x_1) + e^{x_1 x_2})}{(x_1^2 + x_2^2)^2}.
\end{align*}
Como estas funciones son continuas en $\operatorname{dom}(f)$, entonces $f$ es diferenciable en $\operatorname{dom}(f)$.

De la definición de diferenciabilidad se pueden deducir las siguientes propiedades.

>**Proposición**
>
>1. Si $f:\mathbb{R}^n \to \mathbb{R}^m$ es diferenciable en $\bar{x}$ y $\alpha \in \mathbb{R}$, entonces $\alpha f$ es diferenciable en $\bar{x}$ y
$$
                    D(\alpha f) (\bar{x}) = \alpha Df(\bar{x}).
$$
2. Si $f:\mathbb{R}^n \to \mathbb{R}^m$ y $g:\mathbb{R}^n \to \mathbb{R}^m$ son diferenciables en $\bar{x}$, entonces $f+g$ es diferenciable en $\bar{x}$ y
$$
                D(f+g)(\bar{x}) = Df(\bar{x}) + Dg(\bar{x}).
$$
3. Si $f:\mathbb{R}^n \to \mathbb{R}$ y $g:\mathbb{R}^n \to \mathbb{R}$ son diferenciables en $\bar{x}$, entonces $fg$ es diferenciable en $\bar{x}$ y
$$
               D(fg)(\bar{x}) = g(\bar{x}) Df(\bar{x}) + f(\bar{x}) Dg(\bar{x}).
$$
4. Si $f:\mathbb{R}^n \to \mathbb{R}$ y $g:\mathbb{R}^n \to \mathbb{R}$ son diferenciables en $\bar{x}$ con $g(x)\neq 0$ para $x$ en un entorno de $\bar{x}$, entonces $f/g$ es diferenciable en $\bar{x}$ y
$$
                D(f/g)(\bar{x}) = \frac{g(\bar{x}) Df(\bar{x}) - f(\bar{x}) Dg(\bar{x})}{g(\bar{x})^2}.
$$

**Demostración**

El caso del producto escala y de la suma de funciones se demuestran fácilmente. Veamos el producto de funciones. Sea $h=fg$ y como $f$ y $g$ son diferenciables en $\bar{x}$ tenemos que para $x$ en un entorno de $\bar{x}$, $f(x) = f(\bar{x}) + Df(\bar{x}) (x-\bar{x}) + o(\|x-\bar{x}\|)$ y $g(x) = g(\bar{x}) + Dg(\bar{x}) (x-\bar{x}) + o(\|x-\bar{x}\|)$, luego
\begin{align*}
    h(x) - h(\bar{x})  
    &= g(\bar{x}) \bigl(f(x) - f(\bar{x}) \bigr) + f(x) \bigl( g(x) - g(\bar{x}) \bigr) \\
    &= g(\bar{x}) \bigl(f(x) - f(\bar{x}) \bigr) + f(\bar{x}) \bigl( g(x) - g(\bar{x}) \bigr) + o(\|x-\bar{x}\|) \\
    &= g(\bar{x}) Df(\bar{x}) (x-\bar{x}) + f(\bar{x}) Dg(\bar{x}) (x-\bar{x}) + o(\|x-\bar{x}\|),
\end{align*}
donde en la segunda igualdad usamos que $\bigl(f(x) - f(\bar{x}) \bigr)\bigl(g(x) - g(\bar{x}) \bigr) = O(\| x- \bar{x}\|^2) = o(\| x- \bar{x}\|)$, en la tercera igualdad usamos que $\alpha o(\| x- \bar{x}\|) = o(\| x- \bar{x}\|)$ y $o(\| x- \bar{x}\|) + o(\| x- \bar{x}\|) = o(\| x- \bar{x}\|)$. Como $v \mapsto g(\bar{x}) Df(\bar{x}) (v) + f(\bar{x}) Dg(\bar{x}) (v)$ es lineal, obtenemos que debe ser $Dh(\bar{x})$.

Para el cociente, basta demostrar para $h=1/g$ y utilizar lo anterior. Como
\begin{align*}
    h(x) - h(\bar{x}) &= \frac{1}{g(x)} - \frac{1}{g(\bar{x})} \\
    &= \frac{g(\bar{x}) - g(x)}{g(\bar{x})^2} + \frac{\bigl( g(\bar{x}) - g(x)\bigr)^2}{g(x) g(\bar{x})} \\
    &= \frac{-Dg(\bar{x})(x - \bar{x})}{g(\bar{x})^2} + o(\| x- \bar{x}\|),
\end{align*}
donde en la última igualdad usamos que $\bigl( g(\bar{x}) - g(x)\bigr)^2 = O(\|x-\bar{x}\|^2) = o(\|x-\bar{x}\|)$. Como $v \mapsto -\frac{1}{g(\bar{x})^2} Dg(\bar{x})(v)$ es lineal, entonces $Dh(\bar{x}) = -\frac{1}{g(\bar{x})^2} Dg(\bar{x})$. $\spadesuit$



Veamos cómo usar estas propiedades para simplificar cuentas.

#### Ejemplo

Considere la función $h:\mathbb{R}^3 \to \mathbb{R}$ tal que
$$
    h(x) = \frac{x_1^2 + x_2^2 + x_3^2}{x_1^2 + 1}.
$$
La matriz Jacobiana es
\begin{align*}
    h'(x) &= \left[ \frac{\partial h}{\partial x_1}(x) \quad \frac{\partial h}{\partial x_2}(x) \quad \frac{\partial h}{\partial x_3}(x) \right] \\
    &= \left[\frac{2x_1(x_1^2 +1) - 2x_1(x_1^2 + x_2^2 + x_3^2)}{(x_1^2 + 1)^2} \quad
      \frac{2 x_2}{x_1^2 + 1} \quad \frac{2 x_3}{x_1^2 + 1} \right].
\end{align*}

Considerando $h=f/g$ tenemos que
\begin{align*}
    h'(x) &= \frac{g(x) f'(x) - f(x) g'(x)}{g(x)^2} \\
    &= \frac{(x_1^2 + 1) [2x_1 \quad 2x_2 \quad 2x_3] - (x_1^2 + x_2^2 + x_3^2) [2x_1 \quad 0 \quad 0]}{(x_1^2 + 1)^2}.
\end{align*}

Otra propiedad importante es la conocida como **regla de la cadena**.

>**Teorema**
>
> Sean $f:\mathbb{R}^n \to \mathbb{R}^m$ y $g:\mathbb{R}^m \to \mathbb{R}^p$ tal que $\operatorname{img}(f) \subset \operatorname{dom}(g)$. Si $f$ es diferenciable en $\bar{x}$ y $g$ es diferenciable en $f(\bar{x})$, entonces $g \circ f$ es diferenciable en $\bar{x}$ y
$$
    D(g \circ f) (\bar{x}) = Dg(f(\bar{x})) \circ Df(\bar{x}).
$$
La representación matricial es $(g \circ f)'(\bar{x}) = g'(f(\bar{x})) f'(\bar{x})$, donde $f'(\bar{x}) \in \mathbb{R}^{m \times n}$ y  $g'(f(\bar{x})) \in \mathbb{R}^{p \times m}$.


**Demostración**

Sabemos que $f(x) = f(\bar{x}) + Df(\bar{x}) (x-\bar{x}) + o(\|x-\bar{x}\|)$ para $x$ en un entorno de $\bar{x}$ y $g(y) = g(f(\bar{x})) + Dg(f(\bar{x})) (y-f(\bar{x})) + o(\|y-f(\bar{x})\|)$ para $y$ en un entorno de $f(\bar{x})$. Como
$$
    \frac{o(\|f(x)-f(\bar{x})\|)}{\|x-\bar{x}\|}
    = \frac{o(\|f(x)-f(\bar{x})\|)}{\|f(x)-f(\bar{x})\|} \frac{\|f(x)-f(\bar{x})\|}{\|x-\bar{x}\|} \to 0,
$$
cuando $x \to \bar{x}$, tenemos que $o(\|f(x)-f(\bar{x})\|) = o(\|x-\bar{x}\|)$. Entonces
\begin{align*}
    g(f(x)) &= g(f(\bar{x})) + Dg(f(\bar{x})) (f(x)-f(\bar{x})) + o(\|f(x)-f(\bar{x})\|) \\
    &= g(f(\bar{x})) + Dg(f(\bar{x})) \bigl( Df(\bar{x}) (x-\bar{x}) \bigr) + o(\|x-\bar{x}\|),
\end{align*}
donde en la última igualdad usamos que $Dg(f(\bar{x}))\bigl( o(\|x-\bar{x}\|)\bigr) = o(\|x-\bar{x}\|)$.

Como ya se vio, la representación matricial de composición de funciones lineales resulta en el producto de las matrices asociadas a cada función lineal. $\spadesuit$



Veamos un ejemplo concreto de tal propiedad.

#### Ejemplo

Considere $h:\mathbb{R}^3 \to \mathbb{R}$ tal que $h(x) = (x_1^2 x_2)^2 + x_2^4 - e^{x_1 x_3}$. Su matriz Jacobiana es
$$
 h'(x) = [ 4 x_1^3 x_2^2 - x_3 e^{x_1 x_3} \quad 2x_1^4 x_2 + 4x_2^3 \quad -x_1 e^{x_1 x_3} ].
$$
Si consideramos $h = g \circ f$ con $f:\mathbb{R}^3 \to \mathbb{R}^3$ y $g:\mathbb{R}^3 \to \mathbb{R}$ tal que $f(x) = (x_1^2 x_2, x_2^2, e^{x_1 x_3})$ y $g(y) = y_1^2 + y_2^2 - y_3$ tenemos que
$$
 f'(x) = \left[\begin{array}{ccc}
                2x_1x_2& x_1^2 & 0 \\
                    0  & 2x_2  & 0 \\
        x_3 e^{x_1 x_3}&  0    & x_1 e^{x_1 x_3}
 \end{array} \right],
 \qquad
 g'(y) = [2y_1 \quad 2y_2 \quad -1].
$$
Por lo tanto
\begin{align*}
    h'(x) &= g'(f(x)) f'(x) \\
    &= [2x_1^2 x_2 \quad 2x_2^2 \quad -1]
    \left[\begin{array}{ccc}
                2x_1x_2& x_1^2 & 0 \\
                    0  & 2x_2  & 0 \\
        x_3 e^{x_1 x_3}&  0    & x_1 e^{x_1 x_3}
 \end{array} \right].
\end{align*}

Note que si asociamos $\mathbb{R}^3$ con los vectores columna $\mathbb{R}^{3 \times 1}$ tendremos que
\begin{align*}
    \nabla h(x) &= h'(x)^T =  \bigl( g'(f(x)) f'(x) \bigr)^T \\
    & = f'(x)^T g'(f(x))^T = f'(x)^T \nabla g(f(x)) \\
    &= \left[\begin{array}{ccc}
                2x_1x_2& 0 & x_3 e^{x_1 x_3} \\
                x_1^2  & 2x_2  & 0 \\
                0&  0    & x_1 e^{x_1 x_3}
 \end{array} \right] \left[\begin{array}{c}    2x_1^2 x_2 \\ 2x_2^2 \\ -1 \end{array} \right].
\end{align*}

Para realizar cuentas de manera rápida, un artificio muy utilizado consiste en usar la misma letra para una variable y una función. Con este abuso de notación el ejemplo anterior puede escribirse como $h=g(y_1, y_2, y_3)$, con  $y_1=x_1^2 x_2$, $y_2=x_2^2$, $y_3= e^{x_1 x_3}$. Con esto se puede escribir
$$
    \frac{\partial h}{\partial x_1}
    = \frac{\partial g}{\partial y_1} \frac{\partial y_1}{\partial x_1}
        + \frac{\partial g}{\partial y_2} \frac{\partial y_2}{\partial x_1}
        + \frac{\partial g}{\partial y_3} \frac{\partial y_3}{\partial x_1},
$$
que luego de calcular y evaluar donde corresponde coincide con lo obtenido anteriormente via producto matricial. En este curso no utilizaremos esta notación ya que es ambigua e imprecisa, pues denota funciones y variables con la misma letra y no especifica en qué punto deben evaluarse las funciones.


Si $f:\mathbb{R}^n \to \mathbb{R}$ y queremos analizar la tasa de cambio de $f$ a partir de un punto $\bar{x}$ a lo largo de una trayectoria $\gamma:\mathbb{R} \to \mathbb{R}^n$, con $\gamma(0) = \bar{x}$ y $\gamma'(0)=v$, esto se traduce en estudiar el comportamiento de la función $f \circ \gamma$ en el origen. Suponiendo que las funciones son diferenciables, la regla de la cadena nos garantiza que
$$
    (f \circ \gamma)'(0) = f'(\gamma(0)) \gamma'(0)
            = f'(\bar{x}) v
            = \langle \nabla f(\bar{x}), v \rangle.
$$
De esta manera podemos determinar, por ejemplo, la dirección $v$ que debe tomar la trayectoria para incrementar o disminuir el valor $f(\bar{x})$.

>**Proposición**
>
> Si $f:\mathbb{R}^n \to \mathbb{R}$ y $\nabla f(\bar{x}) \neq 0$, entonces $\nabla f(\bar{x})$  y $-\nabla f(\bar{x})$ apuntan en la dirección de máximo crecimiento y decrecimiento de $f$ a partir de $\bar{x}$, respectivamente.

**Demostración**

Como estamos interesados en la dirección y no en la magnitud, veamos cuál es la dirección $v$ con $\|v\|=1$ tal que $\langle \nabla f(\bar{x}), v \rangle $ alcanza su valor máximo y mínimo.

Por la desigualdad de Cauchy-Schwarz, $|\langle \nabla f(\bar{x}), v \rangle | \leq \| \nabla f(\bar{x})\|$. Por lo tanto, los valores máximo y mínimo que puede alcanzar el producto interno son
$$
    \langle \nabla f(\bar{x}), v \rangle = \| \nabla f(\bar{x})\|
    \quad \textrm{ y } \quad
    \langle \nabla f(\bar{x}), v \rangle = - \| \nabla f(\bar{x})\|.
$$
Como sabemos, por Cauchy-Schwarz, estas igualdades se alcanzan si y sólo si $v = \alpha \nabla f(\bar{x})$. Obteniendo que los valores máximo y mínimo se alcanzan respectivamente en la dirección
$$
    v=\frac{1}{\| \nabla f(\bar{x}))\|} \nabla f(\bar{x})
    \quad \textrm{ y } \quad
    v= - \frac{1}{\| \nabla f(\bar{x}))\|} \nabla f(\bar{x}). \spadesuit
$$


Además, utilizando la composición de una función con una trayectoria y la regla de la cadena, podemos obtener una extensión del Teorema del valor medio para funciones a valores escalares.

>**Teorema**
>
> Si $f:\mathbb{R}^n \to \mathbb{R}$ es diferenciable en un abierto que contiene al segmento que une $x$ y $\bar{x}$ de $\mathbb{R}^n$, entonces existe $\hat{x}$ en ese segmento tal que
$$
    f(x) - f(\bar{x}) = f'(\hat{x}) (x - \bar{x}).
$$

**Demostración**

Sea $\gamma:\mathbb{R} \to \mathbb{R}^n$ tal que $\gamma(t) = \bar{x} + t (x-\bar{x})$ y $g:\mathbb{R} \to \mathbb{R}$ tal que $g=f \circ \gamma$. Por el Teorema del valor medio para funciones reales, tenemos que existe $\tau \in (0,1)$ tal que
\begin{align*}
    f(x) - f(\bar{x}) &= g(1) - g(0) = g'(\tau) \\
                    &= f'(\gamma(\tau)) \gamma'(\tau)
                    = f'(\hat{x}) (x - \bar{x}),
\end{align*}
donde $\hat{x}  = \bar{x} + \tau (x-\bar{x})$. $\spadesuit$




En el resultado anterior utilizamos una trayectoria formada por el segmento que une dos puntos. Cuando la trayectoria está formada por la unión finita de segmentos, se dice que es una trayectoria poligonal. Con esto, un conjunto $S$ se dice poligonalmente conexo si dos puntos cualquiera puden unirse mediante una trayectoria poligonal en $S$. Para este tipo de conjuntos obtenemos el siguiente resultado.

>**Teorema**
>
> Si $f:\mathbb{R}^n \to \mathbb{R}^m$ es diferenciable en un abierto poligonalmente conexo $U$ y $f'(x)=0$ para todo $x\in U$, entonces $f$ es constante en $U$.


**Demostración**

Para ver que $f$ es constante veremos que cada función coordenada es constante. Sea $f_i$ la $i$-ésima función coordenada y tome dos puntos arbitrarios $x,\bar{x} \in U$. Como $U$ es poligonalmente conexo, existen $x^0, \ldots, x^p \in U$ con $x^0=\bar{x}$, $x^p=x$ y tal que el segmento de $x^{k-1}$ a $x^{k}$ está contenido en $U$. Luego, por el Teorema anterior existe $\hat{x}^k \in U$ tal que $f_i(x^k) - f_i(x^{k-1}) = f_i'(\hat{x}^k) (x^k - x^{k-1}) = 0$. Por lo tanto
$$
    f_i(x) - f_i(\bar{x}) = f_i(x^p) - f_i(x^0) = \sum_{k=1}^p f_i(x^k) - f_i(x^{k-1}) = 0.
$$
Obteniendo así que $f_i$ es constante en $U$. $\spadesuit$



### Derivadas de orden superior

Como ya vimos, si $f:\mathbb{R}^n \to \mathbb{R}^m$ y sus derivadas parciales $\frac{\partial f_i}{\partial x_j}$ son continuas entonces $f$ es continuamente diferenciable, o de clase $C^1$. Si estas derivadas parciales tienen derivadas parciales continuas, decimos que $f$ es dos veces continuamente diferenciable, o de clase $C^2$. Repitiendo esta lógica se definen todas las derivadas de orden superior. Cuando una función tiene derivada de todos los órdenes, decimos que $f$ es infinitamente diferenciable, o de clase $C^\infty$.

Por ejemplo, si $f:\mathbb{R}^2 \to \mathbb{R}$,
$$
    \frac{\partial^2 f}{\partial x_1^2}(x) = \frac{\partial}{\partial x_1}\left( \frac{\partial f}{\partial x_1}\right)(x), \quad
    \frac{\partial^2 f}{\partial x_1 \partial x_2}(x) = \frac{\partial}{\partial x_1}\left( \frac{\partial f}{\partial x_2}\right)(x), \quad
    \frac{\partial^2 f}{\partial x_2 \partial x_1}(x) = \frac{\partial}{\partial x_2}\left( \frac{\partial f}{\partial x_1}\right)(x).
$$
Estas derivadas se conocen como **derivadas parciales iteradas**. Cuando la derivación se realiza en dos o más variables distintas suelen llamarlas **derivadas parciales mixtas**.
Algunos autores utilizan $f_{x_j}$ para denotar $\frac{\partial f}{\partial x_j}$ y de esta manera $f_{x_j x_k} = (f_{x_j})_{x_k}$, con esta notación
$$
f_{x_j x_k} = \frac{\partial^2 f}{\partial x_k \partial x_j}.
$$
Se puede probar que si $\frac{\partial^2 f}{\partial x_1 \partial x_2}$ y $\frac{\partial^2 f}{\partial x_2 \partial x_1}$ son continuas en un abierto $U$ entonces
$$
\frac{\partial^2 f}{\partial x_1 \partial x_2} (x) = \frac{\partial^2 f}{\partial x_2 \partial x_1} (x),
$$
para todo $x\in U$.
