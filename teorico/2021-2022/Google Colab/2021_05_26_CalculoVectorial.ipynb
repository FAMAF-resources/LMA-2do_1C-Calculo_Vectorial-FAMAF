## Integrales sobre superficies

Luego de haber estudiado lo relacionado a medir curvas, veamos cómo extender alguna de esas nociones para superficies. Recordemos que si $\mathcal{C}$ es una curva simple curva parametrizada por $\gamma:[a,b] \to \mathbb{R}^n$, su longitud es
$$
    \mathscr{l}(\mathcal{C}) = \int_a^b \|\gamma'(t)\| dt.
$$
Note que la manera de medir se basa en sumar para cada punto la medida unidimensional del vector tangente, o sea, la longitud de $\gamma'(t)$ es la medida del tangente a $\mathcal{C}$ en el punto $x=\gamma(t)$.

Suponga ahora que tenemos una superficie regular $\mathcal{S}$ parametrizada por una función $\phi:U \subset \mathbb{R}^2 \to \mathbb{R}^3$, o sea, $\mathcal{S}=\{\phi(u) \mid u \in U\}$.
El plano tangente a $\mathcal{S}$ en un punto $x=\phi(u)$ es el paralelogramo formado por los vectores $\frac{\partial \phi}{\partial u_1}(u)$ y $\frac{\partial \phi}{\partial u_2}(u)$. Además, el área de este paralelogramo es
$$
    \left\| {\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)}\right\|.
$$


De manera análoga a lo realizado para curvas, si sumamos estas cantidades obtendremos la medida bidimensional de esta superficie.

> **Definición**
>
> Si $\mathcal{S}$ es una superficie regular parametrizada por $\phi: U \subset \mathbb{R}^2 \to \mathbb{R}^3$ y la frontera de $U$ está contenida en una cantidad finita de conjuntos suaves, diremos que el **área** de $\mathcal{S}$ es
$$
  \mathscr{a}(\mathcal{S}) = \int_U \left\| {\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)}\right\| du.
$$
Si $\mathcal{S}$ es la unión finita de superficies no solapadas $\mathcal{S}_i$, entonces su área es la suma de las áreas de cada $\mathcal{S}_i$.


Veamos cómo usar esta definición.

#### Ejemplo

Suponga que $\mathcal{S}$ es un toro donde $R$ es la distancia del centro del toro al centro del tubo y $r$ es el radio del tubo, o sea $R > r$. Esta superficie puede parametrizarse mediante
$$
 \phi(u) = \bigl( (R+r\cos(u_1)) \cos(u_2), (R+r\cos(u_1)) \operatorname{sen}(u_2), r \operatorname{sen}(u_1) \bigr),
$$
con $U=\{u\in\mathbb{R}^2 \mid 0 < u_1 < 2\pi, 0< u_2 < 2\pi \}$.


```
#@title Toro { display-mode: "form" }
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

fig = plt.figure()
ax = fig.add_subplot(1,1,1,projection='3d')

u = np.linspace(0, 2*np.pi, 50)
v = np.linspace(0, 2*np.pi, 50)
R = 3
r = 1
x = np.outer(R + r*np.cos(u), np.cos(u))
y = np.outer(R + r*np.cos(u), np.sin(u))
z = np.outer(r*np.sin(u), np.ones(u.shape))
# ejes
ax.plot3D([0,0],[-4.1,4.1], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([-4.1,4.1], [0,0], [0,0], color=(0.5,0.5,0.5))
ax.plot3D([0,0], [0,0], [-4.1,4.1], color=(0.5,0.5,0.5))
# toro
ax.plot_surface(x,y,z, alpha=0.7)
ax.axis('off');
#plt.savefig('ejm_toro.pdf')
```


Se puede ver que $\phi$ es inyectiva y mapea $\operatorname{cl}(U)$ en $\mathcal{S}$. Además, la frontera de $U$ tiene medida nula en $\mathbb{R}^2$. Por otro lado
\begin{align*}
  \frac{\partial \phi}{\partial u_1}(u) &= \bigl( -r \operatorname{sen}(u_1) \cos(u_2), -r \operatorname{sen}(u_1) \operatorname{sen}(u_2), r \cos(u_1) \bigr), \\
	\frac{\partial \phi}{\partial u_2}(u) &= \bigl( -(R+r\cos(u_1)) \operatorname{sen}(u_2), (R+r\cos(u_1)) \cos(u_2), 0 \bigr), \\
{\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)} &= - r(R+r\cos(u_1)) \bigl( \cos(u_1) \cos(u_2),  \cos(u_1) \operatorname{sen}(u_2) ,   \operatorname{sen}(u_1) \bigr).
\end{align*}
Por lo tanto,
\begin{align*}
 \mathscr{a}(\mathcal{S}) &= \int_U \left\|{\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)} \right\| du \\
	&= \int_0^{2\pi} \int_0^{2 \pi} r (R+r\cos(u_1)) du_2 du_1 \\
	&= 2\pi r  \left[ R u_1 + r\operatorname{sen}(u_1) \right]_0^{2\pi} \\
	&= 4 \pi^2 R r.
\end{align*}

---


Al estudiar superficies vimos que si $\mathcal{S} = \operatorname{img}(\phi)$ entonces $\frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)$ es el vector normal a $\mathcal{S}$ en $p=\phi(u)$. Por lo tanto, el elemento de área en $p$ viene dado por la norma del vector normal a $\mathcal{S}$ en $p$.

Como yá vimos, cuando la superficie está definida explícitamente, o sea $\mathcal{S} = \operatorname{gr}(f)$ para $f: D \subset \mathbb{R}^2 \to \mathbb{R}$, el vector normal a $\mathcal{S}$ en $p = (x,f(x))$ puede expresarse como
$$
    \left( -\frac{\partial f}{\partial x_1}(x), -\frac{\partial f}{\partial x_2}(x), 1 \right).
$$
Obteniendo que si $\mathcal{S} = \operatorname{gr}(f)$, entonces
$$
    \mathscr{a}(\mathcal{S}) = \int_D \sqrt{\textstyle \bigl(\frac{\partial f}{\partial x_1}(x) \bigr)^2 + \bigl(\frac{\partial f}{\partial x_2}(x) \bigr)^2 + 1} \, dx.
$$


Tomando las ideas vistas para integrales múltiples, podemos pensar la medida bidimensional $\mathscr{a}(\mathcal{S})$ como la integral de una función constantemente igual a uno.
$$
\int_U 1\, \left\| {\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)}\right\| du.
$$




Veamos cómo extender eso a cualquier función a valores escalares.

>**Definición**
>
> Si $f:\mathbb{R}^3 \to \mathbb{R}$ está definida en la superficie regular $\mathcal{S}$, parametrizada por $\phi: U \subset \mathbb{R}^2 \to \mathbb{R}^3$, la **integral de $f$ sobre $\mathcal{S}$** es
$$
  \int_U f(\phi(u)) \left\| {\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)}\right\| du.
$$


Para mantener la mantener la coherencia con lo visto para curvas, denotaremos a la integral de $f$ sobre $\mathcal{S}$ por
$$
\int_{\phi} f(x) dS,
$$
ya que la integral depende de la parametrización $\phi$. Veremos luego que esta integral es independiente de la parametrización.




#### Ejemplo

Se desea calcular $\int_{\phi} f(x) dS$ donde $\mathcal{S}$ es el helicoide definido paramétricamente por $\phi: U \subset \mathbb{R}^2 \to \mathbb{R}^3$ tal que
$$
\phi(r,\theta) = \bigl( r\cos(\theta), r\operatorname{sen}(\theta), r \bigr),\quad U=[0,1] \times [0, 2 \pi],
$$
y  $f:\mathbb{R}^3 \to \mathbb{R}$ tal que $f(x) = \sqrt{x_1^2 + x_2^2 + 1}$.

Como
\begin{align*}
    \frac{\partial \phi}{\partial r}(r,\theta) &= \bigl( \cos(\theta), \operatorname{sen}(\theta), 1\bigr), \\
    \frac{\partial \phi}{\partial \theta}(r,\theta) &= \bigl( -r \operatorname{sen}(\theta), r \cos(\theta), 0\bigr), \\
    {\textstyle \frac{\partial \phi}{\partial r}(r,\theta) \times \frac{\partial \phi}{\partial \theta}(r,\theta)} &= \bigl( -r \cos(\theta), -r \operatorname{sen}(\theta), r  \bigr),
\end{align*}
entonces
\begin{align*}
    \int_{\mathcal{S}} f(x) dS &= \int_U f(\phi(r,\theta)) \| {\textstyle \frac{\partial \phi}{\partial r}(r,\theta) \times \frac{\partial \phi}{\partial \theta}(r,\theta)} \| dr d\theta \\
     &=\int_0^{2\pi} \int_0^1 \sqrt{r^2 +1} \sqrt{2} r dr d\theta
     =\sqrt{2} 2 \pi \int_0^1 \sqrt{r^2 +1}  r dr \\
     &=2^{3/2} \pi \left[{\textstyle \frac{1}{3}} (r^2 +1)^{3/2} \right]_0^1
     = \frac{2^{3/2}}{3} (2^{3/2} - 1)\pi .
\end{align*}

Suponga que un fluido atraviesa una membrana permeable y se desea medir la cantidad de fluido que atraviesa la membrana. El movimiento del fluido puede expresarse por medio de una función $F:\mathbb{R}^3 \to \mathbb{R}^3$ y la membrana por una superficie $\mathcal{S}$. Si la superficie está parametrizada por una función $\phi$, tenemos que en cada punto $x=\phi(u)$ de la superficie, el flujo saliente va a ser la componente del fluido en la dirección normal a la superficie, o sea
$$
\langle F(x), N(x) \rangle = \left\langle F(\phi(u)), \frac{\frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)}{\bigl\| \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)\bigr\|}\right\rangle.
$$
Por lo tanto, sumando los flujos salientes para cada punto de la superficie, obtenemos que el flujo del fluido sobre $\mathcal{S}$ es
$$
\int_U \left\langle F(\phi(u)), N(u) \right\rangle {\textstyle \bigl\| \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)\bigr\|} du = \int_U \left\langle F(\phi(u)), {\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)}  \right\rangle du .
$$



> **Definición**
>
> Si $F:\mathbb{R}^3 \to \mathbb{R}^3$ está definida en la superficie regular $\mathcal{S}$, parametrizada por $\phi: U \subset \mathbb{R}^2 \to \mathbb{R}^3$, la **integral de superficie** de  $F$ sobre $\mathcal{S}$ es
$$
  \int_U \langle F(\phi(u)) , {\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)} \rangle du.
$$


Por analogía con la integral anterior, una notación coherente sería
$$
    \int_{\phi} F(x) dS \quad \textrm{ o } \quad \int_{\phi} \langle F(x), N(x) \rangle dS,
$$
ya que la integral de superficie tiene involucrada en su definición la parametrización $\phi$. Cuando trabajamos con integrales de línea vimos que para tener una definición independiente de la parametrización fue necesario hablar de orientación. En este caso, para que la integral de superficie dependa de la superficie independientemente de su parametrización, deberemos dar una noción de orientación.




De la definición de superficie, para $\bar{p}\in \mathcal{S}$ tenemos que $\bar{p} = \phi(\bar{u})$ para una parametrización $\phi: U \to V \cap \mathcal{S}$ continua, inyectiva y sobreyectiva, donde $U \subset \mathbb{R}^2$ y $V \subset \mathbb{R}^3$ son abiertos. Suponga que tenemos otra parametrización $\psi: W \to V \cap \mathcal{S}$ continua, inyectiva y sobreyectiva, con $W \subset \mathbb{R}^2$ abierto y $\bar{p} = \psi(\bar{w})$. Si la superficie es regular tenemos que el vector normal a $\bar{p}$ dado por cada parametrización es
$$
\frac{\partial \phi}{\partial u_1}(\bar{u}) \times \frac{\partial \phi}{\partial u_2}(\bar{u}) ,
\quad \textrm{ y } \quad
\frac{\partial \psi}{\partial w_1}(\bar{w}) \times \frac{\partial \psi}{\partial w_2}(\bar{w}).
$$
Como la recta normal tiene una única dirección, uno de estos vectores debe ser múltiplo del otro y cuando ese escalar es positivo ambas parametrizaciones inducen la misma orientación, cuando es negativo la orientación se invierte.

Para obtener ese escalar, note que de las propiedades de $\phi$ y $\psi$ podemos garantizar la existencia de un cambio de coordenada $\xi: W \to U$ continua, inyectiva y sobreyectiva tal que
$$
    \psi(w) = \phi(\xi(w)).
$$
Como $\psi'(w) = \phi'(\xi(w)) \xi'(w)$ y $\bar{u} = \xi(\bar{w})$, entonces
$$
\psi'(\bar{w}) = \phi'(\bar{u}) \xi'(\bar{w})
= \left[\begin{array}{cc}
    \frac{\partial \phi_1}{\partial u_1}(\bar{u}) & \frac{\partial \phi_1}{\partial u_2}(\bar{u}) \\
    \frac{\partial \phi_2}{\partial u_1}(\bar{u}) & \frac{\partial \phi_2}{\partial u_2}(\bar{u}) \\
    \frac{\partial \phi_3}{\partial u_1}(\bar{u}) & \frac{\partial \phi_3}{\partial u_2}(\bar{u}) \\
    \end{array}  \right]
  \left[\begin{array}{cc}
    \frac{\partial \xi_1}{\partial w_1}(\bar{w}) & \frac{\partial \xi_1}{\partial w_2}(\bar{w}) \\
    \frac{\partial \xi_2}{\partial w_1}(\bar{w}) & \frac{\partial \xi_2}{\partial w_2}(\bar{w})
    \end{array}  \right].
$$
Analizando cada columna de este producto de matrices obtenemos que
\begin{align*}
\frac{\partial \psi}{\partial w_1}(\bar{w}) &=
      \frac{\partial \xi_1}{\partial w_1}(\bar{w}) \frac{\partial \phi}{\partial u_1}(\bar{u})
    + \frac{\partial \xi_2}{\partial w_1}(\bar{w}) \frac{\partial \phi}{\partial u_2}(\bar{u}) \\
\frac{\partial \psi}{\partial w_2}(\bar{w}) &=
      \frac{\partial \xi_1}{\partial w_2}(\bar{w}) \frac{\partial \phi}{\partial u_1}(\bar{u})
    + \frac{\partial \xi_2}{\partial w_2}(\bar{w}) \frac{\partial \phi}{\partial u_2}(\bar{u}).
\end{align*}
De las propiedades del producto vectorial obtenemos que
\begin{align*}
\frac{\partial \psi}{\partial w_1}(\bar{w}) \times \frac{\partial \psi}{\partial w_2}(\bar{w})
&= \left( \frac{\partial \xi_1}{\partial w_1}(\bar{w}) \frac{\partial \xi_2}{\partial w_2}(\bar{w})
        - \frac{\partial \xi_2}{\partial w_1}(\bar{w}) \frac{\partial \xi_1}{\partial w_2}(\bar{w})\right)
  \frac{\partial \phi}{\partial u_1}(\bar{u}) \times \frac{\partial \phi}{\partial u_2}(\bar{u})\\
&= \det\bigl(\xi'(\bar{w})\bigr) \frac{\partial \phi}{\partial u_1}(\bar{u}) \times \frac{\partial \phi}{\partial u_2}(\bar{u}).
\end{align*}




> **Definición**
>
> Diremos que dos parametrizaciones de una superficie regular **preservan orientación** si en la región en común la matriz Jacobiana del cambio de variables tiene determinante positivo, y que **invierten orientación** si el determinante es negativo.
>
> Diremos que una superficie regular $\mathcal{S}$ es **orientable** si puede cubrirse por una familia de parametrizaciones que preservan orientación. Si tal familia no existe, la superficie se dice **no orientable**.




#### Ejemplo

Veamos que $\mathcal{S}=\{x\in\mathbb{R}^3 \mid \|x\|=1\}$ es una superficie regular orientable. Las parametrizaciones
\begin{align*}
\phi(u) &= \left( u_1, u_2, \sqrt{1-u_1^2 - u_2^2 } \right),\quad \operatorname{dom}(\phi)=B_1(0), \\
\psi_1(w) &= \left( w_1, \sqrt{1-w_1^2 - w_2^2}, w_2  \right),\quad \operatorname{dom}(\psi_1)=B_1(0),\\
\psi_2(w) &= \left( w_1, -\sqrt{1-w_1^2 - w_2^2}, w_2  \right),\quad \operatorname{dom}(\psi_2)=B_1(0),
\end{align*}
cubren regiones en común. Se vé que $\phi$ y $\psi_1$ coinciden en $V_1\cap\mathcal{S}$ con $V_1=\{x\in\mathbb{R}^3 \mid x_2 >0, \, x_3 >0 \}$, $\phi$ y $\psi_2$ coinciden en $V_2\cap\mathcal{S}$ con $V_2=\{x\in\mathbb{R}^3 \mid x_2 <0, \, x_3 >0 \}$.


```
#@title Esfera orientada { display-mode: "form" }
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

#fig = plt.figure(figsize=(12,4))
#ax = fig.add_subplot(1,3,1,projection='3d')
subplot_args = {'projection':'3d'}
fig, ax = plt.subplots(1,3,subplot_kw=subplot_args,figsize=(13,4))

th = np.linspace(0, 2*np.pi, 50)
r = np.linspace(0, 1, 30)
x = np.outer(r, np.cos(th))
y = np.outer(r, np.sin(th))
z = np.outer(np.sqrt(1-r**2), np.ones(th.shape))

# phi
ax[0].plot_surface(x, y, z, color='b', alpha=0.7)
ax[0].plot_surface(x, y, -z, color='c', alpha=0.1)
ax[0].plot3D([0,0],[-1.1,1.1], [0,0], color=(0.5,0.5,0.5))
ax[0].plot3D([-1.1,1.1], [0,0], [0,0], color=(0.5,0.5,0.5))
ax[0].plot3D([0,0], [0,0], [-1.1,1.1], color=(0.5,0.5,0.5))
ax[0].view_init(30,30)
ax[0].text(1,-1,1,r'$\phi$',fontsize=14)
ax[0].axis('off')
# psi1
ax[1].plot_surface(x, z, y, color='b')
ax[1].plot_surface(x, -z, y, color='c', alpha=0.1)
ax[1].plot3D([0,0],[-1.1,1.1], [0,0], color=(0.5,0.5,0.5))
ax[1].plot3D([-1.1,1.1], [0,0], [0,0], color=(0.5,0.5,0.5))
ax[1].plot3D([0,0], [0,0], [-1.1,1.1], color=(0.5,0.5,0.5))
ax[1].view_init(30,30)
ax[1].text(1,-1,1,r'$\psi_1$',fontsize=14)
ax[1].axis('off')
# psi2
ax[2].plot_surface(x, -z, y, color='b', alpha=0.7)
ax[2].plot_surface(x, z, y, color='c', alpha=0.1)
ax[2].plot3D([0,0],[-1.1,1.1], [0,0], color=(0.5,0.5,0.5))
ax[2].plot3D([-1.1,1.1], [0,0], [0,0], color=(0.5,0.5,0.5))
ax[2].plot3D([0,0], [0,0], [-1.1,1.1], color=(0.5,0.5,0.5))
ax[2].view_init(30,30)
ax[2].text(1,-1,1,r'$\psi_2$',fontsize=14)
ax[2].axis('off');
#plt.savefig('ejm_esferaorientada.pdf')


```

Luego $\phi: U_1 \to V_1 \cap \mathcal{S}$, $\psi_1: W \to V_1 \cap \mathcal{S}$ para $U_1=\{ u\in B_1(0) \mid u_2>0 \}$ y $W=\{ w\in B_1(0) \mid w_2>0 \}$;
y $\phi: U_2 \to V_2 \cap \mathcal{S}$, $\psi_2: W \to V_2 \cap \mathcal{S}$ para $U_2=\{ u\in B_1(0) \mid u_2<0 \}$.
Las funciones de cambio de variables son $\xi_1:W \to U_1$ y $\xi_2:W \to U_2$ tales que
\begin{align*}
\xi_1(w) &= \bigl( w_1, \sqrt{1-w_1^2 - w_2^2} \bigr), \quad
\xi_1'(w) = \left[\begin{array}{cc}
    1 & 0 \\
    \frac{-w_1}{\sqrt{1-w_1^2 - w_2^2}} & \frac{-w_2}{\sqrt{1-w_1^2 - w_2^2}}
    \end{array}  \right],\\
\xi_2(w) &= \bigl( w_1,  -\sqrt{1-w_1^2 - w_2^2} \bigr), \quad
\xi_2'(w) = \left[\begin{array}{cc}
    1 & 0 \\
    \frac{w_1}{\sqrt{1-w_1^2 - w_2^2}} & \frac{w_2}{\sqrt{1-w_1^2 - w_2^2}}
    \end{array}  \right].
\end{align*}
Como $\det\bigl( \xi_1'(w) \bigr)<0$ y  $\det\bigl( \xi_2'(w) \bigr)>0$ en $W$ entonces $\psi_2$ preserva la orientación de $\phi$ y $\psi_1$ invierte la orientación de $\phi$.

Se pueden escoger parametrizaciones que preservan la orientación y cubren $\mathcal{S}$, obteniendo que la esfera es una superficie orientable.

#### Ejemplo

La superficie $\mathcal{S}$, conocida como **banda de Möbius**, puede parametrizarse por $\phi: U \to \mathbb{R}^3$ tal que
$$
\phi(u) = \left(
\bigl(2 - u_2 \operatorname{sen}\bigl({\textstyle \frac{u_1}{2}} \bigr) \bigr) \operatorname{sen}(u_1), \,
\bigl(2 - u_2 \operatorname{sen}\bigl({\textstyle \frac{u_1}{2}} \bigr) \bigr) \cos(u_1),\,
u_2 \cos\bigl({\textstyle \frac{u_1}{2}} \bigr) \right),
$$
con $U=(0, 2\pi) \times (-1,1)$.


Note que esta superficie no tiene una cara interna y otra externa. Se puede ver que cualquier otra parametrización tendrá dos regiones comunes donde en una se preserva la orientación y en la otra se invierte.

Veamos que la integral de superficie es independiente de la parametrización.

>**Teorema**
>
> Sea $f:\mathbb{R}^3 \to \mathbb{R}$ definida sobre la superficie regular $\mathcal{S}$, parametrizada por $\phi: U \subset \mathbb{R}^2 \to \mathbb{R}^3$ y por $\psi: W \subset \mathbb{R}^2 \to \mathbb{R}^3$. Entonces
$$
 \int_\phi f(x) dS = \int_\psi f(x) dS.
$$


**Demostración**

Sea $\psi(w) = \phi(\xi(w))$, como $\mathcal{S}$ es regular entonces $|\det\bigl( \xi'(w)\bigr)| >0$. Como  $U = \xi(W)$ y $\psi'(w) = \phi'(\xi(w)) \xi'(w)$, entonces
\begin{align*}
    \int_\psi f(x) dS &= \int_W  f(\psi(w)) \left\|{\textstyle \frac{\partial \psi}{\partial w_1}(w) \times \frac{\partial \psi}{\partial w_2}(w)} \right\| dw \\
    &= \int_W f(\phi(\xi(w)))  \left\|{\textstyle \frac{\partial \phi}{\partial u_1}(\xi(w)) \times \frac{\partial \phi}{\partial u_2}(\xi(w))} \right\| |\det\bigl( \xi'(w)\bigr)| dw \\
    &= \int_U f(\phi(u)) \left\|{\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)} \right\| du\\
    &= \int_\phi f(x) dS. \quad \spadesuit
\end{align*}





De esta manera, la integral de $f$ sobre una superficie $\mathcal{S}$ se puede denotar por
$$
    \int_{\mathcal{S}} f(x) dS.
$$
Claramente, si $\mathcal{S}$ es la unión de $k$ superficies no solapadas $\mathcal{S}_i$, entonces
$$
    \int_{\mathcal{S}} f(x) dS = \sum_{i=1}^k \int_{\mathcal{S}_i} f(x) dS.
$$



Cuando $\mathcal{S}$ es una superficie regular orientable y $\phi: U \to \mathcal{S}$ una parametrización, se dice que la superficie tiene orientación positiva si, cuando la frontera de $U\subset \mathbb{R}^2$ es recorrida en sentido antihorario por una trayectoria $\gamma$, entonces $t \mapsto \frac{\partial \phi}{\partial u_1}(\gamma(t)) \times \frac{\partial \phi}{\partial u_2}(\gamma(t))$ recorre la frontera de $\mathcal{S}$ de manera que la superficie siempre quede a la izquierda del mismo.

>**Teorema**
>
> Sea $F:\mathbb{R}^3 \to \mathbb{R}^3$ definida sobre la superficie regular orientable $\mathcal{S}$, parametrizada por $\phi: U \subset \mathbb{R}^2 \to \mathbb{R}^3$ con orientación positiva, y por $\psi: W \subset \mathbb{R}^2 \to \mathbb{R}^3$. Si $\psi$ preserva la orientación, entonces
$$
 \int_\psi F(x) dS = \int_\phi F(x) dS,
$$
y si $\psi$ invierte la orientación, entonces
$$
 \int_\psi F(x) dS = - \int_\phi F(x) dS,
$$


**Demostración**

Como $\psi(w) = \phi(\xi(w))$, entonces $\psi'(w) = \phi'(\xi(w)) \xi'(w)$ con $\det\bigl( \xi'(w)\bigr) >0$ si $\psi$ preserva la orientación y $\det\bigl( \xi'(w)\bigr) <0$ si invierte la orientación. Luego, usando que $U = \xi(W)$ obtenemos que
\begin{align*}
    \int_\psi F(x) dS &= \int_W  \bigl\langle F(\psi(w)), {\textstyle \frac{\partial \psi}{\partial w_1}(w) \times \frac{\partial \psi}{\partial w_2}(w)} \bigr\rangle dw \\
    &= \int_W \bigl\langle F(\phi(\xi(w))), {\textstyle \frac{\partial \phi}{\partial u_1}(\xi(w)) \times \frac{\partial \phi}{\partial u_2}(\xi(w))} \bigr\rangle \det\bigl( \xi'(w)\bigr) dw \\
    &= \pm \int_W \bigl\langle F(\phi(\xi(w)), {\textstyle \frac{\partial \phi}{\partial u_1}(\xi(w)) \times \frac{\partial \phi}{\partial u_2}(\xi(w))} \bigr\rangle |\det\bigl( \xi'(w)\bigr)| dw \\    
    &= \pm \int_U \bigl\langle F(\phi(u)), {\textstyle \frac{\partial \phi}{\partial u_1}(u) \times \frac{\partial \phi}{\partial u_2}(u)} \bigr\rangle  du\\
    &= \pm \int_\phi F(x) dS,
\end{align*}
donde $\pm$ es justamente el signo de $\det\bigl( \xi'(w)\bigr)$. $\spadesuit$





Con esto, la integral de superficie de $F$ sobre una superficie regular orientada $\mathcal{S}$ puede denotarse por
$$
    \int_{\mathcal{S}} F(x) dS \quad \textrm{ o }\quad  \int_{\mathcal{S}} \langle F(x), N(x) \rangle dS.
$$
Si $\mathcal{S}$ es una superficie orientada positivamente y $\mathcal{S}^-$ tiene la orientación opuesta, entonces
$$
    \int_{\mathcal{S}} F(x) dS = - \int_{\mathcal{S}^-} F(x) dS
$$
Además, si $\mathcal{S}$ es la unión de $k$ superficies no solapadas $\mathcal{S}_i$ orientadas positivamente, se puede ver que las curvas dadas por sus fronteras se recorren en sentidos opuestos y por lo tanto sus integrales se cancelan, obteniendo que
$$
    \int_{\mathcal{S}} F(x) dS = \sum_{i=1}^k \int_{\mathcal{S}_i} F(x) dS.
$$


#### Ejemplo

Suponga que se desea hallar el flujo de calor a través de una superficie $\mathcal{S}$. Considere que la función de temperatura es $f:\mathbb{R}^3 \to \mathbb{R}$ tal que $f(x) = x_1^2 + x_2^2 + x_3^2$, y $\mathcal{S}=\{x\in \mathbb{R}^3 \mid \|x\|=1\}$ orientada con la normal saliente. Como la temperatura va de los lugares de mayor concentración a los de menor concentración, entonces el campo que genera es dado por $F(x) = -\nabla f(x) = -2x$. Por otro lado, el vector normal saliente en $x\in \mathcal{S}$ es $N(x) = x$. Con esto obtenemos que el flujo de calor es
\begin{align*}
\int_{\mathcal{S}} F(x) dS &= \int_{\mathcal{S}} \langle F(x), N(x) \rangle dS = -2 \int_{\mathcal{S}} 1 dS \\
&= -2 \mathscr{a}(\mathcal{S}) = -2 (4\pi) = -8 \pi.
\end{align*}
