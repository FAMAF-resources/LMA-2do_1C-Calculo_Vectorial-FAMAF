# Funciones de varias variables

## Funciones diferenciables

Queremos aproximar la función no lineal $v \mapsto f(\bar{x} + v)$ por la función afín $v \mapsto f(\bar{x}) + L(v)$. Donde la condición para ser una "muy buena" aproximación es que la distancia de $f(\bar{x} + v)$ a $f(\bar{x}) + L(v)$ tienda a cero más rápido de lo que $v$ tiende a cero. **¿Quién es $L$?**

La clase pasada definimos
* derivada parcial:
$$
\frac{\partial f}{\partial x_j}(\bar{x}) = \lim_{t \to 0} \frac{f(\bar{x} + t e^j) - f(\bar{x})}{t},
$$
* derivada direccional:
$$
f'(\bar{x}; v) = \lim_{t \to 0} \frac{f(\bar{x} + t v) - f(\bar{x})}{t},
$$
y vimos que el candidato sería tomar $L(v) = f'(\bar{x}; v)$ y por lo tanto $L(e^j) = \frac{\partial f}{\partial x_j}(\bar{x})$.

En los casos donde la función $v \mapsto f'(\bar{x};v)$ es lineal podemos tener una noción de derivada en varias variables.

>**Definición**
>
> Para $f:\mathbb{R}^n \to \mathbb{R}^m$, si la derivada direccional de $f$ en $\bar{x}$ existe para toda dirección $v\in\mathbb{R}^n$ y $v \mapsto f'(\bar{x}; v)$ es una función lineal, diremos que la **derivada de $f$ en $\bar{x}$**, denotada por $Df(\bar{x})$, es la función lineal tal que  $Df(\bar{x})(v) = f'(\bar{x}; v)$.
>
> Más aún, su representación matricial, llamada **matriz Jacobiana de $f$ en $\bar{x}$**, será denotada por $f'(\bar{x})$ y es la matriz $\mathbb{R}^{m \times n}$ cuyo elemento en fila $i$ columna $j$ es $\frac{\partial f_i}{\partial x_j}(\bar{x})$, o sea,
$$
    f'(\bar{x}) = \left[ \begin{array}{ccc}
         \frac{\partial f_1}{\partial x_1}(\bar{x}) & \ldots & \frac{\partial f_1}{\partial x_n}(\bar{x}) \\
         \vdots & \ddots & \vdots\\
         \frac{\partial f_m}{\partial x_1}(\bar{x}) & \ldots & \frac{\partial f_m}{\partial x_n}(\bar{x})  \end{array}\right].
$$



Recuerde que de lo visto para funciones lineales, como  $Df(\bar{x}):\mathbb{R}^n \to \mathbb{R}^m$ es lineal entonces la $j$-ésima columna de su matriz asociada debe tener los elementos de $Df(\bar{x})(e^j)$, y en este caso
$$
    Df(\bar{x})(e^j) = f'(\bar{x}; e^j) = \frac{\partial f}{\partial x_j}(\bar{x}).
$$




En el caso particular de funciones a valores escalares, o sea $f:\mathbb{R}^n \to \mathbb{R}$, lo anterior vale con $m=1$. De esta manera $Df(\bar{x}):\mathbb{R}^n \to \mathbb{R}$ es una función lineal que puede expresarse en términos del producto interno de la siguiente manera
\begin{align*}
    Df(\bar{x})(v) &= Df(\bar{x})(v_1 e^1 + \ldots + v_n e^n) \\
                    &= v_1 Df(\bar{x})(e^1) + \ldots + v_n Df(\bar{x})(e^n) \\
                    &= v_1 \frac{\partial f}{\partial x_1}(\bar{x}) + \ldots + v_n \frac{\partial f}{\partial x_n}(\bar{x}) \\
                    &= \left\langle \left( {\textstyle \frac{\partial f}{\partial x_1}(\bar{x}), \ldots, \frac{\partial f}{\partial x_n}(\bar{x})} \right)\, , \, (v_1, \ldots, v_n)\right\rangle.
\end{align*}
Con esto, obtenemos una correspondencia entre $Df(\bar{x})$ y un elemento de $\mathbb{R}^n$. Dicho de otra manera, para toda función $f$ a valores reales podremos obtener un representante de la derivada en el espacio dominio de $f$.


>**Definición**
>
> Sea $f:\mathbb{R}^n \to \mathbb{R}$ tal que existe su derivada en $\bar{x}$. Llamaremos **gradiente de $f$ en $\bar{x}$**, denotado por $\nabla f (\bar{x})$ (o $\operatorname{grad(f)}(\bar{x})$), al elemento de $\mathbb{R}^n$
$$
    \nabla f(\bar{x}) = \left(  \frac{\partial f}{\partial x_1}(\bar{x}), \ldots, \frac{\partial f}{\partial x_n}(\bar{x}) \right).
$$




Recapitulando, si $f:\mathbb{R}^n \to \mathbb{R}$ tenemos que $Df(x)$ es una función lineal, $f'(x)$ es una matriz $1 \times n$ y $\nabla f(x)$ es un elemento de $\mathbb{R}^n$. Para evitar confusiones, vale la pena aclarar que para obtener $f'(x)$, la representación matricial de $Df(x)$, tuvimos que asociar $\mathbb{R}^n$ con los vectores columna $\mathbb{R}^{n \times 1}$, y con esta estructura obtendríamos que $\nabla f(x)$ es un vector columna, o sea, una matriz $n \times 1$. En este caso es fácil ver que
$$
 \nabla f(x) = f'(x)^T.
$$

Una de las ventajas de asociar $Df(x)$ con el gradiente $\nabla f(x)$, es que a este último lo podemos analizar como la función $\nabla f:\mathbb{R}^n \to \mathbb{R}^n$.



#### Ejemplo

Como vimos anteriormente, una matriz $A\in \mathbb{R}^{m\times n}$ puede asociarse a un vector $a\in\mathbb{R}^{mn}$ ordenando los elementos de $A$ por columnas. En este caso $\langle a, b \rangle = \operatorname{tr}(A^T B)$ y, para trabajar en $\mathbb{R}^{m\times n}$, usaremos que $\langle A, B \rangle = \operatorname{tr}(A^T B)$.

De esta manera, dotando a los elementos de $\mathbb{R}^{mn}$ con la estructura matricial $\mathbb{R}^{m\times n}$ podemos definir la función $f:\mathbb{R}^{m \times n} \to \mathbb{R}$ tal que
$$
    f(A) = \frac{1}{2} \operatorname{tr}(A^T A).
$$
Para ver su derivada direccional en una dirección $V\in \mathbb{R}^{m\times n}$, note que por la linealidad de la función traza y el hecho que $\operatorname{tr}(A^T B) = \operatorname{tr}(B^T A)$,
\begin{align*}
    f(A+ tV) &= \frac{1}{2} \operatorname{tr}\bigl( (A+tV)^T (A+tV) \bigr)
             = \frac{1}{2} \operatorname{tr}\bigl( A^T A + t A^T V + t V^T A + t^2 V^T V \bigr) \\
            &= f(A) + t \operatorname{tr}(A^T V) + \frac{t^2}{2} \operatorname{tr}(V^T V).
\end{align*}
Por lo tanto
$$
    f'(A; V) = \lim_{t \to 0} \frac{f(A+tV) - f(A)}{t}
            = \lim_{t \to 0} \operatorname{tr}(A^T V) +  \frac{t}{2} \operatorname{tr}(V^T V)
            = \operatorname{tr}(A^T V).
$$
La linealidad de la traza nos garantiza que $V \mapsto f'(A; V)$ es una función lineal. Por lo tanto existe $Df(A):\mathbb{R}^{m \times n} \to \mathbb{R}$ y al evaluarla obtenemos que $Df(A)(V) = \operatorname{tr}(A^T V)$. Para obtener el gradiente, como
$$
    Df(A)(V) = \operatorname{tr}(A^T V) = \langle A, V\rangle,
$$
concluímos que
$$
\nabla f(A) = A.
$$
Para obtener la matriz Jacobiana tendremos que asociar $\mathbb{R}^{m \times n}$ con los vectores columna $\mathbb{R}^{mn \times 1}$. La $k$-ésima columna de la matriz se obtiene de la derivada a lo largo de la dirección canónica $e^k$, que con el orden pactado tiene asociada una matriz $E_k$ con ceros y un uno en el elemento con fila $i$ columna $j$ tal que $k = i + m(j -1)$. Como $Df(A)(E_k) = \langle A, E_k \rangle = a_{ij}$, concluímos que $f'(A) \in \mathbb{R}^{1 \times mn}$ tal que
$$
    f'(A) = [a_{11} \ \ldots \ a_{m1} \ a_{12} \ \ldots \ a_{m2} \ \ldots \ a_{1n} \ \ldots \ a_{mn} ].
$$

---


Cuando $f:\mathbb{R}^n \to \mathbb{R}^m$ con $m>1$ también se puede hacer una correspondencia de $Df$ con una función que en este caso será vectorial. Para ello, note que la función lineal $v \mapsto Df(x)(v)$ tiene una correspondencia con la matriz $f'(x)$ que tiene dimensión $m \times n$, y ésta se corresponde con un elemento de $\mathbb{R}^{mn}$.
Por otro lado, para cada $x \in \mathbb{R}^n$ tenemos que existe una única función lineal $Df(x)$, o sea, $Df$ es una función del espacio $\mathbb{R}^n$ al espacio de funciones lineales de $\mathbb{R}^n$ a $\mathbb{R}^m$. Usando la correspondencia mencionada anteriormente, podemos analizar a la derivada como si fuera una función $Df: \mathbb{R}^n \to \mathbb{R}^{mn}$.




Ahora que tenemos la noción de derivada sólo resta ver que se cumple la condición de aproximación que queremos, o sea, que la distancia de $f(\bar{x} + v)$ a $f(\bar{x}) + Df(\bar{x})(v)$ tienda a cero más rápido de lo que $v$ tiende a cero. Claramente para cualquier $v\in\mathbb{R}^n\setminus\{0\}$ tenemos que
\begin{align*}
\lim_{t \to 0} \frac{\|f(\bar{x} + t v) - \bigl( f(\bar{x}) + Df(\bar{x})(t v) \bigr)\|}{\|t v\|}
    &= \lim_{t \to 0} \frac{1}{\|v\|} \left\|\frac{f(\bar{x} + t v) - f(\bar{x})}{t} -  Df(\bar{x})(v) \right\| \\
    &= 0.
\end{align*}



Lamentablemente esta condición debería cumplirse para cualquier curva (o sucesión) que tiende a cero y no sólo para rectas. Veamos a continuación un caso donde esta condición vale para rectas pero no para curvas.


#### Ejemplo

Considere la función $f:\mathbb{R}^2 \to \mathbb{R}$ tal que
$$
    f(x) = \left\{ \begin{array}{cc}
         \frac{x_1^3 x_2}{x_1^6 + x_2^2}& \textrm{ si } x\neq 0,  \\
         0 & \textrm{ si } x=0.
    \end{array}\right.
$$


```
#@title  { run: "auto", form-width: "50%" }
elevacion = 52 #@param {type:"slider", min:0, max:90, step:1}
acimut = 30 #@param {type:"slider", min:0, max:180, step:1}

import numpy as np
import matplotlib.pyplot as plt

x1, x2 = np.meshgrid(np.linspace(-1,1,200), np.linspace(-1,1,200))
f = x1**3*x2/(x1**6 + x2**2)

ax = plt.axes(projection='3d')
ax.plot_surface(x1, x2, f, rstride=1, cstride=1, cmap='viridis')
ax.view_init(elevacion, acimut);
#ax.axis('off');
#plt.savefig('ejm_nodif2.pdf')
```

Si $\bar{x}=0$, para cualquier $v\in\mathbb{R}^2$ con $v\neq 0$ se tiene que
$$
f'(\bar{x};v) = \lim_{t \to 0} \frac{f(\bar{x} + tv) - f(\bar{x})}{t}
            = \lim_{t \to 0}  \frac{1}{t}\frac{t^3 v_1^3 t v_2}{t^6 v_1^6 + t^2 v_2^2} = 0.
$$
Por lo tanto, la función $v \mapsto f'(\bar{x};v)$ es lineal, pues es identicamente nula, y existe $Df(\bar{x})$ tal que $Df(\bar{x})(v)=0$. No obstante, la aproximación afín no cumple con la condición pedida.

Considere la curva $v(t) = (t,t^3)$, entonces $v(t) \to 0$ cuando $t\to 0$ pero
$$
\lim_{t \to 0} f(\bar{x} + v(t)) - f(\bar{x}) - Df(\bar{x})(v(t))
            = \lim_{t \to 0}  \frac{t^3 t^3}{t^6 + t^6} = \frac{1}{2}.
$$

---


De esta manera, la aproximación que buscamos debe ser buena si nos aproximamos a cero por cualquier curva y no solo a lo largo de rectas. Las funciones que buscamos son las que definimos a continuación.

>**Definición**
>
> Diremos que $f:\mathbb{R}^n \to \mathbb{R}^m$ es **diferenciable en $\bar{x}$** si
\begin{equation*}
    \lim_{v \to 0} \frac{f(\bar{x} + v) - f(\bar{x}) - Df(\bar{x})(v)}{\|v\|} = 0.    
\end{equation*}
Diremos simplemente que $f$ es **diferenciable** si es diferenciable en todos los puntos de su dominio.



Note que si uno tiene una función lineal cualquiera $L:\mathbb{R}^n \to \mathbb{R}^m$ que cumple
$$
    \lim_{v \to 0} \frac{f(\bar{x} + v) - f(\bar{x}) - L(v)}{\|v\|} = 0,
$$
al restringir el límite a los ejes Cartesianos obtendremos que $L(e^j) = \frac{\partial f}{\partial x_j} (\bar{x})$. O sea $L = Df(\bar{x})$.




En muchos casos es más práctico utilizar la notación asintótica **o pequeño** y **O grande**, que consiste en escribir $r(v) = o(\|v\|)$ para cualquier función $r:\mathbb{R}^n \to \mathbb{R}^m$ tal que
$$
    \lim_{v \to 0} \frac{r(v)}{\|v\|} = 0,
$$
y $r(v) = O(\|v\|)$ para cualquier función $r:\mathbb{R}^n \to \mathbb{R}^m$ tal que
$$
    \| r(v)\| \leq \kappa \|v\|,
$$
para algún $\kappa>0$ y para todo $v$ suficientemente cerca de $0$.

Con esta notación, podemos decir que $f$ es diferenciable en $\bar{x}$ si $f(\bar{x} + v) - f(\bar{x}) - Df(\bar{x})(v) = o(\|v\|)$, o equivalentemente
$$
    f(\bar{x} + v) = f(\bar{x}) + Df(\bar{x})(v) + o(\|v\|),
$$
para $v$ suficientemente cerca de cero.


Haciendo el cambio de variable $v= x - \bar{x}$, podemos decir que $f$ es diferenciable en $\bar{x}$ si
\begin{equation*}
    f(x) = f(\bar{x}) + Df(\bar{x})(x - \bar{x}) + o(\| x - \bar{x} \|),    
\end{equation*}
para $x$ suficientemente cerca de $\bar{x}$.





Ya que en la mayoría de los casos es más práctico trabajar con la matriz Jacobiana, que proviene de escribir $\mathbb{R}^n$ y $\mathbb{R}^m$ como vectores columna, la ecuación anterior con esta estructura es
$$
    \left[\begin{array}{c} f_1(x) \\ \vdots \\ f_m(x) \end{array} \right]
     = \left[\begin{array}{c} f_1(\bar{x}) \\ \vdots \\ f_m(\bar{x}) \end{array} \right]
     + \left[ \begin{array}{ccc}
    \frac{\partial f_1}{\partial x_1}(\bar{x}) & \ldots & \frac{\partial f_1}{\partial x_n}(\bar{x}) \\
         \vdots & \ddots & \vdots\\
    \frac{\partial f_m}{\partial x_1}(\bar{x}) & \ldots & \frac{\partial f_m}{\partial x_n}(\bar{x})  \end{array}\right]
    \left[\begin{array}{c} x_1 - \bar{x}_1 \\ \vdots \\ x_n - \bar{x}_n \end{array} \right]
    + o(\| x - \bar{x} \|).
$$




Veamos cómo calcular la matriz Jacobiana para una función concreta.

#### Ejemplo

Sea $f:\mathbb{R}^3 \to \mathbb{R}^2$ tal que $f(x) = (x_3 e^{x_1}, -x_2 e^{x_3})$. Pensando $\mathbb{R}^3$ como $\mathbb{R}^{3 \times 1}$ tenemos que
$$
    f(x) = \left[\begin{array}{c} x_3 e^{x_1} \\ -x_2 e^{x_3} \end{array} \right],
    \qquad
    f'(x) = \left[\begin{array}{ccc}
                    x_3 e^{x_1} &   0       & e^{x_1}     \\
                          0     & -e^{x_3}  & -x_2 e^{x_3}
            \end{array} \right].
$$
Por lo tanto
\begin{align*}
     f(x) &- f(\bar{x}) - f'(\bar{x})(x-\bar{x}) \\
    &= \left[\begin{array}{c} x_3 e^{x_1} \\ -x_2 e^{x_3} \end{array} \right]
      -\left[\begin{array}{c} \bar{x}_3 e^{\bar{x}_1} \\ -\bar{x}_2 e^{\bar{x}_3} \end{array} \right]
      -\left[\begin{array}{ccc}
                    \bar{x}_3 e^{\bar{x}_1} &   0       & e^{\bar{x}_1}     \\
                          0     & -e^{\bar{x}_3}  & -\bar{x}_2 e^{\bar{x}_3}
            \end{array} \right]
        \left[\begin{array}{c} x_1 - \bar{x}_1 \\ x_2 - \bar{x}_2 \\ x_3 - \bar{x}_3 \end{array} \right] \\
    &= \left[\begin{array}{c}
    x_3 e^{x_1} - \bar{x}_3 e^{\bar{x}_1} - \bar{x}_3 e^{\bar{x}_1}(x_1 - \bar{x}_1) - e^{\bar{x}_1}(x_3 - \bar{x}_3) \\
    -x_2 e^{x_3} + \bar{x}_2 e^{\bar{x}_3} + e^{\bar{x}_3}(x_2 - \bar{x}_2) + \bar{x}_2 e^{\bar{x}_3}(x_3 - \bar{x}_3)
    \end{array} \right].
\end{align*}
Como $e^\xi = e^{\bar{\xi}} + e^{\bar{\xi}}(\xi - \bar{\xi}) + \frac{1}{2}e^{\bar{\xi} + \tau (\xi - \bar{\xi})}(\xi - \bar{\xi})^2$ para algún $\tau \in (0,1)$ y $|x_j - \bar{x}_j| \leq \|x - \bar{x}\|$ para todo $j$, para la primera función coordenada tenemos
\begin{align*}
     |x_3 & e^{x_1} - \bar{x}_3 e^{\bar{x}_1} - \bar{x}_3 e^{\bar{x}_1}(x_1 - \bar{x}_1) - e^{\bar{x}_1}(x_3 - \bar{x}_3)| \\
    & = |(x_3 - \bar{x}_3)e^{\bar{x}_1} + (x_3 - \bar{x}_3) e^{\bar{x}_1}(x_1 - \bar{x}_1)
        - e^{\bar{x}_1}(x_3 - \bar{x}_3) + {\textstyle\frac{1}{2}}x_3 e^{\bar{x}_1 + \tau_1 (x_1 - \bar{x}_1)}(x_1 - \bar{x}_1)^2| \\
    & \leq \left( e^{\bar{x}_1} + {\textstyle\frac{1}{2}}|x_3| e^{\bar{x}_1 + \tau_1 (x_1 - \bar{x}_1)} \right) \|x - \bar{x}\|^2 = o(\|x - \bar{x}\|).
\end{align*}
Para la segunda función coordenada,
\begin{align*}
    |-x_2 & e^{x_3} + \bar{x}_2 e^{\bar{x}_3} + e^{\bar{x}_3}(x_2 - \bar{x}_2) + \bar{x}_2 e^{\bar{x}_3}(x_3 - \bar{x}_3)| \\
    & = | (\bar{x}_2 - x_2) e^{\bar{x}_3} + e^{\bar{x}_3}(x_2 - \bar{x}_2) + (\bar{x}_2 - x_2) e^{\bar{x}_3}(x_3 - \bar{x}_3) - \frac{1}{2} x_2 e^{\bar{x}_3 + \tau_3 (x_3 - \bar{x}_3)}(x_3 - \bar{x}_3)^2| \\
    & \leq \left( e^{\bar{x}_3} + \frac{1}{2} |x_2| e^{\bar{x}_3 + \tau_3 (x_3 - \bar{x}_3)}\right) \|x - \bar{x} \|^2 = o(\|x - \bar{x}\|).
\end{align*}

Concluyendo que $f$ es diferenciable.

---

Como puede apreciar, verificar la validez del límite que garantiza la diferenciabilidad de una función puede ser complejo en algunos casos. Por ello es bueno contar con herramientas que nos permitan verificar esa condición fácilmente.



Existen condiciones necesarias que una función diferenciable debe cumplir, como tener derivada, derivada direccional en toda dirección y por lo tanto derivadas parciales. Otra condición necesaria es que debe ser continua.

>**Teorema**
>
> Si $f:\mathbb{R}^n \to \mathbb{R}^m$ es diferenciable en $\bar{x}$, entonces $f$ es continua en $\bar{x}$. Más aún, existe $\kappa>0$ tal que $\|f(x) - f(\bar{x})\| \leq \kappa \|x - \bar{x}\|$ para todo $x$ en un entorno de $\bar{x}$.


**Demostración**

Como $\lim_{x \to \bar{x}} \frac{o(\|x-\bar{x}\|)}{\|x-\bar{x}\|} = 0$, entonces dado $\varepsilon>0$ existe $\delta>0$ tal que $\| o(\|x-\bar{x}\|)\| \leq \varepsilon \|x-\bar{x}\|$ si $x\in B_\delta(\bar{x})$.
Por otro lado, como $Df(\bar{x})$ es lineal, sabemos que existe $\tilde{\kappa}$ tal que $\| Df(\bar{x})(v) \| \leq \tilde{\kappa} \|v\|$ para todo $v\in\mathbb{R}^n$. Luego para todo $x\in B_\delta(\bar{x})$ se tiene
\begin{align*}
    \|f(x) - f(\bar{x}) \| &= \bigl\| Df(\bar{x})(x - \bar{x}) + o(\|x - \bar{x}\|) \bigr\| \\
                            &\leq \| Df(\bar{x})(x - \bar{x}) \| + \|o(\|x - \bar{x}\|)\| \\
                            &\leq \tilde{\kappa} \|x - \bar{x}\| + \varepsilon \|x-\bar{x}\|\\
                            &= (\tilde{\kappa} + \varepsilon) \|x-\bar{x}\|.
\end{align*}
Tomando $\kappa = \tilde{\kappa} + \varepsilon$ conseguimos la cota deseada. Ahora, claramente $\lim_{x \to \bar{x}} f(x) = f(\bar{x})$. $\spadesuit$


Utilizando notación asintótica, este resultado nos dice que $f(x) - f(\bar{x}) = O(\|x - \bar{x}\|)$ para $x$ suficientemente cerca de $\bar{x}$.




Veamos cómo utilizar este resultado para determinar la no diferenciabilidad de una función

#### Ejemplo

Como vimos en un ejemplo anterior, la función $f:\mathbb{R}^2 \to \mathbb{R}$ tal que
$$
    f(x) = \left\{ \begin{array}{cc}
         \frac{x_1 x_2}{x_1^2 + x_2^2}& \textrm{ si } x\neq 0,  \\
         0 & \textrm{ si } x=0,
    \end{array}\right.
$$
tiene derivada direccional en toda dirección. No obstante, si $x^k = (1/k, 1/k)$ tenemos que $x^k \to 0$ y
$$
\lim_{k \to \infty} f(x^k) = \lim_{k \to \infty} \frac{1}{2} = \frac{1}{2} \neq 0 = f(0).
$$
Claramente esta función no es continua en el origen y por lo tanto no puede ser diferenciable en ese punto.


Como vimos en el ejemplo anterior, la función $f:\mathbb{R}^2 \to \mathbb{R}$ tal que
$$
    f(x) = \left\{ \begin{array}{cc}
         \frac{x_1^3 x_2}{x_1^6 + x_2^2}& \textrm{ si } x\neq 0,  \\
         0 & \textrm{ si } x=0,
    \end{array}\right.
$$
tiene derivada en el origen y su matriz Jacobiana es $f'(0)=[0 \ 0]$. No obstante, si $x^k = (1/k, 1/k^3)$ tenemos que $x^k \to 0$ y
$$
\lim_{k \to \infty} f(x^k) = \lim_{k \to \infty} \frac{1}{2} = \frac{1}{2} \neq 0 = f(0).
$$
Como esta función no es continua en el origen no puede ser diferenciable en ese punto.

---
