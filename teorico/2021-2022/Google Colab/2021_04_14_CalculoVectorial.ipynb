### Topología del espacio Euclídeo

Recuerde que:
* $x$ es un **punto interior** de $S\subset \mathbb{R}^n$ si existe $r>0$ tal que
$$
B_r(x) \subset S.
$$
El conjunto de puntos interiores $S$ lo denotaremos por $\operatorname{int}(S)$ y diremos que $S$ es **abierto** si $\operatorname{int}(S)=S$.
* $x$ es un **punto de acumulación** de $S\subset \mathbb{R}^n$ si para todo $r>0$ vale que
$$
    \left( B_r(x)\setminus\{x\}\right) \cap S \neq \emptyset.
$$
$x$ es un **punto aislado** de $S$ si existe $r>0$ tal que $B_r(x) \cap S = \{x\}$. La unión de puntos aislados y puntos de acumulación de $S$ es la clausura de $S$, denotado por $\operatorname{cl}(S)$.

Se puede entender $\operatorname{cl}(S)$ como la unión de $S$ y los puntos de acumulación de $S$ que no pertenecen a $S$.


>**Definición**
>
> Diremos que $C \subset \mathbb{R}^n$ es un conjunto **cerrado** si $C = \operatorname{cl}(C)$.


Veamos cómo se relacionan con los abiertos.


>**Teorema**
>
> Un conjunto $C\subset \mathbb{R}^n$ es cerrado si y sólo si su complemento $\mathbb{R}^n \setminus C$ es abierto. Equivalentemente, $U \subset \mathbb{R}^n$ es abierto si y sólo si $\mathbb{R}^n \setminus U$ es cerrado.


**Demostración:**


El conjunto $C$ es cerrado,
    
$\Leftrightarrow$ para todo $x\notin C$ vale que $x$ no es un punto de acumulación de $C$,
    
$\Leftrightarrow$ para todo $x\notin C$ existe $r>0$ tal que $\bigl( B_r(x)\setminus\{x\}\bigr) \cap C = \emptyset$,
    
$\Leftrightarrow$ para todo $x\in \mathbb{R}^n \setminus C$ existe $r>0$ tal que $ B_r(x)\setminus\{x\} \subset \mathbb{R}^n \setminus C$,
    
$\Leftrightarrow$ para todo $x\in \mathbb{R}^n \setminus C$ existe $r>0$ tal que $ B_r(x) \subset \mathbb{R}^n \setminus C$,
    
$\Leftrightarrow$ todo $x\in \mathbb{R}^n \setminus C$ es un punto interior,
    
$\Leftrightarrow$ $\mathbb{R}^n \setminus C$ es abierto. $\spadesuit$





Ahora el siguiente resultado se deduce fácilmente tomando complemento y utilizando las propiedades vistas para los conjuntos abiertos.

>**Teorema**
>
> Los cerrados cumplen las siguientes propiedades:
1. Si $\{ C_\lambda\}_{\lambda \in \Lambda}$ es una familia arbitraria de conjuntos cerrados $C_\lambda \subset \mathbb{R}^n$, entonces la intersección
    $$
    \bigcap_{\lambda \in \Lambda} C_\lambda,
    $$
    es un conjunto cerrado.
2.  Si $C_1$ y $C_2$ son cerrados, entonces $C_1 \cup C_2$ es cerrado.


Con esto obtenemos además que la unión de un número finito de cerrados $C_1 \cup C_2 \cup \ldots \cup C_k$ es cerrado. Note que la unión infinita de cerrados puede no ser un cerrado, como por ejemplo
$$
    \bigcup_{k=1}^\infty \{y\in \mathbb{R}^n \mid \|y-x\| \leq 1-1/k \} = B_1(x).
$$


Otro conjunto que nos será de utilidad en este curso es el siguiente.

>**Definición**
>
> Diremos que $x$ es un **punto frontera** de $S\subset\mathbb{R}^n$ si para todo $r>0$ vale que
$$
B_r(x)\cap S \neq \emptyset \quad \textrm{ y } \quad B_r(x)\cap \bigl(\mathbb{R}^n \setminus S\bigr) \neq \emptyset.
$$
El conjunto formado por los puntos frontera se llama la **frontera** de $S$ y lo denotaremos por $\operatorname{fr}(S)$.





Con las herramientas que tenemos, podemos escribir la frontera en términos de la clausura y el interior.

>**Proposición**
>
> Si $S\subset \mathbb{R}^n$, entonces $\operatorname{fr}(S) = \operatorname{cl}(S) \setminus \operatorname{int}(S)$.


**Demostración:**

$x\in \operatorname{fr}(S)$
    
$\Leftrightarrow$ $B_r(x)\cap S \neq \emptyset$ y $B_r(x)\cap \bigl(\mathbb{R}^n \setminus S\bigr) \neq \emptyset$ para todo $r>0$,
    
$\Leftrightarrow$ $B_r(x)\cap S = \{x\}$ o $\bigl( B_r(x)\setminus\{x\} \bigr)\cap S \neq \emptyset$ y $B_r(x) \not\subset S$ para todo $r>0$,
    
$\Leftrightarrow$ $x\in\operatorname{cl}(S)$ y $x\notin \operatorname{int}(S)$,
    
$\Leftrightarrow$ $x\in\operatorname{cl}(S)\setminus \operatorname{int}(S)$. $\spadesuit$



De este resultado se puede deducir que:
* un conjunto es cerrado si y sólo si contiene a su frontera;
* un conjunto es abierto si y sólo si es disjunto con su frontera.



#### Ejemplo

Considere el siguiente conjunto.


```
#@title Tipos de puntos { display-mode: "form" }
import matplotlib.pyplot as plt
import numpy as np

th = np.linspace(0,2*np.pi,100)
circ = np.array([np.cos(th), np.sin(th)])
rect = np.array([[0.5, 1.5, 1.5, 0.5], [-0.5, -0.5, 0.5, 0.5]])
for r in np.linspace(0,1,50):
  plt.fill(r*circ[0], r*circ[1], color='red', alpha=0.1)
plt.fill(rect[0], rect[1], color='red')

punto1 = 0.9*circ[:,60]
punto2 = np.array([1, 0.5])
punto3 = np.array([-1,1])
punto4 = np.array([0.5,-0.25])

plt.plot(punto1[0], punto1[1], 'o', markerfacecolor='None', markeredgecolor='k')
plt.text(punto1[0], punto1[1], '$p^1$', verticalalignment='top', horizontalalignment='left', fontsize=14)
plt.plot(punto2[0], punto2[1], 'o', markerfacecolor='None', markeredgecolor='k')
plt.text(punto2[0], punto2[1], '$p^2$', verticalalignment='top', horizontalalignment='left', fontsize=14)
plt.plot(punto3[0], punto3[1], 'o', markerfacecolor='r', markeredgecolor='k')
plt.text(punto3[0], punto3[1], '$p^3$', verticalalignment='top', horizontalalignment='left', fontsize=14)
plt.plot(punto4[0], punto4[1], 'o', markerfacecolor='w', markeredgecolor='k')
plt.text(punto4[0], punto4[1], '$p^4$', verticalalignment='top', horizontalalignment='left', fontsize=14)
plt.axis('equal')
plt.axis('off');
#plt.savefig('ejm_cjto.pdf')
```

$p^2$ | si | no | si | no | si
$p^3$ | si | no | no | si | si
$p^4$ | no | no | si | no | si

Para finalizar esta breve introducción a conjuntos daremos una última definición.

>**Definición**
>
> Diremos que $S\subset \mathbb{R}^n$ es **acotado** si existe $r>0$ tal que $S\subset B_r(0)$.
>
> Diremos que $S\subset \mathbb{R}^n$ es un **compacto** si es cerrado y acotado.




En muchos casos el uso de sucesiones será de gran utilidad para el análisis a desarrollar, en este caso una **sucesión** de puntos en $\mathbb{R}^n$ será denotada por $\{x^k\}_{k\in\mathbb{N}}$ o simplemente por  $\{x^k\}$ cuando no existan ambigüedades. Una **subsucesión** es la restricción de la sucesión a un conjunto infinito de subíndices $\mathcal{K} = \{ k_1 < k_2 < \ldots < k_m < \ldots\} \subset \mathbb{N}$ y suele denotarse con $\{x^k\}_{k\in \mathcal{K}}$ o $\{x^{k_j}\}$.

>**Definición**
>
> Diremos que una sucesión $\{x^k\}$ converge a $\bar{x}$ si para cada $\varepsilon>0$ existe $k_0 \in \mathbb{N}$ tal que
$$
    x^k \in B_\varepsilon(\bar{x}) \textrm{ para todo } k\geq k_0.
$$
En este caso escribiremos $\displaystyle \lim_{k \in \mathbb{N}} x^k = \bar{x}$, $\displaystyle \lim_{k \to \infty} x^k = \bar{x}$ o simplemente $x^k \to \bar{x}$.



De acuerdo a esta definición, tenemos que la sucesión de vectores $\{x^k\}$ converge a $\bar{x}$ si y sólo si la sucesión de escalares $\{\|x^k - \bar{x}\|\}$ converge a cero. Claramente, cualquier subsucesión de una sucesión convergente también es convergente y converge al mismo límite.


>**Teorema**
>
> Una sucesión $\{x^k\}\subset\mathbb{R}^n$ converge a $\bar{x}$ si y sólo si $\{x^k_i\}\subset \mathbb{R}$ converge a $\bar{x}_i$ para $i=1, \ldots, n$.

**Demostración:**
Para cada coordenada $i$ se tiene que
$$
|x^k_i - \bar{x}_i| \leq \sqrt{|x^k_1 - \bar{x}_1|^2 + \ldots + |x^k_n - \bar{x}_n|^2}
    = \|x^k - \bar{x}\|,
$$
por lo tanto si $x^k \to \bar{x}$ entonces $x^k_i \to \bar{x}_i$ para $i=1, \ldots, n$.

Recíprocamente, si $x^k_i \to \bar{x}_i$ para $i=1, \ldots, n$, entoces dado $\varepsilon>0$ para cada $i$ existe $k_i\in \mathbb{N}$ tal que $|x^k_i - \bar{x}_i| < \varepsilon/\sqrt{n}$ si $k\geq  k_i$. Definiendo $k_0=\max\{k_1, \ldots, k_n\}$ tenemos que para todo $k\geq k_0$
$$
\|x^k - \bar{x}\| = \sqrt{|x^k_1 - \bar{x}_1|^2 + \ldots + |x^k_n - \bar{x}_n|^2}
< \sqrt{n \left(\frac{\varepsilon}{\sqrt{n}} \right)^2} = \varepsilon.
$$
Luego $x^k \to \bar{x}$. $\spadesuit$



De esta manera, podemos extender de manera natural las propiedades conocidas de sucesiones en $\mathbb{R}$ para sucesiones en $\mathbb{R}^n$. Entre estas propiedades se encuentra la siguiente.

>**Teorema (Bolzano-Weierstrass)**
>
> Toda sucesión acotada en $\mathbb{R}^n$ tiene una subsucesión convergente.

**Demostración:**
Suponga que $\{x^k\} \subset \mathbb{R}^n$ es una sucesión acotada. Mirando la primera coordenada se obtiene que $\{x^k_1\} \subset \mathbb{R}$ es una sucesión acotada, y por el Teorema de Bolzano-Weierstrass en $\mathbb{R}$ existe una subsucesión convergente. O sea, existe un subconjunto infinito $\mathcal{K}_1 \subset \mathbb{N}$ y $\bar{x}_1$ tal que $\displaystyle \lim_{k\in \mathcal{K}_1} x^k_1 = \bar{x}_1$.

Ahora, $\{x^k\}_{k\in \mathcal{K}_1} \subset \mathbb{R}^n$ es una sucesión acotada. Como $\{x^k_2\}_{k\in \mathcal{K}_1} \subset \mathbb{R}$ es una sucesión acotada, entonces existe un subconjunto infinito $\mathcal{K}_2 \subset \mathcal{K}_1$ y $\bar{x}_2$ tal que $\displaystyle \lim_{k\in \mathcal{K}_2} x^k_2 = \bar{x}_2$.

Siguiendo de esta manera hasta el índice $n$ obtendremos $\mathcal{K}_n \subset \ldots \subset \mathcal{K}_1 \subset \mathbb{N}$ y $\bar{x}_1, \ldots, \bar{x}_n$ tales que $\displaystyle \lim_{k\in \mathcal{K}_n} x^k_i = \bar{x}_i$ para $i=1,\ldots,n$. Por lo tanto $\displaystyle \lim_{k\in \mathcal{K}_n} x^k = \bar{x}$. $\spadesuit$



Las sucesiones nos serán de utilidad para entender el concepto de clausura desde otro punto de vista.

>**Proposición**
>
> Sea $S\subset \mathbb{R}^n$, entonces $x\in\operatorname{cl}(S)$ si y sólo si existe una sucesión $\{x^k\} \subset S$ tal que $x^k \to x$.


**Demostración:**
($\Rightarrow$) Si $x\in\operatorname{cl}(S)$ es un punto aislado, entonces $x\in S$ y la sucesión constante $\{x^k\}$ tal que $x^k = x$ para todo $k$ cumple que $\{x^k\} \subset S$ y $x^k \to x$.

Sino, $x$ es un punto de acumulación y por lo tanto $\bigl( B_r(x)\setminus\{x\} \bigr) \cap S \neq \emptyset$ para todo $r>0$. Luego, para cada $k\in\mathbb{N}$ definimos $x^k$ tal que $x^k \in \bigl( B_{1/k}(x)\setminus\{x\} \bigr) \cap S$. Obteniendo que $\{x^k\} \subset S$ y $x^k \to x$.


($\Leftarrow$) Si $\{x^k\} \subset S$ y $x^k \to x$, entonces para todo $\varepsilon>0$ existe $k_0$ tal que $x^k \in B_\varepsilon(x) \cap S $ para $k\geq k_0$. Por lo tanto, para todo $\varepsilon>0$ tenemos que $B_\varepsilon(x) \cap S \neq \emptyset$. Concluyendo que para todo $\varepsilon>0$ o bien $B_\varepsilon(x) \cap S = \{x\}$ o $\bigl( B_\varepsilon(x)\setminus\{x\} \bigr) \cap S \neq \emptyset$. O sea, $x\in\operatorname{cl}(S)$. $\spadesuit$



Este resultado nos permite afirmar que un conjunto $C$ es cerrado si y sólo si para toda sucesión $\{x^k\} \subset C$ tal que $x^k \to x$ se cumple $x\in C$.
De esta manera, $U\subset \mathbb{R}^n$ es abierto si ninguna sucesión en $\mathbb{R}^n \setminus U$ converge a puntos de $U$.



Para los conjuntos compactos obtenemos lo siguiente.

>**Proposición**
>
> Sea $K\subset \mathbb{R}^n$, entonces $K$ es compacto si y sólo si toda sucesión $\{x^k\} \subset K$ tiene una subsucesión que converge a un punto de $K$.


**Demostración:**
($\Rightarrow$) Si $K$ es compacto y $\{x^k\} \subset K$, entonces la sucesión es acotada ya que $K$ es acotado. Luego, por el Teorema de Bolzano-Weierstrass existe una subsucesión $\{x^{k_j}\}$ y $\bar{x}$ tal que $x^{k_j} \to \bar{x}$. Como $K$ es cerrado, entonces $\bar{x}\in K$.

($\Leftarrow$) Veamos que $K$ es cerrado y acotado. Para ver que $K$ es cerrado tomemos $\{x^k\} \subset K$ tal que $x^k \to x$, por hipótesis debe existir una subsucesión $\{x^{k_j}\}$ que converge a un punto de $K$ y como toda subsucesión de $\{x^k\}$ converge a $x$ se obtiene que $x\in K$. Concluyendo que $K$ es cerrado.

Para ver que $K$ es acotado, supongamos lo contrario, o sea, para cada $k\in\mathbb{N}$, $K\not\subset B_k(0)$. Esto garantiza la existencia de una sucesión $\{x^k\} \subset K$ tal que $\|x^k\| \geq k$ y claramente esta sucesión no posee ninguna subsucesión convergente. $\spadesuit$



### Funciones continuas

La noción de límite de una función involucra el cumplimiento conjunto de un límite en el dominio y otro en la imagen de la función. Por lo tanto serán de mucha utilidad los conceptos vistos anteriormente.

>**Definición**
>
> Sean $f:\mathbb{R}^n \to \mathbb{R}^m$, $\bar{y}\in\mathbb{R}^m$ y $\bar{x}$ un punto de acumulación de $\operatorname{dom}(f)$. Diremos que $\bar{y}$ es el límite de $f$ en $\bar{x}$, denotado $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y}$, si para todo $\varepsilon>0$ existe $\delta>0$ tal que si $x\in\operatorname{dom}(f)$ y $0<\|x-\bar{x}\|<\delta$ entonces $\| f(x) - \bar{y} \|<\varepsilon$.


En término de los conjuntos anteriormente estudiados, podemos afirmar que $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y}$ si y sólo si
$$
    \forall\, \varepsilon>0, \ \exists\, \delta>0 \textrm{ tal que }
    x \in \bigl( B_\delta(\bar{x})\setminus \{\bar{x}\} \bigr) \cap \operatorname{dom}(f)
    \quad \Rightarrow \quad f(x)\in B_\varepsilon(\bar{y}).
$$
Muchas veces, para referirnos a este límite diremos que $f(x)$ tiende a $\bar{y}$ cuando $x$ tiende a $\bar{x}$, o en símbolos $f(x) \to \bar{y}$ cuando $x \to \bar{x}$.



La necesidad de considerar elementos en un entorno de $\bar{x}$ que no contenga a $\bar{x}$, proviene de evitar aquellos casos donde $f$ está definida en $\bar{x}$ pero el límite de $f$ cuando $x$ tiende a $\bar{x}$ es distinto de $f(\bar{x})$.

#### Ejemplo

Considere la función $f:\mathbb{R} \to \mathbb{R}$ tal que
$$
    f(x) = \left\{\begin{array}{cc}
         0 & \textrm{ si } x\neq 0, \\
         1 & \textrm{ si } x=0.
    \end{array} \right.
$$

Como $\operatorname{dom}(f) = \mathbb{R}$, si $\bar{x}=0$ entonces
$$
\bigl( B_\delta(\bar{x})\setminus \{\bar{x}\} \bigr) \cap \operatorname{dom}(f)
= (-\delta, 0) \cup (0,\delta),
$$
y claramente $f(x)=0$ para $x$ en este conjunto. Por lo tanto, $\lim_{x \to 0} f(x) = 0$.

Si en la definición de límite consideramos los $x$ en el conjunto
$$
 B_\delta(\bar{x}) \cap \operatorname{dom}(f)
= (-\delta,\delta),
$$
tendríamos que $f(x) \in \{0,1\}$ y por lo tanto el límite no existiría.
