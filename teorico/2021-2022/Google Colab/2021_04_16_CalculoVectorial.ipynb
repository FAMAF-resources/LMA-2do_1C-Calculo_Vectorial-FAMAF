# Funciones de varias variables

## Límites y continuidad

### Funciones continuas

Recuerde que $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y}$ si y sólo si
$$
    \forall\, \varepsilon>0, \ \exists\, \delta>0 \textrm{ tal que }
    x \in \bigl( B_\delta(\bar{x})\setminus \{\bar{x}\} \bigr) \cap \operatorname{dom}(f)
    \quad \Rightarrow \quad f(x)\in B_\varepsilon(\bar{y}).
$$


En la práctica, trabajar con esta definición de límite puede ser muy complejo, por ello veremos un equivalente mediante sucesiones.

>**Teorema**
>
> Dada $f:\mathbb{R}^n \to \mathbb{R}^m$, tenemos que $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y}$ si y sólo si $\displaystyle \lim_{k \to \infty} f(x^k) = \bar{y}$ para cualquier $\{x^k\}\subset\operatorname{dom}(f)\setminus \{\bar{x}\}$ tal que $\displaystyle \lim_{k \to \infty} x^k = \bar{x}$.

**Demostración:**
($\Rightarrow$) Sea $\{x^k\}\subset\operatorname{dom}(f)\setminus \{\bar{x}\}$ una sucesión arbitraria tal que $x^k \to \bar{x}$. Tome $\varepsilon>0$ y veamos que existe $k_0$ tal que $f(x^k) \in B_\varepsilon(\bar{y})$ para todo $k\geq k_0$. Como $\lim_{x \to \bar{x}} f(x) = \bar{y}$, por definición, $\varepsilon$ tiene asociado un $\delta$. Usando que $x^k \to \bar{x}$ tenemos que existe $k_0$ tal que $x^k \in B_\delta(\bar{x})$ para todo $k\geq k_0$ y sumado a que $\{x^k\}\subset\operatorname{dom}(f)\setminus \{\bar{x}\}$, obtenemos
$$
    x^k \in \bigl( B_\delta(\bar{x})\setminus \{\bar{x}\} \bigr) \cap \operatorname{dom}(f),
$$
para todo $k\geq k_0$. Concluyendo que $f(x^k) \in B_\varepsilon(\bar{y})$. para todo $k\geq k_0$.


($\Leftarrow$) Usando el contrarecíproco, si $\bar{y}$ no es el límite de $f$ en $\bar{x}$ entonces existe $\varepsilon$ tal que para todo $\delta$ existe $x$ con
$$
    x \in \bigl( B_\delta(\bar{x})\setminus \{\bar{x}\} \bigr) \cap \operatorname{dom}(f)
    \ \textrm{ y } \ f(x)\notin B_\varepsilon(\bar{y}).
$$
Tomando $\delta=1/k$ para $k\in\mathbb{N}$ se puede construir $\{x^k\}$ tal que
$$
    x^k \in \bigl( B_{1/k}(\bar{x})\setminus \{\bar{x}\} \bigr) \cap \operatorname{dom}(f)
    \ \textrm{ y } \ f(x^k)\notin B_\varepsilon(\bar{y}).
$$
Obteniendo así que existe $\{x^k\}\subset\operatorname{dom}(f)\setminus \{\bar{x}\}$ que converge a $\bar{x}$ y que $\{f(x^k)\}$ no converge a $\bar{y}$. $\spadesuit$



#### Ejemplo

Considere $f:\mathbb{R}^2 \to \mathbb{R}$ tal que
$$
f(x) = \frac{x_1^2 - x_2^2}{x_1^2 + x_2^2}.
$$
Claramente $\operatorname{dom}(f) = \mathbb{R}^2 \setminus \{0\}$. Veamos que esta función no tiene límite en $\bar{x} = 0$. Para ello, fije un escalar $\alpha$ y considere la sucesión $x^k = (1/k, \alpha/k)$. Luego $x^k \to \bar{x}$ y
$$
    \lim_{k \to \infty} f(x^k) = \lim_{k \to \infty} \frac{1-\alpha^2}{1+\alpha^2} = \frac{1-\alpha^2}{1+\alpha^2}.
$$
Note que para cada $\alpha>0$ obtenemos un límite distinto, algo que no puede ocurrir si el límite de $f$ existiera en $\bar{x}$

Combinando los Teoremas anteriores podemos obtener una gran variedad de resultados gracias a lo visto para sucesiones. Listaremos a continuación alguno de ellos.

>**Corolario**
>
> Para $f:\mathbb{R}^n \to \mathbb{R}^m$, $g:\mathbb{R}^n \to \mathbb{R}^m$ y $\alpha \in \mathbb{R}$, las siguientes valen.
1. Si $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y}$ y $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{z}$, entonces $\bar{y} = \bar{z}$.
2. $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y}$ si y sólo si $\displaystyle \lim_{x \to \bar{x}} f_i(x) = \bar{y}_i$ para $i=1, \ldots, m$.
3. Si $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y}$, entonces $\displaystyle \lim_{x \to \bar{x}} \alpha f(x) = \alpha \bar{y}$, donde $(\alpha f)(x) = \alpha f(x)$.
4. Si $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y}$ y $\displaystyle \lim_{x \to \bar{x}} g(x) = \bar{z}$, entonces $\displaystyle \lim_{x \to \bar{x}} (f+g)(x) = \bar{y} + \bar{z}$, donde $(f+g)(x) = f(x) + g(x)$.
5. Si $m=1$, $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y}$ y $\displaystyle \lim_{x \to \bar{x}} g(x) = \bar{z}$, entonces $\displaystyle \lim_{x \to \bar{x}} (fg)(x) = \bar{y}\bar{z}$, donde $(fg)(x)=f(x) g(x)$.
6. Si $m=1$, $\displaystyle \lim_{x \to \bar{x}} f(x) = \bar{y} \neq 0$ y $f(x)\neq 0$ para $x\neq \bar{x}$ en un entorno de $\bar{x}$, entonces $\displaystyle \lim_{x \to \bar{x}} (1/f)(x) = 1/\bar{y}$, donde $(1/f)(x) = 1/f(x)$.





Para poder trabajar con funciones, suele ser importante considerar aquellas cuyo valor no cambia de manera abrupta. O sea, aquellas funciones $f$ tales que si $x$ está cerca de $\bar{x}$ entonces $f(x)$ debe estar cerca de $f(\bar{x})$. Claramente esta propiedad se encuentra fuertemente relacionada con la noción de límite y su definición formal es la siguiente.

>**Definición**
>
> Diremos que $f:\mathbb{R}^n \to \mathbb{R}^m$ es **continua en $\bar{x}$**, si $\bar{x} \in \operatorname{dom}(f)$ y es un punto aislado o
$$
    \lim_{x \to \bar{x}} f(x) = f(\bar{x}).
$$
Diremos simplemente que $f$ es **continua** si es continua en todos los puntos de su dominio.

Note que cualquier función es continua en los puntos aislados de su dominio.


Usando sucesiones en la noción de límite y del hecho que la función es continua en puntos aislados, obtenemos la siguiente caracterización.

>**Teorema**
>
>Una función $f:\mathbb{R}^n \to \mathbb{R}^m$ es continua en $\bar{x}$ si y sólo si $\displaystyle \lim_{k \to \infty} f(x^k) = f(\bar{x})$ para cualquier $\{x^k\}\subset\operatorname{dom}(f)$ tal que $\displaystyle \lim_{k \to \infty} x^k = \bar{x}$.



De las propiedades vistas para límites de funciones obtenemos que:
* Una función es continua en un punto si y sólo si sus funciones coordenadas son continuas en ese punto.
* El producto escalar de una función continua es una función continua.
* La suma y producto de funciones continuas es una función continua.



Otra propiedad importante de las funciones continuas es que al componerlas la continuidad se mantiene.

>**Teorema**
>
> Sean $f:\mathbb{R}^n \to \mathbb{R}^m$ y $g:\mathbb{R}^m \to \mathbb{R}^p$ tal que $\operatorname{img}(f) \subset \operatorname{dom}(g)$. Si $f$ es continua en $\bar{x}$ y $g$ es continua en $f(\bar{x})$, entonces $g \circ f$ es continua en $\bar{x}$.


**Demostración:**
Tome una sucesión arbitraria $\{x^k\}\subset\operatorname{dom}(f)$ tal que $\displaystyle \lim_{k \to \infty} x^k = \bar{x}$. Como $f$ es continua en $\bar{x}$, entonces $\displaystyle \lim_{k \to \infty} f(x^k) = f(\bar{x})$. Además $\{f(x^k)\}\subset \operatorname{dom}(g)$. Como $g$ es continua en $f(\bar{x})$, entonces $\displaystyle \lim_{k \to \infty} g(f(x^k)) = g(f(\bar{x}))$. $\spadesuit$




#### Ejemplo

La función $f:\mathbb{R}^2 \to \mathbb{R}^2$ tal que
$$
f(x) = \left( x_1^2 x_2 , \frac{x_2 + x_1^3}{1+x_1^2}\right),
$$
es continua, pues sus funciones coordenadas son continuas por ser cada una de ellas composición de funciones continuas.

---

Un ejemplo muy simple de función continua es el siguiente.

>**Teorema**
>
> Si $L:\mathbb{R}^n \to \mathbb{R}^m$ es lineal, entonces es continua. Más aún, existe $\kappa>0$ tal que
$$
    \| L(x) \| \leq \kappa \|x\|, \quad \textrm{ para todo } x\in\mathbb{R}^n.
$$

**Demostración:**
Veamos primero la desigualdad. Como $L$ es lineal, usando la desigualdad triangular tenemos que
\begin{align*}
    \| L(x) \|  &= \| x_1 L(e^1) + \ldots + x_n L(e^n)\| \\
                &\leq |x_1| \|L(e^1)\| + \ldots + |x_n| \|L(e^n)\|,
\end{align*}
donde $e^j\in \mathbb{R}^n$ son los vectores canónicos. Como $|x_j| \leq \|x\|$, definiendo $\kappa = \|L(e^1)\| + \ldots + \|L(e^n)\|$, obtenemos que $\| L(x) \| \leq \kappa \|x\|$.

Ahora, dados $x$ y $\bar{x}$ en $\mathbb{R}^n$ concluímos que
\begin{equation*}
    \|L(x) - L(\bar{x})\| = \| L (x-\bar{x})\| \leq \kappa \| x-\bar{x}\|.    
\end{equation*}
Por lo tanto $L(x)$ tiende a $L(\bar{x})$ cuando $x$ tiende a $\bar{x}$. $\spadesuit$




Algo importante a resaltar, es que la última desigualdad nos dice que la distancia entre los valores de la función es a lo sumo del mismo orden que la distancia de los puntos del dominio.
Además de las funciones lineales, existen otras funciones que cumplen con esto, o sea, para $f:\mathbb{R}^n \to \mathbb{R}^m$ existe $\kappa>0$ tal que para todo $x,\bar{x}\in \mathbb{R}^n$
$$
    \|f(x) - f(\bar{x}) \| \leq \kappa \| x - \bar{x}\|.
$$
Este subconjunto de funciones continuas se conocen como funciones **Lipschitz continuas**.



## Funciones diferenciables

Cuando se trabaja con funciones no lineales, es de mucha utilidad contar con un modelo lineal que aproxime bien a la función cerca del punto en estudio. En otras palabras, para $f:\mathbb{R}^n \to \mathbb{R}^m$ y $\bar{x} \in \mathbb{R}^n$ se desea tener una "muy buena" aproximación de la función $v \mapsto f(\bar{x} + v) - f(\bar{x})$ por medio de una función lineal $L:\mathbb{R}^n \to \mathbb{R}^m$ para $v$ cerca de cero. O sea,
$$
    f(\bar{x} + v) - f(\bar{x}) \approx L(v),
$$
para $v$ cerca de $0$. Otra manera de ver ésto es que para $v$ próximo a cero queremos aproximar la función no lineal $v \mapsto f(\bar{x} + v)$ por la función afín $v \mapsto f(\bar{x}) + L(v)$. Donde la condición para ser una "muy buena" aproximación es que la distancia de $f(\bar{x} + v)$ a $f(\bar{x}) + L(v)$ tienda a cero más rápido de lo que $v$ tiende a cero.



```
#@title Diferenciabilidad { display-mode: "form" }
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1,3,200)
a = 0.1
f = -a*x**2
x0 = x[30]
af = -a*x0**2 -2*a*x0*(x-x0)
plt.plot(x,f,x,af)
plt.plot(x[30:200:50], f[30:200:50], 'r.')
plt.plot(x[30:200:50], af[30:200:50], 'r.')
for i in range(30,200,50):
  plt.plot([x[30], x[i], x[i]], [f[i], f[i], af[i]], 'r:')

plt.text(x[190], f[190], r'$f(\bar{x} + v)$', verticalalignment='bottom', horizontalalignment='left', fontsize=14)
plt.text(x[190], af[190], r'$f(\bar{x}) + L(v)$', verticalalignment='top', horizontalalignment='left', fontsize=14)
plt.text(x[30], f[190], r'$\bar{x}$', verticalalignment='top', horizontalalignment='center', fontsize=14)
plt.text(x[180], f[190], r'$\bar{x} + v$', verticalalignment='top', horizontalalignment='center', fontsize=14)
plt.axis('equal')
plt.axis('off');
#plt.savefig('diferenciable.pdf')
```

Cuando se tienen funciones de una variable, tal aproximación afín viene dada en términos de la derivada de la función. Este tipo de aproximación también es de utilidad en funciones de varias variables.

>**Definición**
>
> Para $f:\mathbb{R}^n \to \mathbb{R}^m$, la **derivada parcial de $f$ en $\bar{x}$ con respecto a la $j$-ésima variable** es
\begin{align*}
\frac{\partial f}{\partial x_j}(\bar{x}) &= \lim_{t \to 0} \frac{f(\bar{x} + t e^j) - f(\bar{x})}{t} \\
                &= \lim_{t \to 0} \frac{f(\bar{x}_1, \ldots, \bar{x}_j + t, \ldots, \bar{x}_n) - f(\bar{x}_1, \ldots, \bar{x}_j, \ldots, \bar{x}_n)}{t}.
\end{align*}
Donde la función $\frac{\partial f}{\partial x_j}: \mathbb{R}^n \to \mathbb{R}^m$ tiene como dominio aquellos puntos de $\operatorname{dom}(f)$ donde el límite existe.


Ya que el límite anterior existe si y sólo si existe en cada coordenada, podemos ver que
$$
\frac{\partial f}{\partial x_j}(x) = \left( \frac{\partial f_1}{\partial x_j}(x), \ldots, \frac{\partial f_m}{\partial x_j}(x) \right).
$$




#### Ejemplo

Sea $f:\mathbb{R}^3 \to \mathbb{R}$ tal que
$$
f(x) = x_1^2 x_2 + x_2^2 x_3 + x_3^2 x_1.
$$
Entonces
\begin{equation*}
    \frac{\partial f}{\partial x_1}(x) = 2 x_1 x_2 + x_3^2, \quad
    \frac{\partial f}{\partial x_2}(x) = x_1^2 + 2 x_2 x_3, \quad
    \frac{\partial f}{\partial x_3}(x) = x_2^2 + 2 x_3 x_1.
\end{equation*}
Si por ejemplo $\bar{x} = (1,2,3)$, tenemos que
\begin{equation*}
    \frac{\partial f}{\partial x_1}(\bar{x}) = 4 + 9 = 13, \quad
    \frac{\partial f}{\partial x_2}(\bar{x}) = 1 + 12 = 13, \quad
    \frac{\partial f}{\partial x_3}(\bar{x}) = 4 + 6 = 10.
\end{equation*}


---


Con esto, si considero una función $L$ lineal en cada eje Cartesiano tal que $L(\alpha e^j) = \alpha \frac{\partial f}{\partial x_j}(\bar{x})$ para $j=1, \ldots, n$, entonces la distancia de $f(\bar{x} + t e^j)$ a $f(\bar{x}) + L(t e^j)$ tienda a cero más rápido de lo que $t e^j$ tiende a cero, pues
\begin{align*}
\lim_{t \to 0} \frac{\|f(\bar{x} + t e^j) - ( f(\bar{x}) + L(t e^j) )\|}{\|t e^j\|}
    &= \lim_{t \to 0} \left\|\frac{f(\bar{x} + t e^j) - f(\bar{x})}{t} -  \frac{\partial f}{\partial x_j}(\bar{x}) \right\| \\
    &= 0.
\end{align*}



Ya que en cada coordenada se cumple con la condición deseada, se necesitaría que $L$ sea lineal en todo el espacio y no solamente en los ejes Cartesianos. Como todo $v\in \mathbb{R}^n$ puede escribirse como combinación lineal de la base canónica, un candidato sería calcular el límite de la definición anterior cambiando el eje por la recta definida por $v$.

>**Definición**
>
> Para $f:\mathbb{R}^n \to \mathbb{R}^m$, la **derivada direccional de $f$ en $\bar{x}$ en la dirección $v$** es
\begin{align*}
    f'(\bar{x}; v) &= \lim_{t \to 0} \frac{f(\bar{x} + t v) - f(\bar{x})}{t} \\
                &= \lim_{t \to 0} \frac{f(\bar{x}_1 + t v_1, \ldots, \bar{x}_n + t v_n) - f(\bar{x}_1, \ldots, \bar{x}_n)}{t},
\end{align*}
si tal límite existe.




Algunos autores piden en la definición anterior que $v$ sea un vector unitario, o sea $\|v\|=1$, para enfatizar que lo importante es la dirección y no la magnitud del vector $v$.
Es fácil ver que la definición dada también cumple con esa característica, ya que si $\alpha \neq 0$,
\begin{align*}
    f'(\bar{x}; \alpha v) &= \lim_{t \to 0} \frac{f(\bar{x} + t \alpha v) - f(\bar{x})}{t} \\
                        &= \lim_{\alpha t \to 0} \alpha \frac{f(\bar{x} + \alpha t v) - f(\bar{x})}{\alpha t} \\
                        &= \alpha f'(\bar{x};  v).
\end{align*}




Claramente,
$$
    f'(\bar{x}; e^j) = \frac{\partial f}{\partial x_j}(\bar{x}).
$$
Por lo tanto el candidato natural para la función lineal $L$ sería $L(v) = f'(\bar{x}; v)$. Pero lamentablemente esta función no siempre es lineal.

#### Ejemplo

Considere la función $f:\mathbb{R}^2 \to \mathbb{R}$ tal que
$$
    f(x) = \left\{ \begin{array}{cc}
         \frac{x_1 x_2}{x_1^2 + x_2^2}& \textrm{ si } x\neq 0,  \\
         0 & \textrm{ si } x=0.
    \end{array}\right.
$$



```
#@title No diferenciable { run: "auto", form-width: "50%" }
elevacion = 46 #@param {type:"slider", min:0, max:90, step:1}
acimut = 33 #@param {type:"slider", min:0, max:180, step:1}

import numpy as np
import matplotlib.pyplot as plt

x1, x2 = np.meshgrid(np.linspace(-1,1,100), np.linspace(-1,1,100))
f = x1*x2/(x1**2 + x2**2)

ax = plt.axes(projection='3d')
ax.plot_surface(x1, x2, f, rstride=1, cstride=1, cmap='viridis')
ax.view_init(elevacion, acimut);
#ax.axis('off');
#plt.savefig('ejm_nodif1.pdf')
```

Si $\bar{x}=0$, para cualquier $v\in\mathbb{R}^2$ con $v\neq 0$ se tiene que
$$
f'(\bar{x};v) = \lim_{t \to 0} \frac{f(\bar{x} + tv) - f(\bar{x})}{t}
            = \lim_{t \to 0}  \frac{t^2 v_1 v_2}{t^2 v_1^2 + t^2 v_2^2}
            = \frac{v_1 v_2}{v_1^2 + v_2^2}.
$$
Puede verse fácilmente que $v \mapsto f'(\bar{x};v)$ no es lineal, pues para $v = (1,\alpha) = e^1 + \alpha e^2$, $\alpha\neq 0$,
$$
 f'(\bar{x};v) = \frac{\alpha}{1 + \alpha^2} \neq 0 = f'(\bar{x};e^1) + \alpha f'(\bar{x};e^2).
$$
