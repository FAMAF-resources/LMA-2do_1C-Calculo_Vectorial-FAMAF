# Integración múltiple

## Integrales iteradas






Veamos que el orden de integración no afecta el resultado.

> **Teorema (Fubini)**
>
> Sea $f:\mathbb{R}^n \to \mathbb{R}$ continua en $D$ excepto quizás en una cantidad finita de conjuntos suaves y
 $$
 D= \{x\in \mathbb{R}^n \mid a_1\leq x_1 \leq b_1, \, \eta_2(x_1) \leq x_2 \leq \xi_2(x_1), \,
 \ldots,\, \eta_n(x_{n-1}) \leq x_n \leq \xi_n(x_{n-1}) \}.
 $$
 Si existe la integral iterada
 $$
    \int_{a_1}^{b_1} \int_{\eta_2(x_1)}^{\xi_2(x_1)} \ldots \int_{\eta_n(x_{n-1})}^{\xi_n(x_{n-1})} f(x_1, x_2, \ldots, x_n) dx_n \ldots dx_2 dx_1,
 $$
 y además existe $\int_D f(x) dx$, entonces las integrales son iguales.


**Demostración:**

Veamos la prueba para $\mathbb{R}^2$, ya que el caso general sale de manera análoga. Supongamos primero que $D$ es un rectángulo, o sea, $D=[a_1,b_1] \times [a_2, b_2]$ y veamos que
$$
    \int_{a_1}^{b_1} \int_{a_2}^{b_2} f(x_1,x_2) dx_2 dx_1 = \int_D f(x) dx.
$$
Sea
$$
    F(x_1) = \int_{a_2}^{b_2} f(x_1, x_2) dx_2.
$$
Considere la partición $a_2 = s_0 < s_1 < \ldots < s_l = b_2$, entonces
$$
    F(x_1) = \sum_{j=0}^{l-1} \int_{s_j}^{s_{j+1}} f(x_1, x_2) dx_2.
$$
Por el Teorema del valor medio para integrales, existen valores $\zeta_j(x_1)$ tal que
$$
    \int_{s_j}^{s_{j+1}} f(x_1, x_2) dx_2 = f(x_1, \zeta_j(x_1)) (s_{j+1} - s_j),
$$
obteniendo así que
$$
    F(x_1) = \sum_{j=0}^{l-1} f(x_1, \zeta_j(x_1)) (s_{j+1} - s_j)
$$
Por otro lado, por lo visto para integración en una variable, tenemos que
$$
    \int_{a_1}^{b_1} F(x_1) dx_1 = \lim_{l\to \infty} \sum_{i=0}^{l-1} F(\tau_i)(t_{i+1}-t_i),
$$
para cualquier partición $a_1 =t_0 < t_2 < \ldots < t_l = b_1$ y $\tau_i\in [t_i, t_{i+1}]$. Tomando $R_{ij}=[t_i, t_{i+1}] \times [s_j, s_{j+1}]$ y $x^{ij}=(\tau_i, \zeta_j(\tau_i))$, tenemos que
\begin{align*}
    \int_{a_1}^{b_1} \int_{a_2}^{b_2} f(x_1,x_2) dx_2 dx_1
    &= \int_{a_1}^{b_1} F(x_1) dx_1
    = \lim_{l\to \infty} \sum_{i=0}^{l-1} F(\tau_i)(t_{i+1}-t_i) \\
    &= \lim_{l\to \infty} \sum_{i=0}^{l-1} \sum_{j=0}^{l-1} f(x^{ij}) |R_{ij}| \\
    &= \int_D f(x) dx.
\end{align*}

Si $D\subset \mathbb{R}^2$ no es un rectángulo, tomemos un rectángulo $R=[a_1,b_1] \times [a_2, b_2]$ tal que $D \subset R$ y definamos $\bar{f}:R \to \mathbb{R}$ tal que
$$
    \bar{f}(x)=\left\{ \begin{array}{cc} f(x) & \textrm{ si } x\in D,\\ 0& \textrm{ si } x\notin D.
    \end{array} \right.
$$
De esta manera
\begin{align*}
    \int_D f(x) dx &= \int_R \bar{f}(x) dx \\
    &= \int_{a_1}^{b_1} \int_{a_2}^{b_2} \bar{f}(x_1,x_2) dx_2 dx_1 \\
    &= \int_{a_1}^{b_1} \int_{\eta_2(x_1)}^{\xi_2(x_1)} f(x_1,x_2) dx_2 dx_1. \qquad \spadesuit
\end{align*}




Con esto podemos concluir lo siguiente.

>**Corolario**
>
> Si $\int_D f(x) dx$ existe y las integrales interadas existen para algún orden de integración, entonces todas estas integrales son iguales.




Por ejemplo para $g: \mathbb{R}^2 \to \mathbb{R}$, si la función $\frac{\partial g}{\partial x}$ es continua en $ [c,d] \times [a,b]$ tenemos que
\begin{align*}
    \int_{c}^{x} G(\tilde{x}) d\tilde{x}
    &=\int_{c}^{x} \int_a^b \frac{\partial g}{\partial x}(\tilde{x}, t) dt d\tilde{x} \\
    &= \int_a^b \int_{c}^{x} \frac{\partial g}{\partial x}(\tilde{x}, t) d\tilde{x} dt \\
    &= \int_a^b \bigl( g(x,t) - g(c,t) \bigr)  dt = F(x) - F(c).
\end{align*}
Por lo tanto, $F'(x) = G(x)$, obteniendo la conocida **regla de Leibnitz**
$$
    \frac{d}{dx} \int_a^b  g(x,t) dt = \int_a^b \frac{\partial g}{\partial x}(x, t) dt.
$$



Se puede ver que valen las siguientes propiedades para las integrales.

>**Proposición**
>
> 1. Si $f$ y $g$ son integrables sobre $D$ y $\alpha,\beta \in \mathbb{R}$, entonces $\alpha f + \beta g$ es integrable y
$$
    \int_D \bigl(\alpha f(x) + \beta g(x) \bigr)dx = \alpha \int_D f(x) dx + \beta \int_D g(x) dx.
$$
2. Si $f$ y $g$ son integrables sobre $D$ y $f(x) \leq g(x)$ para todo $x\in D$, entonces
$$
    \int_D f(x) dx \leq \int_D g(x) dx.
$$
3. Si $f$ es integrable sobre $D_1$ y $D_2$ con $D_1 \cap D_2 = \emptyset$, entonces
$$
 \int_{D_1 \cup D_2} f(x) dx = \int_{D_1} f(x) dx + \int_{D_2} f(x) dx.
$$



Claramente, de la propiedad 2 obtenemos que
$$
\int_D f(x) dx \geq 0,
$$
si $f(x) \geq 0$ para todo $x\in D$. Además, como $-|f(x)| \leq f(x) \leq |f(x)|$ se obtiene que $-\int_D |f(x)|dx \leq \int_D f(x) dx \leq \int_D |f(x)| dx$, o sea
$$
  \left| \int_D f(x) dx \right| \leq \int_D |f(x)| dx.
$$

Por otro lado, en la propiedad 3 podemos tener $D_1 \cap D_2 \neq \emptyset$ siempre y cuando $|D_1 \cap D_2 | = 0$. Esto es consecuencia de que la integral no cambia cuando se modifica en conjuntos de medida nula. Trivialmente podemos extender esta propiedad a una cantidad finita de conjuntos,
$$
    \int_{D_1 \cup D_2 \cup \ldots \cup D_p} f(x) dx
    = \int_{D_1} f(x) dx + \int_{D_2} f(x) dx + \ldots + \int_{D_p} f(x) dx.
$$




## Cambio de variables

Recordemos que para integración en una variable sabemos que si $\tau$ es continuamente diferenciable en $[a,b]$ entonces
$$
\int_{\tau(a)}^{\tau(b)} f(y) dy = \int_a^b f(\tau(x)) \tau'(x) dx.
$$
Note que si $\tau$ es una función no decreciente y $f = 1$, entonces obtenemos
$$
 \tau(b) - \tau(a) = \int_{\tau(a)}^{\tau(b)} 1 dy = \int_a^b \tau'(x) dx,
$$
o sea, la medida (longitud) del conjunto $\tau([a,b])$.

En varias variables sucede algo similar, y para entenderlo veamos primero cómo medir la imagen de transformaciones lineales.




Supongamos que $T:\mathbb{R}^2 \to \mathbb{R}^2$ es lineal y sea $R=[0,1] \times [0,1]$ el rectángulo unitario. Claramente, el rectángulo unitario es el paralelogramo formado por los vectores canónicos $e^1$ y $e^2$ en $\mathbb{R}^2$, pues $R=\{\alpha_1 e^1 + \alpha_2 e^2 \mid 0\leq \alpha_1 \leq 1,\, 0\leq \alpha_2 \leq 1\}$. Ahora, como $T$ es lineal, tenemos que $T(R)=\{\alpha_1 v^1 + \alpha_2 v^2 \mid 0\leq \alpha_1 \leq 1,\, 0\leq \alpha_2 \leq 1\}$ donde $v^1=T(e^1)$ y $v^2=T(e^2)$, luego $T(R)$ es el paralelogramo formado por los vectores $v^1$ y $v^2$. Más aún, la representación matricial de $T$ viene dada por la matriz
$$
    A = \left[ \begin{array}{cc}
        v^1_1 & v^2_1 \\
         v^1_2& v^2_2
    \end{array}\right].
$$
Por otro lado, la superficie del paralelogramo formado por los vectores $v^1$ y $v^2$ es igual a la superficie del paralogramo formado por los vectores $\bar{v}^1=(v^1_1, v^1_2, 0)$ y $\bar{v}^2=(v^2_1, v^2_2, 0)$. Como vimos en la primera unidad del curso, la superficie de este último es igual a $\| \bar{v}^1 \times \bar{v}^2\|$. Además, sabemos que $\bar{v}^1 \times \bar{v}^2$ es un múltiplo de $\bar{e}^3 \in \mathbb{R}^3$. Usando que Cauchy-Schwarz vale con igualdad, obtenemos que
$$
    |T(R)| = \| \bar{v}^1 \times \bar{v}^2\| = | \langle \bar{v}^1 \times \bar{v}^2, \bar{e}^3 \rangle |
    = \left|\det\left(\left[ \begin{array}{ccc}
             0 & 0 & 1  \\
             v^1_1 & v^1_2 & 0 \\
             v^2_1 & v^2_2 & 0
        \end{array}\right] \right) \right| = \left| \det(A^T)\right| = |\det(A)|.
$$



Supongamos ahora $T:\mathbb{R}^3 \to \mathbb{R}^3$ lineal, cuya representación matricial tiene como columna los vectores $v^1, v^2, v^3 \in \mathbb{R}^3$, y $R\subset \mathbb{R}^3$ es el rectángulo unitario. Claramente, $T(R)$ es el paralelepípedo formado por los vectores $v^1$, $v^2$, $v^3$ y, por lo visto en la primera unidad del curso, su volumen es $|T(R)| = |\det(A)|$. No obstante, veamos una deducción alternativa que nos permitirá generalizar este concepto. Como $T(R)$ es un paralelepípedo en $\mathbb{R}^3$ formado por los vectores en las columnas de $A$, éste puede rotarse de manera de que dos de esos vectores queden en el plano $x_3=0$. Matricialmente, existe una matriz $Q\in \mathbb{R}^{3 \times 3}$ ortogonal tal que
$$
    QA = Q\left[ \begin{array}{ccc}
             v^1_1 & v^2_1 & v^3_1  \\
             v^1_2 & v^2_1 & v^3_2 \\
             v^1_3 & v^2_3 & v^3_3
        \end{array}\right]
        = \left[ \begin{array}{ccc}
             \tilde{v}^1_1 & \tilde{v}^2_1 & \tilde{v}^3_1  \\
             \tilde{v}^1_2 & \tilde{v}^2_2 & \tilde{v}^3_2 \\
             0 & 0 & \tilde{v}^3_3
        \end{array}\right].
$$
Las primeras dos columnas generan un paralelogramo en el plano $x_3=0$. Como $|\det(Q)| = 1$, y usando lo visto para $\mathbb{R}^2$, obtenemos que
$$
    |\det(A)| = |\det(Q) \det(A)| = |\det(QA)|
    = \left|\det\left(\left[ \begin{array}{cc}
             \tilde{v}^1_1 & \tilde{v}^2_1  \\
             \tilde{v}^1_2 & \tilde{v}^2_2
        \end{array}\right] \right) \right| |\tilde{v}^3_3| = |T(R)|,
$$
donde la última igualdad usamos que el volumen es el producto de la superficie de la base por la altura del paralelepípedo.


La deducción para $\mathbb{R}^4$ es análoga a la anterior, y siguiendo de manera inductiva se consigue demostrar que si $T:\mathbb{R}^n \to \mathbb{R}^n$ es lineal y $R\subset \mathbb{R}^n$ es el hiperrectángulo unitario, entonces
$$
    |T(R)| = |\det(A)|,
$$
donde $A\in\mathbb{R}^{n \times n}$ es la representación matricial de $T$. Además, como $|R|=1$ y $T'(x) = A$, en término de integrales obtenemos que
$$
    \int_{T(R)} 1 dy = |T(R)| = |\det(A)|\, |R| = \int_R 1 |\det(T'(x))| dx.
$$



Ahora, suponga que tenemos un hiperrectángulo general $R=[a_1,b_1] \times \ldots \times [a_n,b_n]$ y la matriz asociada a $T$ tiene columnas $v^1, \ldots, v^n$. Luego, $T(R)=\{\alpha_1 l_1 v^1 + \ldots + \alpha_n l_n v^n \mid 0\leq \alpha_i \leq 1,\, i=1, \ldots, n\}$, donde $l_i = b_i-a_i$. Obteniendo que $T(R)$ es un hiperparalelepípedo formado por los vectores $l_1 v^1, \ldots, l_n v^n$. Si denotamos por $\tilde{A}$ a la matriz cuya $i$-ésima columna es $l_i v^i$, obtenemos que
$$
 |T(R)| = |\det(\tilde{A})| = l_1 \ldots l_n |\det(A)| = |R|\, |\det(A)|,
$$
pues el determinante es lineal en cada columna.



Si suponemos que tenemos un conjunto $D\subset \mathbb{R}^n$ y un cubrimiento $\mathcal{G} = \{R_1, \ldots, R_p\}$ de $D$, entonces
$$
\sum_{i=1}^p |T(R_i)| = \sum_{i=1}^p |\det(A)| |R_i|.
$$
Luego, tomando límites cuando $m(\mathcal{G}) \to 0$ obtenemos
$$
    |T(D)| = |\det(A)|\, |D|.
$$
Como $T$ es lineal tenemos que $T'(x) = A$ para todo $x$, por lo tanto en término de integrales tenemos
$$
    \int_{T(D)} 1 dy = \int_D 1 |\det(T'(x))| dx
$$

Veamos que si $T$ es lineal y su representación matricial es $T(x) = Ax$, entonces $T'(x) = A$.

**Método 1:** Note que
$$
T(y) - T(x) - A(y-x) = Ay - Ax - A(y-x) = 0,
$$
Luego $T(y) = T(x) + A(y-x) = T(x) + A(y-x) + o(\|y-x\|)$ y como la matriz Jacobiana es única, debemos tener $T'(x) = A$.

**Método 2:** Como
$$
T(x) = Ax = \left[ \begin{array}{c}
            a_{11} x_1 + \ldots + a_{1n} x_n \\
            a_{21} x_1 + \ldots + a_{2n} x_n \\
              \vdots \\
              a_{n1} x_1 + \ldots + a_{nn} x_n
\end{array}\right].
$$
Luego
$$
\frac{\partial T_i}{\partial x_j}(x) = a_{ij}.
$$
Por lo tanto $T'(x) = A$.

Cuando $T$ es no lineal o la función a integrar no es necesariamente uno, tenemos el siguiente resultado.


>**Teorema**
>
> Sea $T:\mathbb{R}^n \to \mathbb{R}^n$ continuamente diferenciable y $D\subset\mathbb{R}^n$ con $\operatorname{cl}(D) \subset \operatorname{int}(\operatorname{dom}(T))$ tal que su frontera está contenida en una cantidad finita de conjuntos suaves. Suponga que $T$ es inyectiva en $D$ y $\det(T'(x))\neq 0$ para $x\in D$ excepto quizás en una cantidad finita de conjuntos suaves. Si $f:\mathbb{R}^n \to \mathbb{R}$ es continua (excepto quizás en conjuntos suaves) y acotada sobre $T(D)$, entonces
 $$
    \int_{T(D)} f(y) dy = \int_D f(T(x)) |\det(T'(x))| dx.
 $$

**Demostración:**

Sea $\bar{x}\in \operatorname{int}(D)$ arbitrario y considere la función $F(x) = T'(\bar{x})^{-1} T(x)$. Como $F$ es diferenciable y $F'(x) = T'(\bar{x})^{-1} T'(x)$, tenemos que $F(x) = F(\bar{x}) + (x-\bar{x}) + o(\|x-\bar{x}\|)$. Fijo $0<\varepsilon<\tau < 1$.
Luego, existe $r>0$ tal que $\|o(\|x-\bar{x}\|)\| \leq \varepsilon \|x-\bar{x}\|$ para todo $x\in B_r(\bar{x})$.
Considero un hiperrectángulo $R$ tal que $B_{\tau r}(\bar{x}) \subset R \subset B_r(\bar{x})$, obteniendo que $|R| \leq |B_r(\bar{x})| = \alpha r^n$ y $|R| \geq \alpha (\tau r)^n$. Además, si $x\in R$,
$$
    \|F(x) - F(\bar{x})\| \leq (1+\varepsilon) \|x-\bar{x}\| \leq (1+\varepsilon) r,
$$
concluyendo que $F(R) \subset B_{(1+\varepsilon)r}(F(\bar{x}))$. Por lo tanto
$$
    |F(R)| \leq \alpha (1+\varepsilon)^n r^n
    \leq \left(\frac{1+\varepsilon}{\tau} \right)^n |R|.
$$
Como $v \mapsto T'(\bar{x})^{-1} v$ es lineal y $\det\bigl( T'(\bar{x})^{-1} \bigr) = \det\bigl( T'(\bar{x}) \bigr)^{-1}$, por lo visto para transformaciones lineales,
$$
 |F(R)|  = | T'(\bar{x})^{-1}(T(R))| = |\det\bigl( T'(\bar{x}) \bigr)|^{-1} |T(R)|.
$$
Concluyendo que
$$
    |T(R)| \leq \left(\frac{1+\varepsilon}{\tau} \right)^n |\det\bigl( T'(\bar{x}) \bigr)| \,|R|.
$$

Consideremos ahora un cubrimiento $\mathcal{G} = \{R_1, \ldots, R_p\}$ de $D$, tal que lo anterior valga remplazando $R$ por $R_i$ y $\bar{x}$ por $x^i$. Por lo tanto
$$
    \sum_{i=1}^p |T(R_i)| \leq \left(\frac{1+\varepsilon}{\tau} \right)^n \sum_{i=1}^p |\det\bigl( T'(x^i) \bigr)| \,|R_i|.
$$
Tomando límite para $m(\mathcal{G}) \to 0$, se tiene
$$
\int_{T(D)} 1 dy \leq \left(\frac{1+\varepsilon}{\tau} \right)^n \int_D |\det\bigl( T'(x) \bigr)| dx.
$$
Como esto vale para cualquier $\varepsilon$ y $\tau$, podemos tomar límite para $\varepsilon \to 0$ y $\tau \to 1$, consiguiendo
$$
\int_{T(D)} 1 dy \leq  \int_D |\det\bigl( T'(x) \bigr)| dx.
$$


Toda función $f$ puede escribirse como $f=f^+ - f^-$ con $f^+(y) = \max\{0,f(y)\}$ y $f^-(y) = \max\{0, -f(y)\}$, ambas funciones no negativas. Por la linealidad de la integral, sólo basta probar que el resultado vale para $f$ no negativa. Note que
$$
    \int_{T(D)} f(y) dy = \int_{T(D)} \int_0^{f(y)} 1 dt dy = \int_{\bar{T}(\bar{D})} 1 d\bar{y},
$$
donde $\bar{T}(x,t) = (T(x),t)$ y $\bar{D}=\{(x,t) \mid x\in D, \, 0\leq t \leq f(T(x))\}$ y por lo tanto
$$
\bar{T}(\bar{D})=\{(T(x),t) \mid x\in D,\, 0 \leq t \leq f(T(x)) \}
    = \{ (y,t) \mid y\in T(D), \, 0\leq t \leq f(y)\}.
$$
Además, como $\det(\bar{T}'(x,t)) = \det(T'(x))$,
$$
\int_{\bar{D}} 1 |\det(\bar{T}'(\bar{x}))| d\bar{x}
    = \int_D \int_0^{f(T(x))} 1 |\det(T'(x))| dt dx
    = \int_D f(T(x)) |\det(T'(x))| dx.
$$
Por lo visto anteriormente, tenemos que $\int_{\bar{T}(\bar{D})} 1 d\bar{y} \leq  \int_{\bar{D}} |\det( \bar{T}'(\bar{x}) )| d\bar{x}$, concluyendo que
$$
\int_{T(D)} f(y) dy \leq \int_D f(T(x)) |\det(T'(x))| dx.
$$

Para mostrar que la otra desigualdad vale, note que
\begin{align*}
    \int_D f(T(x)) |\det(T'(x))| dx &= \int_{T^{-1}(T(D))} f(T(x)) |\det(T'(x))| dx \\
    & \leq \int_{T(D)} f(y) |\det(T'(T^{-1}(y)))| |\det((T^{-1})'(y))| dy \\
    & = \int_{T(D)} f(y) dy,
\end{align*}
donde usamos que $(T^{-1})'(y) = T'(T^{-1}(y))^{-1}$, pues $y=T(T^{-1}(y))$ y derivando $I = T'(T^{-1}(y)) (T^{-1})'(y)$. $\spadesuit$
